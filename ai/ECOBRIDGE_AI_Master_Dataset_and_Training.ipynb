```python
import os
import zipfile
from pathlib import Path

BASE_DIR = Path("/content/ECOBRIDGE_DATA")

BASE_DIR.mkdir(exist_ok=True)

zip_files = [
    "/content/datasets.zip",
    "/content/archive.zip",
    "/content/ctsoc-ewaste-main.zip",
    "/content/E-waste.v9-roboflow-instant-1--eval-.tensorflow.zip"
]

for zip_file in zip_files:

    if not os.path.exists(zip_file):
        print("❌ Missing:", zip_file)
        continue

    print("Extracting:", os.path.basename(zip_file))

    with zipfile.ZipFile(zip_file, "r") as z:
        z.extractall(BASE_DIR)

    print("✓ Done")

print("\nAll extraction completed.")
```

```python
from pathlib import Path
from collections import Counter
import os

IMAGE_EXTENSIONS = {".jpg", ".jpeg", ".png", ".bmp", ".webp"}

print("=" * 80)
print("ECOBRIDGE — STEP 16: COMPLETE DATASET AUDIT")
print("=" * 80)

# ---------------------------------------------------------
# 1. Find all image files
# ---------------------------------------------------------
all_images = []

for path in BASE_DIR.rglob("*"):
    if path.is_file() and path.suffix.lower() in IMAGE_EXTENSIONS:
        all_images.append(path)

print(f"\nTotal image files found: {len(all_images):,}")


# ---------------------------------------------------------
# 2. Categorize images according to their source
# ---------------------------------------------------------
source_counts = Counter()

for path in all_images:
    p = str(path).lower()

    if "/datasets/ai_ml/vision/" in p:
        source = "Original ECOBRIDGE"

    elif "/archive/" in p or "archive" in p:
        source = "Archive Dataset"

    elif "ctsoc-ewaste-main" in p:
        source = "CTSoc"

    elif "e-waste.v9" in p or "roboflow" in p:
        source = "Roboflow E-waste"

    else:
        source = "Other / Unknown"

    source_counts[source] += 1


print("\n" + "-" * 60)
print("IMAGE COUNT BY SOURCE")
print("-" * 60)

for source, count in source_counts.most_common():
    print(f"{source:30s} : {count:,}")


# ---------------------------------------------------------
# 3. Original ECOBRIDGE dataset
# ---------------------------------------------------------
print("\n" + "=" * 80)
print("ORIGINAL ECOBRIDGE VISION DATASET")
print("=" * 80)

vision_dir = BASE_DIR / "datasets" / "ai_ml" / "vision"

for split in ["images/train", "images/val", "images/test"]:

    split_dir = vision_dir / split

    if not split_dir.exists():
        print(f"\n{split}: NOT FOUND")
        continue

    print(f"\n{split}:")

    total = 0

    for class_dir in sorted(split_dir.iterdir()):

        if class_dir.is_dir():

            count = sum(
                1 for p in class_dir.iterdir()
                if p.suffix.lower() in IMAGE_EXTENSIONS
            )

            print(f"  {class_dir.name:25s} : {count}")
            total += count

    print(f"  {'TOTAL':25s} : {total}")


# ---------------------------------------------------------
# 4. Archive dataset
# ---------------------------------------------------------
print("\n" + "=" * 80)
print("ARCHIVE DATASET")
print("=" * 80)

archive_candidates = []

for path in BASE_DIR.rglob("*"):
    if path.is_dir() and path.name.lower() == "train":
        if "archive" in str(path).lower():
            archive_candidates.append(path)

for train_dir in archive_candidates:

    print("\nTRAIN:", train_dir)

    for class_dir in sorted(train_dir.iterdir()):

        if class_dir.is_dir():

            count = sum(
                1 for p in class_dir.rglob("*")
                if p.is_file() and p.suffix.lower() in IMAGE_EXTENSIONS
            )

            print(f"  {class_dir.name:25s} : {count}")


# ---------------------------------------------------------
# 5. CTSoc dataset
# ---------------------------------------------------------
print("\n" + "=" * 80)
print("CTSOC DATASET")
print("=" * 80)

ctsoc_dir = BASE_DIR / "ctsoc-ewaste-main"

if ctsoc_dir.exists():

    ctsoc_images = [
        p for p in ctsoc_dir.rglob("*")
        if p.is_file() and p.suffix.lower() in IMAGE_EXTENSIONS
    ]

    print(f"Total CTSoc images found: {len(ctsoc_images)}")

    print("\nImages by folder:")

    folder_counts = Counter(
        str(p.parent.relative_to(ctsoc_dir))
        for p in ctsoc_images
    )

    for folder, count in folder_counts.most_common():
        print(f"  {folder:50s} : {count}")


# ---------------------------------------------------------
# 6. Roboflow dataset
# ---------------------------------------------------------
print("\n" + "=" * 80)
print("ROBOFLOW E-WASTE DATASET")
print("=" * 80)

roboflow_images = []

for path in BASE_DIR.rglob("*"):

    if (
        path.is_file()
        and path.suffix.lower() in IMAGE_EXTENSIONS
        and (
            "roboflow" in str(path).lower()
            or "e-waste.v9" in str(path).lower()
        )
    ):
        roboflow_images.append(path)

print(f"Roboflow images found: {len(roboflow_images)}")

if roboflow_images:

    folder_counts = Counter(
        str(p.parent)
        for p in roboflow_images
    )

    print("\nImages by folder:")

    for folder, count in folder_counts.most_common():
        print(f"  {folder} : {count}")


# ---------------------------------------------------------
# 7. Final summary
# ---------------------------------------------------------
print("\n" + "=" * 80)
print("AUDIT COMPLETE")
print("=" * 80)
```

```python
# ============================================================
# STEP 19 — SAFE ECOBRIDGE CLASS MAPPING
# ============================================================

ECOBRIDGE_CLASSES = [
    "CAT_CRT",
    "CAT_LCD_LED",
    "CAT_PCB",
    "CAT_CABLES",
    "CAT_BATTERY",
    "CAT_MOTORS",
    "CAT_PLASTICS",
    "CAT_OTHER"
]

SAFE_MAPPING = {
    "Battery": "CAT_BATTERY",
    "PCB": "CAT_PCB",

    "Keyboard": "CAT_OTHER",
    "Microwave": "CAT_OTHER",
    "Mobile": "CAT_OTHER",
    "Mouse": "CAT_OTHER",
    "Player": "CAT_OTHER",
    "Printer": "CAT_OTHER",
    "Washing Machine": "CAT_OTHER",
}

MODIFIED_DIR = BASE_DIR / "modified-dataset"

print("=" * 80)
print("STEP 19 — SAFE CLASS MAPPING")
print("=" * 80)

print("\nECOBRIDGE classes:")
for i, cls in enumerate(ECOBRIDGE_CLASSES):
    print(f"{i}: {cls}")

print("\n" + "-" * 80)
print("MAPPING")
print("-" * 80)

for source_class, target_class in SAFE_MAPPING.items():

    source_dir = MODIFIED_DIR / "train" / source_class

    if source_dir.exists():
        image_count = sum(
            1
            for p in source_dir.iterdir()
            if p.is_file() and p.suffix.lower() in IMAGE_EXTENSIONS
        )

        print(
            f"{source_class:20s} -> "
            f"{target_class:20s} "
            f"({image_count} train images)"
        )

    else:
        print(
            f"❌ Missing source folder: {source_class}"
        )

print("\n" + "-" * 80)
print("UNMAPPED CLASSES")
print("-" * 80)

all_modified_classes = {
    p.name
    for p in (MODIFIED_DIR / "train").iterdir()
    if p.is_dir()
}

unmapped = all_modified_classes - set(SAFE_MAPPING.keys())

for cls in sorted(unmapped):
    print(f"⚠️ {cls}")

print("\n" + "=" * 80)
print("Mapping created successfully.")
print("Television intentionally remains UNMAPPED.")
print("=" * 80)
```

```python
# ============================================================
# STEP 20 — BUILD CLEAN ECOBRIDGE MASTER DATASET
# ============================================================

import shutil
import csv
from pathlib import Path

print("=" * 80)
print("STEP 20 — BUILDING ECOBRIDGE MASTER DATASET")
print("=" * 80)

# ------------------------------------------------------------
# Paths
# ------------------------------------------------------------

MASTER_DIR = BASE_DIR / "ECOBRIDGE_MASTER_DATASET"

# Remove an old master dataset if this cell was run before
if MASTER_DIR.exists():
    print("\nRemoving previous master dataset...")
    shutil.rmtree(MASTER_DIR)

# Create directories
for split in ["train", "val", "test"]:
    for class_name in ECOBRIDGE_CLASSES:
        (MASTER_DIR / split / class_name).mkdir(
            parents=True,
            exist_ok=True
        )

metadata = []

# ------------------------------------------------------------
# Helper function
# ------------------------------------------------------------

def copy_image(
    source_path,
    destination_dir,
    source_dataset,
    source_class,
    target_class,
    source_split
):
    """
    Copy an image to the master dataset with a unique filename.
    """

    source_path = Path(source_path)
    destination_dir = Path(destination_dir)

    # Create unique filename
    new_name = (
        f"{source_dataset}__"
        f"{source_split}__"
        f"{source_class}__"
        f"{source_path.name}"
    )

    destination_path = destination_dir / new_name

    shutil.copy2(source_path, destination_path)

    metadata.append({
        "filename": new_name,
        "target_class": target_class,
        "source_dataset": source_dataset,
        "source_class": source_class,
        "source_split": source_split,
        "original_path": str(source_path)
    })


# ============================================================
# PART 1 — ORIGINAL ECOBRIDGE DATASET
# ============================================================

print("\n" + "-" * 80)
print("ADDING ORIGINAL ECOBRIDGE DATA")
print("-" * 80)

VISION_DIR = BASE_DIR / "datasets" / "ai_ml" / "vision"

# Original ECOBRIDGE train images
original_train = VISION_DIR / "images" / "train"

for image_path in sorted(original_train.iterdir()):

    if not image_path.is_file():
        continue

    if image_path.suffix.lower() not in IMAGE_EXTENSIONS:
        continue

    filename = image_path.name.lower()

    # Determine class from filename
    if filename.startswith("batteries_"):
        target = "CAT_BATTERY"
        source_class = "BATTERIES"

    elif filename.startswith("cables_"):
        target = "CAT_CABLES"
        source_class = "CABLES"

    elif filename.startswith("crt_"):
        target = "CAT_CRT"
        source_class = "CRT"

    elif filename.startswith("lcd_led_panel_"):
        target = "CAT_LCD_LED"
        source_class = "LCD_LED_PANEL"

    elif filename.startswith("mixed_plastics_"):
        target = "CAT_PLASTICS"
        source_class = "MIXED_PLASTICS"

    elif filename.startswith("motors_magnets_"):
        target = "CAT_MOTORS"
        source_class = "MOTORS_MAGNETS"

    elif filename.startswith("other_ewaste_"):
        target = "CAT_OTHER"
        source_class = "OTHER_EWASTE"

    elif filename.startswith("pcb_"):
        target = "CAT_PCB"
        source_class = "PCB"

    else:
        print("⚠️ Could not identify:", image_path.name)
        continue

    copy_image(
        image_path,
        MASTER_DIR / "train" / target,
        "ECOBRIDGE",
        source_class,
        target,
        "train"
    )


# Original ECOBRIDGE validation images
original_val = VISION_DIR / "images" / "val"

for image_path in sorted(original_val.iterdir()):

    if not image_path.is_file():
        continue

    if image_path.suffix.lower() not in IMAGE_EXTENSIONS:
        continue

    filename = image_path.name.lower()

    if filename.startswith("batteries_"):
        target = "CAT_BATTERY"
        source_class = "BATTERIES"

    elif filename.startswith("cables_"):
        target = "CAT_CABLES"
        source_class = "CABLES"

    elif filename.startswith("crt_"):
        target = "CAT_CRT"
        source_class = "CRT"

    elif filename.startswith("lcd_led_panel_"):
        target = "CAT_LCD_LED"
        source_class = "LCD_LED_PANEL"

    elif filename.startswith("mixed_plastics_"):
        target = "CAT_PLASTICS"
        source_class = "MIXED_PLASTICS"

    elif filename.startswith("motors_magnets_"):
        target = "CAT_MOTORS"
        source_class = "MOTORS_MAGNETS"

    elif filename.startswith("other_ewaste_"):
        target = "CAT_OTHER"
        source_class = "OTHER_EWASTE"

    elif filename.startswith("pcb_"):
        target = "CAT_PCB"
        source_class = "PCB"

    else:
        print("⚠️ Could not identify:", image_path.name)
        continue

    copy_image(
        image_path,
        MASTER_DIR / "val" / target,
        "ECOBRIDGE",
        source_class,
        target,
        "val"
    )


# ============================================================
# PART 2 — MODIFIED DATASET
# ============================================================

print("\n" + "-" * 80)
print("ADDING MODIFIED DATASET")
print("-" * 80)

for source_split, target_split in [
    ("train", "train"),
    ("val", "val"),
    ("test", "test")
]:

    split_dir = MODIFIED_DIR / source_split

    for source_class, target_class in SAFE_MAPPING.items():

        source_dir = split_dir / source_class

        if not source_dir.exists():
            print(
                f"⚠️ Missing: {source_split}/{source_class}"
            )
            continue

        images = [
            p for p in source_dir.iterdir()
            if p.is_file()
            and p.suffix.lower() in IMAGE_EXTENSIONS
        ]

        for image_path in sorted(images):

            copy_image(
                image_path,
                MASTER_DIR / target_split / target_class,
                "MODIFIED_DATASET",
                source_class,
                target_class,
                source_split
            )


# ============================================================
# PART 3 — SAVE METADATA
# ============================================================

metadata_file = MASTER_DIR / "metadata.csv"

with open(
    metadata_file,
    "w",
    newline="",
    encoding="utf-8"
) as f:

    writer = csv.DictWriter(
        f,
        fieldnames=[
            "filename",
            "target_class",
            "source_dataset",
            "source_class",
            "source_split",
            "original_path"
        ]
    )

    writer.writeheader()
    writer.writerows(metadata)


# ============================================================
# PART 4 — SUMMARY
# ============================================================

print("\n" + "=" * 80)
print("MASTER DATASET CREATED")
print("=" * 80)

print(f"\nLocation:")
print(MASTER_DIR)

print(f"\nTotal images copied: {len(metadata):,}")

print("\nImages by split and class:")

for split in ["train", "val", "test"]:

    print(f"\n{split.upper()}")

    split_total = 0

    for class_name in ECOBRIDGE_CLASSES:

        class_dir = MASTER_DIR / split / class_name

        count = sum(
            1
            for p in class_dir.iterdir()
            if p.is_file()
            and p.suffix.lower() in IMAGE_EXTENSIONS
        )

        print(f"  {class_name:20s} : {count}")
        split_total += count

    print(f"  {'TOTAL':20s} : {split_total}")


print("\n" + "=" * 80)
print("STEP 20 COMPLETE")
print("=" * 80)
```

```python
# ============================================================
# STEP 21 — MASTER DATASET QUALITY & SOURCE AUDIT
# ============================================================

import pandas as pd
from pathlib import Path

print("=" * 80)
print("STEP 21 — MASTER DATASET QUALITY AUDIT")
print("=" * 80)

MASTER_DIR = BASE_DIR / "ECOBRIDGE_MASTER_DATASET"

# ------------------------------------------------------------
# Load metadata
# ------------------------------------------------------------

metadata_path = MASTER_DIR / "metadata.csv"

df = pd.read_csv(metadata_path)

print("\nTotal metadata records:", len(df))

# ------------------------------------------------------------
# 1. Source dataset × target class
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("SOURCE DATASET × TARGET CLASS")
print("=" * 80)

source_class_table = pd.crosstab(
    df["target_class"],
    df["source_dataset"]
)

print(source_class_table.to_string())


# ------------------------------------------------------------
# 2. Source class × target class
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("SOURCE CLASS × TARGET CLASS")
print("=" * 80)

source_target_table = pd.crosstab(
    df["source_class"],
    df["target_class"]
)

print(source_target_table.to_string())


# ------------------------------------------------------------
# 3. Split × target class
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("SPLIT × TARGET CLASS")
print("=" * 80)

split_class_table = pd.crosstab(
    df["target_class"],
    df["source_split"]
)

print(split_class_table.to_string())


# ------------------------------------------------------------
# 4. Total class counts
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("TOTAL IMAGES PER CLASS")
print("=" * 80)

class_counts = (
    df["target_class"]
    .value_counts()
    .reindex(ECOBRIDGE_CLASSES, fill_value=0)
)

for cls, count in class_counts.items():
    print(f"{cls:20s} : {count}")


# ------------------------------------------------------------
# 5. Check whether every class has test images
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("TEST SET CHECK")
print("=" * 80)

test_counts = (
    df[df["source_split"] == "test"]["target_class"]
    .value_counts()
    .reindex(ECOBRIDGE_CLASSES, fill_value=0)
)

for cls, count in test_counts.items():

    if count == 0:
        print(f"⚠️ {cls:20s} : NO TEST IMAGES")
    else:
        print(f"✓  {cls:20s} : {count}")


# ------------------------------------------------------------
# 6. Check for duplicate filenames
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("DUPLICATE FILENAME CHECK")
print("=" * 80)

duplicate_filenames = df[
    df["filename"].duplicated(keep=False)
]

if len(duplicate_filenames) == 0:
    print("✓ No duplicate filenames")
else:
    print(
        f"⚠️ Duplicate filenames found: "
        f"{len(duplicate_filenames)}"
    )
    print(duplicate_filenames.head(20).to_string(index=False))


# ------------------------------------------------------------
# 7. Final warning
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("STEP 21 AUDIT COMPLETE")
print("=" * 80)

print("""
IMPORTANT:
Do not train yet.

We need to solve:
1. Severe CAT_OTHER imbalance
2. Missing test images for 5 classes
3. Very small specialized classes
4. Potential source-domain imbalance

Next step will create a better training/evaluation strategy
without inventing labels.
""")
```

```python
# ============================================================
# STEP 25A — ROBOFLOW CLASS COUNTS
# ============================================================

import pandas as pd
from pathlib import Path

print("=" * 80)
print("STEP 25A — ROBOFLOW ANNOTATION CLASS COUNTS")
print("=" * 80)

ROBOFLOW_SPLITS = {
    "train": BASE_DIR / "train",
    "valid": BASE_DIR / "valid",
    "test": BASE_DIR / "test"
}

all_annotations = []

for split_name, split_dir in ROBOFLOW_SPLITS.items():

    csv_path = split_dir / "_annotations.csv"

    df_rf = pd.read_csv(csv_path)
    df_rf["dataset_split"] = split_name

    all_annotations.append(df_rf)

    print("\n" + "-" * 70)
    print(split_name.upper())
    print("-" * 70)

    print("Total annotation rows:", len(df_rf))

    print("\nClass counts:")

    print(
        df_rf["class"]
        .value_counts()
        .to_string()
    )


# Combine all annotations

rf_all = pd.concat(
    all_annotations,
    ignore_index=True
)

print("\n" + "=" * 80)
print("ALL ROBOFLOW ANNOTATIONS")
print("=" * 80)

print("Total annotation rows:", len(rf_all))

print("\nOverall class counts:")

print(
    rf_all["class"]
    .value_counts()
    .to_string()
)

print("\nUnique annotated images:")

print(
    rf_all["filename"].nunique()
)

print("\nBattery-related annotations:")

battery_labels = ["Battery", "Powerbank"]

print(
    rf_all[
        rf_all["class"].isin(battery_labels)
    ]["class"]
    .value_counts()
    .to_string()
)
```

```python
# ============================================================
# STEP 25B — CHECK SOURCE IMAGE OVERLAP
# ============================================================

import re

print("=" * 80)
print("STEP 25B — ROBOFLOW SOURCE OVERLAP CHECK")
print("=" * 80)


def get_source_id(filename):
    """
    Remove the Roboflow-generated hash.

    Example:
    image_jpg.rf.abcdef123.jpg

    becomes approximately:
    image
    """

    name = Path(filename).name

    # Remove Roboflow .rf hash
    name = re.sub(
        r"\.rf\.[a-f0-9]+(?=\.[^.]+$)",
        "",
        name,
        flags=re.IGNORECASE
    )

    return name


rf_all["source_id"] = rf_all["filename"].apply(
    get_source_id
)


# ------------------------------------------------------------
# Unique source IDs by split
# ------------------------------------------------------------

for split_name in ["train", "valid", "test"]:

    subset = rf_all[
        rf_all["dataset_split"] == split_name
    ]

    print(
        f"{split_name:8s}: "
        f"{subset['source_id'].nunique()} unique source IDs"
    )


# ------------------------------------------------------------
# Find source IDs appearing in multiple splits
# ------------------------------------------------------------

source_split_counts = (
    rf_all
    .groupby("source_id")["dataset_split"]
    .nunique()
)

overlapping_sources = source_split_counts[
    source_split_counts > 1
]

print("\n" + "=" * 80)
print("CROSS-SPLIT SOURCE OVERLAP")
print("=" * 80)

print(
    "Source IDs appearing in multiple splits:",
    len(overlapping_sources)
)

if len(overlapping_sources) > 0:

    print("\n⚠️ Potential leakage detected.")

    print(
        overlapping_sources
        .sort_values(ascending=False)
        .head(30)
        .to_string()
    )

else:

    print(
        "✓ No source IDs were found across multiple splits."
    )
```

```python
# ============================================================
# STEP 26 — REBUILD ROBOFLOW BATTERY DATASET
#              WITHOUT CROSS-SPLIT SOURCE LEAKAGE
# ============================================================

import os
import shutil
import re
import pandas as pd
from pathlib import Path
from sklearn.model_selection import GroupShuffleSplit

print("=" * 80)
print("STEP 26 — REBUILD ROBOFLOW BATTERY DATASET")
print("=" * 80)

# ------------------------------------------------------------
# 1. LOAD ROBOFLOW ANNOTATIONS
# ------------------------------------------------------------

ROBOFLOW_SPLITS = {
    "train": BASE_DIR / "train",
    "valid": BASE_DIR / "valid",
    "test": BASE_DIR / "test"
}

annotation_list = []

for split_name, split_dir in ROBOFLOW_SPLITS.items():

    csv_path = split_dir / "_annotations.csv"

    df = pd.read_csv(csv_path)

    df["original_split"] = split_name

    annotation_list.append(df)

rf_all = pd.concat(
    annotation_list,
    ignore_index=True
)

print("Total annotation rows:", len(rf_all))


# ------------------------------------------------------------
# 2. KEEP ONLY BATTERY + POWERBANK
# ------------------------------------------------------------

BATTERY_CLASSES = [
    "Battery",
    "Powerbank"
]

battery_df = rf_all[
    rf_all["class"].isin(BATTERY_CLASSES)
].copy()

print("\nBattery-related annotation rows:")
print(len(battery_df))

print("\nClass distribution:")
print(
    battery_df["class"]
    .value_counts()
)


# ------------------------------------------------------------
# 3. CREATE ECOBRIDGE CLASS
# ------------------------------------------------------------

battery_df["mapped_class"] = "CAT_BATTERY"


# ------------------------------------------------------------
# 4. CREATE SOURCE ID
# ------------------------------------------------------------

def get_source_id(filename):

    name = Path(filename).name

    name = re.sub(
        r"\.rf\.[a-f0-9]+(?=\.[^.]+$)",
        "",
        name,
        flags=re.IGNORECASE
    )

    return name


battery_df["source_id"] = battery_df["filename"].apply(
    get_source_id
)


# ------------------------------------------------------------
# 5. SOURCE GROUP SUMMARY
# ------------------------------------------------------------

source_summary = (
    battery_df
    .groupby("source_id")
    .agg(
        annotations=("filename", "count"),
        classes=("class", lambda x: ", ".join(sorted(set(x)))),
        original_splits=(
            "original_split",
            lambda x: ", ".join(sorted(set(x)))
        )
    )
    .reset_index()
)

print("\nUnique source images:")
print(source_summary["source_id"].nunique())

print("\nSources appearing in multiple original splits:")

multi_split = source_summary[
    source_summary["original_splits"].str.contains(",")
]

print(len(multi_split))

print("\nFirst few:")
print(
    multi_split.head(20).to_string(index=False)
)


# ------------------------------------------------------------
# 6. CREATE NEW GROUP-BASED SPLIT
# ------------------------------------------------------------

# First split:
# 80% training
# 20% temporary

groups = battery_df["source_id"]

gss1 = GroupShuffleSplit(
    n_splits=1,
    test_size=0.20,
    random_state=42
)

train_idx, temp_idx = next(
    gss1.split(
        battery_df,
        groups=groups
    )
)

train_df = battery_df.iloc[train_idx].copy()
temp_df = battery_df.iloc[temp_idx].copy()


# ------------------------------------------------------------
# 7. SPLIT TEMP INTO VALIDATION + TEST
# ------------------------------------------------------------

temp_groups = temp_df["source_id"]

gss2 = GroupShuffleSplit(
    n_splits=1,
    test_size=0.50,
    random_state=42
)

valid_idx, test_idx = next(
    gss2.split(
        temp_df,
        groups=temp_groups
    )
)

valid_df = temp_df.iloc[valid_idx].copy()
test_df = temp_df.iloc[test_idx].copy()


# ------------------------------------------------------------
# 8. DISPLAY NEW SPLIT
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("NEW GROUP-BASED SPLIT")
print("=" * 80)

for name, df in [
    ("TRAIN", train_df),
    ("VALIDATION", valid_df),
    ("TEST", test_df)
]:

    print(f"\n{name}")

    print("Annotations:", len(df))

    print(
        "Unique source IDs:",
        df["source_id"].nunique()
    )

    print(
        "Original Roboflow splits:",
        sorted(df["original_split"].unique())
    )


# ------------------------------------------------------------
# 9. VERIFY NO SOURCE LEAKAGE
# ------------------------------------------------------------

train_sources = set(train_df["source_id"])
valid_sources = set(valid_df["source_id"])
test_sources = set(test_df["source_id"])

print("\n" + "=" * 80)
print("LEAKAGE CHECK")
print("=" * 80)

print(
    "Train ∩ Valid:",
    len(train_sources & valid_sources)
)

print(
    "Train ∩ Test:",
    len(train_sources & test_sources)
)

print(
    "Valid ∩ Test:",
    len(valid_sources & test_sources)
)

if (
    len(train_sources & valid_sources) == 0
    and
    len(train_sources & test_sources) == 0
    and
    len(valid_sources & test_sources) == 0
):

    print("\n✅ NO SOURCE LEAKAGE")

else:

    print("\n❌ SOURCE LEAKAGE STILL EXISTS")
```

```python
# ============================================================
# STEP 27 — CHECK ROBOFLOW IMAGE FILES
# ============================================================

print("=" * 80)
print("STEP 27 — CHECK ROBOFLOW IMAGE FILES")
print("=" * 80)

for split_name, split_dir in ROBOFLOW_SPLITS.items():

    image_files = []

    for ext in ["*.jpg", "*.jpeg", "*.png"]:

        image_files.extend(
            split_dir.glob(ext)
        )

    print(
        f"{split_name:8s}: "
        f"{len(image_files)} image files"
    )


print("\nChecking annotation filenames...")

missing = []

for _, row in battery_df.iterrows():

    split_dir = ROBOFLOW_SPLITS[
        row["original_split"]
    ]

    image_path = split_dir / row["filename"]

    if not image_path.exists():

        missing.append(
            (
                row["original_split"],
                row["filename"]
            )
        )


print(
    "Missing annotated image files:",
    len(missing)
)

if len(missing) == 0:

    print("✅ All battery annotation images found.")

else:

    print("⚠️ Some annotation images are missing.")

    print(
        pd.DataFrame(
            missing,
            columns=["split", "filename"]
        ).head(20)
    )
```

```python
# ============================================================
# STEP 28 — EXTRACT BATTERY / POWERBANK CROPS
# ============================================================

import os
import cv2
import pandas as pd
from pathlib import Path

print("=" * 80)
print("STEP 28 — EXTRACT BATTERY / POWERBANK CROPS")
print("=" * 80)


# ------------------------------------------------------------
# 1. CREATE OUTPUT DIRECTORIES
# ------------------------------------------------------------

CROP_BASE = BASE_DIR / "processed_battery"

CROP_TRAIN = CROP_BASE / "train" / "CAT_BATTERY"
CROP_VALID = CROP_BASE / "valid" / "CAT_BATTERY"
CROP_TEST  = CROP_BASE / "test" / "CAT_BATTERY"

for folder in [
    CROP_TRAIN,
    CROP_VALID,
    CROP_TEST
]:
    folder.mkdir(
        parents=True,
        exist_ok=True
    )

print("\nOutput directory:")
print(CROP_BASE)


# ------------------------------------------------------------
# 2. FUNCTION TO EXTRACT CROP
# ------------------------------------------------------------

def extract_crop(
    row,
    output_dir,
    crop_id
):

    # Original image location
    split_dir = ROBOFLOW_SPLITS[
        row["original_split"]
    ]

    image_path = split_dir / row["filename"]

    # Read image
    image = cv2.imread(
        str(image_path)
    )

    if image is None:
        return False, "image_read_error"

    height, width = image.shape[:2]

    # Bounding box
    xmin = int(row["xmin"])
    ymin = int(row["ymin"])
    xmax = int(row["xmax"])
    ymax = int(row["ymax"])

    # --------------------------------------------------------
    # Clamp coordinates to image boundaries
    # --------------------------------------------------------

    xmin = max(0, min(xmin, width - 1))
    ymin = max(0, min(ymin, height - 1))
    xmax = max(0, min(xmax, width))
    ymax = max(0, min(ymax, height))

    # --------------------------------------------------------
    # Validate bounding box
    # --------------------------------------------------------

    if xmax <= xmin:
        return False, "invalid_width"

    if ymax <= ymin:
        return False, "invalid_height"

    crop_width = xmax - xmin
    crop_height = ymax - ymin

    if crop_width < 10:
        return False, "crop_too_small"

    if crop_height < 10:
        return False, "crop_too_small"

    # --------------------------------------------------------
    # Crop object
    # --------------------------------------------------------

    crop = image[
        ymin:ymax,
        xmin:xmax
    ]

    if crop.size == 0:
        return False, "empty_crop"

    # --------------------------------------------------------
    # Save crop
    # --------------------------------------------------------

    output_path = (
        output_dir /
        f"{crop_id:05d}.jpg"
    )

    success = cv2.imwrite(
        str(output_path),
        crop
    )

    if not success:
        return False, "save_error"

    return True, str(output_path)


# ------------------------------------------------------------
# 3. PROCESS EACH SPLIT
# ------------------------------------------------------------

split_data = {
    "train": train_df,
    "valid": valid_df,
    "test": test_df
}

output_dirs = {
    "train": CROP_TRAIN,
    "valid": CROP_VALID,
    "test": CROP_TEST
}

crop_records = []

crop_counter = 0

for split_name, df in split_data.items():

    print("\n" + "-" * 70)
    print(f"PROCESSING {split_name.upper()}")
    print("-" * 70)

    success_count = 0
    failed_count = 0

    for _, row in df.iterrows():

        crop_counter += 1

        success, result = extract_crop(
            row,
            output_dirs[split_name],
            crop_counter
        )

        if success:

            success_count += 1

            crop_records.append({
                "crop_id": crop_counter,
                "crop_path": result,
                "dataset_split": split_name,
                "mapped_class": "CAT_BATTERY",
                "source_class": row["class"],
                "original_filename": row["filename"],
                "source_id": row["source_id"],
                "original_split": row["original_split"],
                "xmin": row["xmin"],
                "ymin": row["ymin"],
                "xmax": row["xmax"],
                "ymax": row["ymax"]
            })

        else:

            failed_count += 1

            crop_records.append({
                "crop_id": crop_counter,
                "crop_path": None,
                "dataset_split": split_name,
                "mapped_class": "CAT_BATTERY",
                "source_class": row["class"],
                "original_filename": row["filename"],
                "source_id": row["source_id"],
                "original_split": row["original_split"],
                "xmin": row["xmin"],
                "ymin": row["ymin"],
                "xmax": row["xmax"],
                "ymax": row["ymax"],
                "failure_reason": result
            })

    print("Successful crops:", success_count)
    print("Failed crops:", failed_count)


# ------------------------------------------------------------
# 4. SAVE METADATA
# ------------------------------------------------------------

crop_metadata = pd.DataFrame(
    crop_records
)

metadata_path = (
    CROP_BASE /
    "battery_crop_metadata.csv"
)

crop_metadata.to_csv(
    metadata_path,
    index=False
)


# ------------------------------------------------------------
# 5. FINAL SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("STEP 28 SUMMARY")
print("=" * 80)

print(
    "Total annotation rows processed:",
    len(crop_metadata)
)

print(
    "Successful crops:",
    crop_metadata["crop_path"].notna().sum()
)

print(
    "Failed crops:",
    crop_metadata["crop_path"].isna().sum()
)

print("\nCrops by split:")

print(
    crop_metadata[
        crop_metadata["crop_path"].notna()
    ]
    .groupby("dataset_split")
    .size()
    .to_string()
)

print("\nCrops by original class:")

print(
    crop_metadata[
        crop_metadata["crop_path"].notna()
    ]["source_class"]
    .value_counts()
    .to_string()
)

print("\nMetadata saved to:")
print(metadata_path)
```

```python
# ============================================================
# STEP 29 — BATTERY CROP QUALITY ANALYSIS
# ============================================================

import cv2
import numpy as np
import pandas as pd
from pathlib import Path

print("=" * 80)
print("STEP 29 — BATTERY CROP QUALITY ANALYSIS")
print("=" * 80)

# ------------------------------------------------------------
# 1. LOAD METADATA
# ------------------------------------------------------------

metadata_path = (
    BASE_DIR /
    "processed_battery" /
    "battery_crop_metadata.csv"
)

crop_metadata = pd.read_csv(metadata_path)

valid_metadata = crop_metadata[
    crop_metadata["crop_path"].notna()
].copy()

print("\nTotal valid crops:", len(valid_metadata))


# ------------------------------------------------------------
# 2. CALCULATE IMAGE PROPERTIES
# ------------------------------------------------------------

widths = []
heights = []
areas = []
aspect_ratios = []
file_sizes = []
mean_brightness = []

for _, row in valid_metadata.iterrows():

    image_path = Path(row["crop_path"])

    image = cv2.imread(
        str(image_path)
    )

    if image is None:
        widths.append(0)
        heights.append(0)
        areas.append(0)
        aspect_ratios.append(0)
        file_sizes.append(0)
        mean_brightness.append(0)
        continue

    h, w = image.shape[:2]

    widths.append(w)
    heights.append(h)
    areas.append(w * h)

    if h > 0:
        aspect_ratios.append(w / h)
    else:
        aspect_ratios.append(0)

    file_sizes.append(
        image_path.stat().st_size
    )

    gray = cv2.cvtColor(
        image,
        cv2.COLOR_BGR2GRAY
    )

    mean_brightness.append(
        float(np.mean(gray))
    )


valid_metadata["crop_width"] = widths
valid_metadata["crop_height"] = heights
valid_metadata["crop_area"] = areas
valid_metadata["aspect_ratio"] = aspect_ratios
valid_metadata["file_size_bytes"] = file_sizes
valid_metadata["mean_brightness"] = mean_brightness


# ------------------------------------------------------------
# 3. BASIC STATISTICS
# ------------------------------------------------------------

print("\n" + "-" * 70)
print("IMAGE SIZE STATISTICS")
print("-" * 70)

print(
    valid_metadata[
        [
            "crop_width",
            "crop_height",
            "crop_area",
            "aspect_ratio",
            "file_size_bytes",
            "mean_brightness"
        ]
    ].describe()
)


# ------------------------------------------------------------
# 4. VERY SMALL CROPS
# ------------------------------------------------------------

small_crops = valid_metadata[
    (valid_metadata["crop_width"] < 32) |
    (valid_metadata["crop_height"] < 32)
]

print("\n" + "-" * 70)
print("VERY SMALL CROPS")
print("-" * 70)

print(
    "Crops with width OR height < 32 pixels:",
    len(small_crops)
)

if len(small_crops) > 0:

    print(
        small_crops[
            [
                "crop_id",
                "source_class",
                "crop_width",
                "crop_height",
                "source_id"
            ]
        ]
        .head(20)
        .to_string(index=False)
    )


# ------------------------------------------------------------
# 5. EXTREME ASPECT RATIOS
# ------------------------------------------------------------

extreme_ratio = valid_metadata[
    (valid_metadata["aspect_ratio"] > 5) |
    (valid_metadata["aspect_ratio"] < 0.2)
]

print("\n" + "-" * 70)
print("EXTREME ASPECT RATIOS")
print("-" * 70)

print(
    "Crops with aspect ratio > 5 or < 0.2:",
    len(extreme_ratio)
)

if len(extreme_ratio) > 0:

    print(
        extreme_ratio[
            [
                "crop_id",
                "source_class",
                "crop_width",
                "crop_height",
                "aspect_ratio",
                "source_id"
            ]
        ]
        .head(20)
        .to_string(index=False)
    )


# ------------------------------------------------------------
# 6. VERY DARK / VERY BRIGHT CROPS
# ------------------------------------------------------------

dark_crops = valid_metadata[
    valid_metadata["mean_brightness"] < 20
]

bright_crops = valid_metadata[
    valid_metadata["mean_brightness"] > 245
]

print("\n" + "-" * 70)
print("BRIGHTNESS CHECK")
print("-" * 70)

print(
    "Very dark crops (<20):",
    len(dark_crops)
)

print(
    "Very bright crops (>245):",
    len(bright_crops)
)


# ------------------------------------------------------------
# 7. CHECK ZERO/INVALID VALUES
# ------------------------------------------------------------

invalid_dimensions = valid_metadata[
    (valid_metadata["crop_width"] <= 0) |
    (valid_metadata["crop_height"] <= 0)
]

print("\n" + "-" * 70)
print("INVALID DIMENSIONS")
print("-" * 70)

print(
    "Invalid crops:",
    len(invalid_dimensions)
)


# ------------------------------------------------------------
# 8. CHECK DUPLICATE SOURCE + BBOX
# ------------------------------------------------------------

duplicate_annotations = valid_metadata[
    valid_metadata.duplicated(
        subset=[
            "source_id",
            "xmin",
            "ymin",
            "xmax",
            "ymax"
        ],
        keep=False
    )
]

print("\n" + "-" * 70)
print("DUPLICATE ANNOTATION CHECK")
print("-" * 70)

print(
    "Potential duplicate annotations:",
    len(duplicate_annotations)
)


# ------------------------------------------------------------
# 9. SAVE ENHANCED METADATA
# ------------------------------------------------------------

enhanced_metadata_path = (
    BASE_DIR /
    "processed_battery" /
    "battery_crop_metadata_enhanced.csv"
)

valid_metadata.to_csv(
    enhanced_metadata_path,
    index=False
)

print("\n" + "=" * 80)
print("STEP 29 COMPLETE")
print("=" * 80)

print(
    "Enhanced metadata saved to:"
)

print(enhanced_metadata_path)
```

```python
# ============================================================
# STEP 30 — VISUAL INSPECTION OF BATTERY CROPS
# ============================================================

import matplotlib.pyplot as plt
import cv2
import pandas as pd
import numpy as np
from pathlib import Path

print("=" * 80)
print("STEP 30 — VISUAL INSPECTION OF BATTERY CROPS")
print("=" * 80)


# ------------------------------------------------------------
# 1. LOAD ENHANCED METADATA
# ------------------------------------------------------------

enhanced_metadata_path = (
    BASE_DIR /
    "processed_battery" /
    "battery_crop_metadata_enhanced.csv"
)

battery_meta = pd.read_csv(
    enhanced_metadata_path
)

print(
    "Total crops:",
    len(battery_meta)
)


# ------------------------------------------------------------
# 2. FUNCTION TO DISPLAY CROPS
# ------------------------------------------------------------

def show_crop_grid(
    df,
    title,
    n=12,
    random_state=42
):

    if len(df) == 0:

        print(f"\nNo images available for: {title}")
        return

    sample_n = min(n, len(df))

    sample = df.sample(
        n=sample_n,
        random_state=random_state
    )

    cols = 4
    rows = int(np.ceil(sample_n / cols))

    fig, axes = plt.subplots(
        rows,
        cols,
        figsize=(16, 4 * rows)
    )

    axes = np.array(axes).reshape(-1)

    for ax in axes:
        ax.axis("off")

    for ax, (_, row) in zip(
        axes,
        sample.iterrows()
    ):

        image_path = Path(
            row["crop_path"]
        )

        image = cv2.imread(
            str(image_path)
        )

        if image is None:
            ax.set_title("IMAGE ERROR")
            continue

        image = cv2.cvtColor(
            image,
            cv2.COLOR_BGR2RGB
        )

        ax.imshow(image)

        ax.set_title(
            f"ID: {row['crop_id']}\n"
            f"{row['source_class']}\n"
            f"{row['crop_width']}×{row['crop_height']}",
            fontsize=9
        )

        ax.axis("off")

    fig.suptitle(
        title,
        fontsize=16
    )

    plt.tight_layout()
    plt.show()


# ------------------------------------------------------------
# 3. RANDOM BATTERY CROPS
# ------------------------------------------------------------

battery_crops = battery_meta[
    battery_meta["source_class"] == "Battery"
]

show_crop_grid(
    battery_crops,
    "RANDOM BATTERY CROPS",
    n=12,
    random_state=42
)


# ------------------------------------------------------------
# 4. RANDOM POWERBANK CROPS
# ------------------------------------------------------------

powerbank_crops = battery_meta[
    battery_meta["source_class"] == "Powerbank"
]

show_crop_grid(
    powerbank_crops,
    "RANDOM POWERBANK CROPS",
    n=12,
    random_state=42
)


# ------------------------------------------------------------
# 5. VERY SMALL CROPS
# ------------------------------------------------------------

small_crops = battery_meta[
    (battery_meta["crop_width"] < 32) |
    (battery_meta["crop_height"] < 32)
]

show_crop_grid(
    small_crops,
    "VERY SMALL CROPS",
    n=12,
    random_state=10
)


# ------------------------------------------------------------
# 6. EXTREME ASPECT RATIO CROPS
# ------------------------------------------------------------

extreme_crops = battery_meta[
    (battery_meta["aspect_ratio"] > 5) |
    (battery_meta["aspect_ratio"] < 0.2)
]

show_crop_grid(
    extreme_crops,
    "EXTREME ASPECT RATIO CROPS",
    n=12,
    random_state=10
)


# ------------------------------------------------------------
# 7. VERY DARK CROPS
# ------------------------------------------------------------

dark_crops = battery_meta[
    battery_meta["mean_brightness"] < 20
]

show_crop_grid(
    dark_crops,
    "VERY DARK CROPS",
    n=12,
    random_state=10
)


# ------------------------------------------------------------
# 8. VERY BRIGHT CROPS
# ------------------------------------------------------------

bright_crops = battery_meta[
    battery_meta["mean_brightness"] > 245
]

show_crop_grid(
    bright_crops,
    "VERY BRIGHT CROPS",
    n=12,
    random_state=10
)


print("\n" + "=" * 80)
print("STEP 30 COMPLETE")
print("=" * 80)
print("Inspect the displayed image grids carefully.")
```

```python
# ============================================================
# STEP 31 — CONSERVATIVE BATTERY DATASET CLEANING
# ============================================================

import pandas as pd
from pathlib import Path
import shutil

print("=" * 80)
print("STEP 31 — CONSERVATIVE BATTERY DATASET CLEANING")
print("=" * 80)


# ------------------------------------------------------------
# 1. LOAD METADATA
# ------------------------------------------------------------

metadata_path = (
    BASE_DIR /
    "processed_battery" /
    "battery_crop_metadata_enhanced.csv"
)

battery_meta = pd.read_csv(
    metadata_path
)

print("\nOriginal crops:", len(battery_meta))


# ------------------------------------------------------------
# 2. DEFINE QUALITY RULES
# ------------------------------------------------------------

battery_meta["remove_reason"] = ""

# Very tiny crops
condition_tiny = (
    (battery_meta["crop_width"] < 24) |
    (battery_meta["crop_height"] < 24)
)

battery_meta.loc[
    condition_tiny,
    "remove_reason"
] = "too_small"


# Extreme horizontal aspect ratio
condition_wide = (
    battery_meta["aspect_ratio"] > 7
)

battery_meta.loc[
    condition_wide &
    (battery_meta["remove_reason"] == ""),
    "remove_reason"
] = "extreme_aspect_ratio"


# Extreme vertical aspect ratio
condition_tall = (
    battery_meta["aspect_ratio"] < 0.15
)

battery_meta.loc[
    condition_tall &
    (battery_meta["remove_reason"] == ""),
    "remove_reason"
] = "extreme_aspect_ratio"


# Extremely dark
condition_dark = (
    battery_meta["mean_brightness"] < 15
)

battery_meta.loc[
    condition_dark &
    (battery_meta["remove_reason"] == ""),
    "remove_reason"
] = "too_dark"


# Extremely bright
condition_bright = (
    battery_meta["mean_brightness"] > 250
)

battery_meta.loc[
    condition_bright &
    (battery_meta["remove_reason"] == ""),
    "remove_reason"
] = "too_bright"


# ------------------------------------------------------------
# 3. SHOW WHAT WILL BE REMOVED
# ------------------------------------------------------------

remove_df = battery_meta[
    battery_meta["remove_reason"] != ""
].copy()

keep_df = battery_meta[
    battery_meta["remove_reason"] == ""
].copy()

print("\n" + "-" * 70)
print("QUALITY FILTER RESULTS")
print("-" * 70)

print(
    "Crops flagged for removal:",
    len(remove_df)
)

print(
    "Crops retained:",
    len(keep_df)
)

print("\nRemoval reasons:")

print(
    remove_df[
        "remove_reason"
    ]
    .value_counts()
    .to_string()
)


# ------------------------------------------------------------
# 4. CHECK SPLIT DISTRIBUTION
# ------------------------------------------------------------

print("\n" + "-" * 70)
print("RETAINED CROPS BY SPLIT")
print("-" * 70)

print(
    keep_df[
        "dataset_split"
    ]
    .value_counts()
    .to_string()
)


print("\n" + "-" * 70)
print("REMOVED CROPS BY SPLIT")
print("-" * 70)

print(
    remove_df[
        "dataset_split"
    ]
    .value_counts()
    .to_string()
)


# ------------------------------------------------------------
# 5. CHECK CLASS DISTRIBUTION
# ------------------------------------------------------------

print("\n" + "-" * 70)
print("RETAINED SOURCE CLASS DISTRIBUTION")
print("-" * 70)

print(
    keep_df[
        "source_class"
    ]
    .value_counts()
    .to_string()
)


# ------------------------------------------------------------
# 6. CREATE CLEAN DATASET FOLDERS
# ------------------------------------------------------------

CLEAN_BASE = (
    BASE_DIR /
    "processed_battery_clean"
)

CLEAN_DIRS = {
    "train": CLEAN_BASE / "train" / "CAT_BATTERY",
    "valid": CLEAN_BASE / "valid" / "CAT_BATTERY",
    "test": CLEAN_BASE / "test" / "CAT_BATTERY"
}

for folder in CLEAN_DIRS.values():

    folder.mkdir(
        parents=True,
        exist_ok=True
    )


# ------------------------------------------------------------
# 7. COPY RETAINED CROPS
# ------------------------------------------------------------

clean_records = []

for _, row in keep_df.iterrows():

    source_path = Path(
        row["crop_path"]
    )

    split = row["dataset_split"]

    destination_dir = CLEAN_DIRS[split]

    destination_path = (
        destination_dir /
        source_path.name
    )

    shutil.copy2(
        source_path,
        destination_path
    )

    clean_records.append({
        "crop_id": row["crop_id"],
        "crop_path": str(destination_path),
        "dataset_split": split,
        "mapped_class": "CAT_BATTERY",
        "source_class": row["source_class"],
        "original_filename": row["original_filename"],
        "source_id": row["source_id"],
        "original_split": row["original_split"],
        "xmin": row["xmin"],
        "ymin": row["ymin"],
        "xmax": row["xmax"],
        "ymax": row["ymax"],
        "crop_width": row["crop_width"],
        "crop_height": row["crop_height"],
        "aspect_ratio": row["aspect_ratio"],
        "mean_brightness": row["mean_brightness"]
    })


# ------------------------------------------------------------
# 8. SAVE CLEAN METADATA
# ------------------------------------------------------------

clean_metadata = pd.DataFrame(
    clean_records
)

clean_metadata_path = (
    CLEAN_BASE /
    "battery_clean_metadata.csv"
)

clean_metadata.to_csv(
    clean_metadata_path,
    index=False
)


# ------------------------------------------------------------
# 9. SAVE REMOVED IMAGE REPORT
# ------------------------------------------------------------

removed_metadata_path = (
    CLEAN_BASE /
    "battery_removed_quality_report.csv"
)

remove_df.to_csv(
    removed_metadata_path,
    index=False
)


# ------------------------------------------------------------
# 10. FINAL SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("STEP 31 COMPLETE")
print("=" * 80)

print(
    "Original crops:",
    len(battery_meta)
)

print(
    "Retained crops:",
    len(clean_metadata)
)

print(
    "Removed crops:",
    len(remove_df)
)

print("\nClean dataset:")

print(
    CLEAN_BASE
)

print("\nClean metadata:")

print(
    clean_metadata_path
)

print("\nRemoved-image report:")

print(
    removed_metadata_path
)
```

```python
# ============================================================
# STEP 32 — AUDIT ARCHIVE DATASET
# ============================================================

from pathlib import Path
import os

print("=" * 80)
print("STEP 32 — ARCHIVE DATASET AUDIT")
print("=" * 80)

# ------------------------------------------------------------
# 1. FIND ARCHIVE DATASET
# ------------------------------------------------------------

archive_candidates = []

for p in BASE_DIR.rglob("*"):

    if p.is_dir():

        name = p.name.lower()

        if (
            "archive" in name
            or
            "train" == name
            or
            "val" == name
            or
            "test" == name
        ):
            archive_candidates.append(p)

print("\nPotential dataset directories:")

for p in archive_candidates[:50]:
    print(p)


# ------------------------------------------------------------
# 2. SEARCH FOR CLASS DIRECTORIES
# ------------------------------------------------------------

KNOWN_ARCHIVE_CLASSES = [
    "Battery",
    "Keyboard",
    "Microwave",
    "Mobile",
    "Mouse",
    "PCB",
    "Player",
    "Printer",
    "Television",
    "Washing Machine"
]

found_classes = {}

for class_name in KNOWN_ARCHIVE_CLASSES:

    matches = []

    for p in BASE_DIR.rglob("*"):

        if (
            p.is_dir()
            and
            p.name.lower() == class_name.lower()
        ):
            matches.append(p)

    found_classes[class_name] = matches


print("\n" + "-" * 70)
print("ARCHIVE CLASS DIRECTORIES")
print("-" * 70)

for class_name, matches in found_classes.items():

    if matches:

        print(f"\n{class_name}:")

        for p in matches:
            print("   ", p)

    else:

        print(f"\n{class_name}: NOT FOUND")


# ------------------------------------------------------------
# 3. COUNT IMAGES
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("IMAGE COUNTS")
print("=" * 80)

IMAGE_EXTENSIONS = {
    ".jpg",
    ".jpeg",
    ".png",
    ".bmp",
    ".webp"
}

archive_counts = []

for class_name, matches in found_classes.items():

    for class_dir in matches:

        image_files = [
            p for p in class_dir.rglob("*")
            if p.is_file()
            and p.suffix.lower() in IMAGE_EXTENSIONS
        ]

        archive_counts.append({
            "class": class_name,
            "directory": str(class_dir),
            "image_count": len(image_files)
        })

        print(
            f"{class_name:18s} "
            f"{len(image_files):5d} images"
        )


# ------------------------------------------------------------
# 4. SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("STEP 32 SUMMARY")
print("=" * 80)

print(
    "Total archive class directories found:",
    len(archive_counts)
)

print(
    "Total images found:",
    sum(
        x["image_count"]
        for x in archive_counts
    )
)
```

```python
# ============================================================
# STEP 33 — VISUAL AUDIT OF ARCHIVE DATASET
# ============================================================

import matplotlib.pyplot as plt
import cv2
import numpy as np
from pathlib import Path

print("=" * 80)
print("STEP 33 — VISUAL AUDIT OF ARCHIVE DATASET")
print("=" * 80)


# ------------------------------------------------------------
# 1. ARCHIVE ROOT
# ------------------------------------------------------------

ARCHIVE_ROOT = (
    BASE_DIR /
    "modified-dataset"
)

ARCHIVE_TRAIN = (
    ARCHIVE_ROOT /
    "train"
)


# ------------------------------------------------------------
# 2. ARCHIVE CLASSES
# ------------------------------------------------------------

ARCHIVE_CLASSES = [
    "Battery",
    "Keyboard",
    "Microwave",
    "Mobile",
    "Mouse",
    "PCB",
    "Player",
    "Printer",
    "Television",
    "Washing Machine"
]


# ------------------------------------------------------------
# 3. IMAGE EXTENSIONS
# ------------------------------------------------------------

IMAGE_EXTENSIONS = {
    ".jpg",
    ".jpeg",
    ".png",
    ".bmp",
    ".webp"
}


# ------------------------------------------------------------
# 4. FIND IMAGES
# ------------------------------------------------------------

archive_images = {}

for class_name in ARCHIVE_CLASSES:

    class_dir = (
        ARCHIVE_TRAIN /
        class_name
    )

    images = [
        p for p in class_dir.iterdir()
        if p.is_file()
        and p.suffix.lower() in IMAGE_EXTENSIONS
    ]

    archive_images[class_name] = images

    print(
        f"{class_name:18s}: "
        f"{len(images)} train images"
    )


# ------------------------------------------------------------
# 5. FUNCTION TO DISPLAY CLASS
# ------------------------------------------------------------

def show_archive_class(
    class_name,
    n=12,
    random_state=42
):

    images = archive_images[class_name]

    if len(images) == 0:

        print(
            f"No images found for {class_name}"
        )

        return

    rng = np.random.default_rng(
        random_state
    )

    n = min(
        n,
        len(images)
    )

    selected_indices = rng.choice(
        len(images),
        size=n,
        replace=False
    )

    selected = [
        images[i]
        for i in selected_indices
    ]

    cols = 4

    rows = int(
        np.ceil(n / cols)
    )

    fig, axes = plt.subplots(
        rows,
        cols,
        figsize=(16, 4 * rows)
    )

    axes = np.array(
        axes
    ).reshape(-1)

    for ax in axes:
        ax.axis("off")

    for ax, image_path in zip(
        axes,
        selected
    ):

        image = cv2.imread(
            str(image_path)
        )

        if image is None:

            ax.set_title(
                "IMAGE ERROR"
            )

            continue

        image = cv2.cvtColor(
            image,
            cv2.COLOR_BGR2RGB
        )

        ax.imshow(image)

        ax.set_title(
            image_path.name,
            fontsize=8
        )

        ax.axis("off")

    fig.suptitle(
        f"ARCHIVE — {class_name}",
        fontsize=16
    )

    plt.tight_layout()
    plt.show()


# ------------------------------------------------------------
# 6. DISPLAY ALL 10 CLASSES
# ------------------------------------------------------------

for class_name in ARCHIVE_CLASSES:

    show_archive_class(
        class_name,
        n=12,
        random_state=42
    )


print("\n" + "=" * 80)
print("STEP 33 COMPLETE")
print("=" * 80)
```

```python
# ============================================================
# STEP 34 — TELEVISION + WASHING MACHINE AUDIT
# ============================================================

import matplotlib.pyplot as plt
import cv2
import numpy as np
from pathlib import Path

print("=" * 80)
print("STEP 34 — TELEVISION + WASHING MACHINE AUDIT")
print("=" * 80)

ARCHIVE_ROOT = BASE_DIR / "modified-dataset"
ARCHIVE_TRAIN = ARCHIVE_ROOT / "train"

IMAGE_EXTENSIONS = {
    ".jpg",
    ".jpeg",
    ".png",
    ".bmp",
    ".webp"
}


# ------------------------------------------------------------
# FUNCTION TO LOAD IMAGES
# ------------------------------------------------------------

def get_class_images(class_name):

    class_dir = ARCHIVE_TRAIN / class_name

    images = [
        p for p in class_dir.iterdir()
        if p.is_file()
        and p.suffix.lower() in IMAGE_EXTENSIONS
    ]

    return images


# ------------------------------------------------------------
# FUNCTION TO DISPLAY GRID
# ------------------------------------------------------------

def display_class_grid(
    class_name,
    n=24,
    random_state=123
):

    images = get_class_images(class_name)

    print(
        f"\n{class_name}: {len(images)} images"
    )

    if len(images) == 0:
        print("❌ No images found.")
        return

    rng = np.random.default_rng(
        random_state
    )

    n = min(n, len(images))

    indices = rng.choice(
        len(images),
        size=n,
        replace=False
    )

    selected = [
        images[i]
        for i in indices
    ]

    cols = 6
    rows = int(np.ceil(n / cols))

    fig, axes = plt.subplots(
        rows,
        cols,
        figsize=(18, 3.2 * rows)
    )

    axes = np.array(
        axes
    ).reshape(-1)

    for ax in axes:
        ax.axis("off")

    for ax, image_path in zip(
        axes,
        selected
    ):

        image = cv2.imread(
            str(image_path)
        )

        if image is None:
            ax.set_title("ERROR")
            continue

        image = cv2.cvtColor(
            image,
            cv2.COLOR_BGR2RGB
        )

        ax.imshow(image)

        ax.set_title(
            image_path.name,
            fontsize=8
        )

        ax.axis("off")

    fig.suptitle(
        f"ARCHIVE — {class_name}",
        fontsize=18
    )

    plt.tight_layout()
    plt.show()


# ------------------------------------------------------------
# TELEVISION
# ------------------------------------------------------------

display_class_grid(
    "Television",
    n=24,
    random_state=123
)


# ------------------------------------------------------------
# WASHING MACHINE
# ------------------------------------------------------------

display_class_grid(
    "Washing Machine",
    n=24,
    random_state=456
)


print("\n" + "=" * 80)
print("STEP 34 COMPLETE")
print("=" * 80)
```

```python
# ============================================================
# STEP 35 — BUILD ARCHIVE DATASET METADATA
# ============================================================

import pandas as pd
from pathlib import Path

print("=" * 80)
print("STEP 35 — BUILD ARCHIVE DATASET METADATA")
print("=" * 80)


# ------------------------------------------------------------
# 1. ARCHIVE LOCATION
# ------------------------------------------------------------

ARCHIVE_ROOT = (
    BASE_DIR /
    "modified-dataset"
)

ARCHIVE_SPLITS = {
    "train": ARCHIVE_ROOT / "train",
    "valid": ARCHIVE_ROOT / "val",
    "test": ARCHIVE_ROOT / "test"
}


# ------------------------------------------------------------
# 2. CLASS MAPPING
# ------------------------------------------------------------

ARCHIVE_MAPPING = {

    "Battery": "CAT_BATTERY",

    "PCB": "CAT_PCB",

    "Keyboard": "CAT_OTHER",
    "Microwave": "CAT_OTHER",
    "Mobile": "CAT_OTHER",
    "Mouse": "CAT_OTHER",
    "Player": "CAT_OTHER",
    "Printer": "CAT_OTHER",
    "Washing Machine": "CAT_OTHER",

    # Television requires separate inspection
    "Television": "UNRESOLVED_TELEVISION"
}


# ------------------------------------------------------------
# 3. IMAGE EXTENSIONS
# ------------------------------------------------------------

IMAGE_EXTENSIONS = {
    ".jpg",
    ".jpeg",
    ".png",
    ".bmp",
    ".webp"
}


# ------------------------------------------------------------
# 4. BUILD METADATA
# ------------------------------------------------------------

archive_records = []

for split_name, split_dir in ARCHIVE_SPLITS.items():

    print(
        f"\nProcessing {split_name.upper()}..."
    )

    for source_class, mapped_class in ARCHIVE_MAPPING.items():

        class_dir = (
            split_dir /
            source_class
        )

        if not class_dir.exists():

            print(
                "⚠️ Missing:",
                class_dir
            )

            continue

        image_files = [
            p for p in class_dir.iterdir()
            if p.is_file()
            and p.suffix.lower()
            in IMAGE_EXTENSIONS
        ]

        print(
            f"{source_class:18s} "
            f"{len(image_files):3d}"
        )

        for image_path in image_files:

            archive_records.append({

                "source_dataset":
                    "archive",

                "dataset_split":
                    split_name,

                "source_class":
                    source_class,

                "mapped_class":
                    mapped_class,

                "filename":
                    image_path.name,

                "image_path":
                    str(image_path)
            })


# ------------------------------------------------------------
# 5. CREATE DATAFRAME
# ------------------------------------------------------------

archive_metadata = pd.DataFrame(
    archive_records
)

print("\n" + "=" * 80)
print("ARCHIVE METADATA SUMMARY")
print("=" * 80)

print(
    "Total images:",
    len(archive_metadata)
)


# ------------------------------------------------------------
# 6. SOURCE CLASS COUNTS
# ------------------------------------------------------------

print("\nSource class counts:")

print(
    archive_metadata[
        "source_class"
    ]
    .value_counts()
    .sort_index()
    .to_string()
)


# ------------------------------------------------------------
# 7. SPLIT COUNTS
# ------------------------------------------------------------

print("\nSplit counts:")

print(
    archive_metadata[
        "dataset_split"
    ]
    .value_counts()
    .to_string()
)


# ------------------------------------------------------------
# 8. MAPPING COUNTS
# ------------------------------------------------------------

print("\nCurrent mapping counts:")

print(
    archive_metadata[
        "mapped_class"
    ]
    .value_counts()
    .to_string()
)


# ------------------------------------------------------------
# 9. TELEVISION COUNT
# ------------------------------------------------------------

television_metadata = archive_metadata[
    archive_metadata["source_class"] ==
    "Television"
].copy()

print("\nTelevision images requiring inspection:")

print(
    len(television_metadata)
)


# ------------------------------------------------------------
# 10. SAVE
# ------------------------------------------------------------

ARCHIVE_METADATA_PATH = (
    BASE_DIR /
    "archive_metadata.csv"
)

archive_metadata.to_csv(
    ARCHIVE_METADATA_PATH,
    index=False
)

print("\nMetadata saved to:")
print(ARCHIVE_METADATA_PATH)


print("\n" + "=" * 80)
print("STEP 35 COMPLETE")
print("=" * 80)
```

```python
# ============================================================
# STEP 36 — TELEVISION MANUAL LABELING ASSISTANT
# ============================================================

import pandas as pd
import matplotlib.pyplot as plt
import cv2
import numpy as np
from pathlib import Path

print("=" * 80)
print("STEP 36 — TELEVISION MANUAL LABELING ASSISTANT")
print("=" * 80)


# ------------------------------------------------------------
# 1. LOAD ARCHIVE METADATA
# ------------------------------------------------------------

ARCHIVE_METADATA_PATH = (
    BASE_DIR /
    "archive_metadata.csv"
)

archive_metadata = pd.read_csv(
    ARCHIVE_METADATA_PATH
)


# ------------------------------------------------------------
# 2. GET TELEVISION IMAGES
# ------------------------------------------------------------

tv_metadata = archive_metadata[
    archive_metadata["source_class"] == "Television"
].copy()

tv_metadata = tv_metadata.reset_index(drop=True)

print(
    "Total television images:",
    len(tv_metadata)
)


# ------------------------------------------------------------
# 3. CREATE LABEL FILE
# ------------------------------------------------------------

TV_LABEL_PATH = (
    BASE_DIR /
    "television_manual_labels.csv"
)


if TV_LABEL_PATH.exists():

    tv_labels = pd.read_csv(
        TV_LABEL_PATH
    )

    print(
        "Existing labels loaded:",
        len(tv_labels)
    )

else:

    tv_labels = pd.DataFrame(
        columns=[
            "image_path",
            "filename",
            "dataset_split",
            "source_class",
            "manual_label",
            "mapped_class"
        ]
    )

    print(
        "No previous labels found."
    )


# ------------------------------------------------------------
# 4. DISPLAY BATCHES
# ------------------------------------------------------------

BATCH_SIZE = 30

total_batches = int(
    np.ceil(
        len(tv_metadata) /
        BATCH_SIZE
    )
)

print(
    "\nNumber of batches:",
    total_batches
)

print(
    "\nFor each image, use:"
)

print(
    "1 = CRT"
)

print(
    "2 = LCD/LED"
)

print(
    "3 = UNCLEAR"
)

print(
    "\nThe batch will be displayed with image numbers."
)


for batch_number in range(total_batches):

    start = batch_number * BATCH_SIZE

    end = min(
        start + BATCH_SIZE,
        len(tv_metadata)
    )

    batch = tv_metadata.iloc[
        start:end
    ]

    print("\n" + "=" * 80)
    print(
        f"BATCH {batch_number + 1}/{total_batches}"
    )
    print(
        f"Images {start + 1} to {end}"
    )
    print("=" * 80)


    cols = 5

    rows = int(
        np.ceil(
            len(batch) /
            cols
        )
    )

    fig, axes = plt.subplots(
        rows,
        cols,
        figsize=(18, 3.5 * rows)
    )

    axes = np.array(
        axes
    ).reshape(-1)


    for ax in axes:
        ax.axis("off")


    for position, (_, row) in enumerate(
        batch.iterrows(),
        start=1
    ):

        ax = axes[position - 1]

        image = cv2.imread(
            row["image_path"]
        )

        if image is None:

            ax.set_title(
                f"{start + position}\nIMAGE ERROR"
            )

            continue

        image = cv2.cvtColor(
            image,
            cv2.COLOR_BGR2RGB
        )

        ax.imshow(image)

        ax.set_title(
            f"{start + position}\n"
            f"{row['filename']}",
            fontsize=7
        )

        ax.axis("off")


    fig.suptitle(
        f"Television Batch {batch_number + 1}",
        fontsize=18
    )

    plt.tight_layout()
    plt.show()


    # --------------------------------------------------------
    # ASK FOR LABELS
    # --------------------------------------------------------

    print(
        f"\nEnter {len(batch)} labels in order:"
    )

    print(
        "Example:"
    )

    print(
        "1,2,2,1,3,1,2,..."
    )

    label_input = input(
        "\nLabels: "
    ).strip()


    labels = [
        x.strip()
        for x in label_input.split(",")
        if x.strip()
    ]


    if len(labels) != len(batch):

        print(
            f"⚠️ Expected {len(batch)} labels "
            f"but received {len(labels)}."
        )

        print(
            "This batch will NOT be saved."
        )

        continue


    # --------------------------------------------------------
    # CONVERT LABELS
    # --------------------------------------------------------

    batch_records = []

    valid_batch = True

    for (_, row), label in zip(
        batch.iterrows(),
        labels
    ):

        if label == "1":

            mapped_class = "CAT_CRT"

        elif label == "2":

            mapped_class = "CAT_LCD_LED"

        elif label == "3":

            mapped_class = "UNRESOLVED_TELEVISION"

        else:

            print(
                f"⚠️ Invalid label: {label}"
            )

            valid_batch = False

            break


        batch_records.append({

            "image_path":
                row["image_path"],

            "filename":
                row["filename"],

            "dataset_split":
                row["dataset_split"],

            "source_class":
                "Television",

            "manual_label":
                label,

            "mapped_class":
                mapped_class
        })


    if not valid_batch:

        print(
            "Batch rejected."
        )

        continue


    # --------------------------------------------------------
    # SAVE LABELS
    # --------------------------------------------------------

    batch_df = pd.DataFrame(
        batch_records
    )


    # Remove any previous entries
    # for these exact image paths

    if len(tv_labels) > 0:

        tv_labels = tv_labels[
            ~tv_labels["image_path"].isin(
                batch_df["image_path"]
            )
        ]


    tv_labels = pd.concat(
        [
            tv_labels,
            batch_df
        ],
        ignore_index=True
    )


    tv_labels.to_csv(
        TV_LABEL_PATH,
        index=False
    )


    print(
        f"✅ Batch {batch_number + 1} saved."
    )


# ------------------------------------------------------------
# 5. FINAL SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("STEP 36 COMPLETE")
print("=" * 80)

print(
    "Total labeled:",
    len(tv_labels)
)

print("\nLabel distribution:")

if len(tv_labels) > 0:

    print(
        tv_labels[
            "mapped_class"
        ]
        .value_counts()
        .to_string()
    )

print("\nSaved to:")

print(
    TV_LABEL_PATH
)
```

```python
# ============================================================
# STEP 37 — VALIDATE TELEVISION MANUAL LABELS
# ============================================================

import pandas as pd

print("=" * 80)
print("STEP 37 — TELEVISION LABEL VALIDATION")
print("=" * 80)

TV_LABEL_PATH = BASE_DIR / "television_manual_labels.csv"

tv_labels = pd.read_csv(TV_LABEL_PATH)

# ------------------------------------------------------------
# 1. TOTAL RECORDS
# ------------------------------------------------------------

print("\nTOTAL RECORDS")
print("-" * 60)

print("Total rows:", len(tv_labels))


# ------------------------------------------------------------
# 2. UNIQUE IMAGES
# ------------------------------------------------------------

print("\nUNIQUE IMAGES")
print("-" * 60)

print(
    "Unique image paths:",
    tv_labels["image_path"].nunique()
)


# ------------------------------------------------------------
# 3. DUPLICATE CHECK
# ------------------------------------------------------------

print("\nDUPLICATE CHECK")
print("-" * 60)

duplicate_count = (
    tv_labels["image_path"]
    .duplicated()
    .sum()
)

print(
    "Duplicate image records:",
    duplicate_count
)


# ------------------------------------------------------------
# 4. SOURCE CLASS CHECK
# ------------------------------------------------------------

print("\nSOURCE CLASS CHECK")
print("-" * 60)

print(
    tv_labels["source_class"]
    .value_counts()
    .to_string()
)


# ------------------------------------------------------------
# 5. FINAL CLASS DISTRIBUTION
# ------------------------------------------------------------

print("\nFINAL CLASS DISTRIBUTION")
print("-" * 60)

print(
    tv_labels["mapped_class"]
    .value_counts()
    .to_string()
)


# ------------------------------------------------------------
# 6. MANUAL LABEL DISTRIBUTION
# ------------------------------------------------------------

print("\nMANUAL LABEL DISTRIBUTION")
print("-" * 60)

print(
    tv_labels["manual_label"]
    .value_counts()
    .sort_index()
    .to_string()
)


# ------------------------------------------------------------
# 7. INVALID LABEL CHECK
# ------------------------------------------------------------

print("\nINVALID LABEL CHECK")
print("-" * 60)

valid_labels = {1, 2, 3}

invalid_labels = set(
    tv_labels["manual_label"].dropna().unique()
) - valid_labels

print(
    "Invalid manual labels:",
    invalid_labels
)


# ------------------------------------------------------------
# 8. EXPECTED CLASS MAPPING CHECK
# ------------------------------------------------------------

print("\nMAPPING CONSISTENCY CHECK")
print("-" * 60)

expected_mapping = {
    1: "CAT_CRT",
    2: "CAT_LCD_LED",
    3: "UNRESOLVED_TELEVISION"
}

mapping_errors = 0

for _, row in tv_labels.iterrows():

    expected = expected_mapping.get(
        row["manual_label"]
    )

    if row["mapped_class"] != expected:
        mapping_errors += 1

print(
    "Mapping errors:",
    mapping_errors
)


# ------------------------------------------------------------
# FINAL STATUS
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("STEP 37 STATUS")
print("=" * 80)

if (
    len(tv_labels) == 300
    and tv_labels["image_path"].nunique() == 300
    and duplicate_count == 0
    and len(invalid_labels) == 0
    and mapping_errors == 0
):
    print("✅ TELEVISION LABEL DATASET PASSED VALIDATION")
else:
    print("⚠️ VALIDATION FOUND ISSUES — DO NOT BUILD MASTER DATASET YET")

print("=" * 80)
```

```python
# ============================================================
# STEP 38 — BUILD CLEAN TELEVISION DATASET
# ============================================================

import pandas as pd
import shutil
from pathlib import Path

print("=" * 80)
print("STEP 38 — BUILD CLEAN TELEVISION DATASET")
print("=" * 80)

# ------------------------------------------------------------
# PATHS
# ------------------------------------------------------------

TV_LABEL_PATH = (
    BASE_DIR /
    "television_manual_labels.csv"
)

TV_OUTPUT_DIR = (
    BASE_DIR /
    "processed_television_clean"
)

# Create class folders
CRT_DIR = TV_OUTPUT_DIR / "CAT_CRT"
LCD_DIR = TV_OUTPUT_DIR / "CAT_LCD_LED"

CRT_DIR.mkdir(
    parents=True,
    exist_ok=True
)

LCD_DIR.mkdir(
    parents=True,
    exist_ok=True
)

# ------------------------------------------------------------
# LOAD LABELS
# ------------------------------------------------------------

tv_labels = pd.read_csv(
    TV_LABEL_PATH
)

# Keep only confidently labeled images
tv_clean = tv_labels[
    tv_labels["mapped_class"].isin(
        [
            "CAT_CRT",
            "CAT_LCD_LED"
        ]
    )
].copy()

print("\nConfidently labeled images:")
print(len(tv_clean))

print("\nClass distribution:")
print(
    tv_clean["mapped_class"]
    .value_counts()
    .to_string()
)

# ------------------------------------------------------------
# COPY IMAGES
# ------------------------------------------------------------

copied = 0
failed = 0

for _, row in tv_clean.iterrows():

    source_path = Path(
        row["image_path"]
    )

    if not source_path.exists():

        failed += 1

        continue

    if row["mapped_class"] == "CAT_CRT":

        destination_dir = CRT_DIR

    else:

        destination_dir = LCD_DIR

    # Prefix filename to avoid collisions
    new_filename = (
        "tv_"
        + str(row["manual_label"])
        + "_"
        + source_path.name
    )

    destination = (
        destination_dir /
        new_filename
    )

    shutil.copy2(
        source_path,
        destination
    )

    copied += 1


# ------------------------------------------------------------
# SAVE CLEAN METADATA
# ------------------------------------------------------------

clean_metadata_path = (
    TV_OUTPUT_DIR /
    "television_clean_metadata.csv"
)

tv_clean.to_csv(
    clean_metadata_path,
    index=False
)

# ------------------------------------------------------------
# SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("STEP 38 SUMMARY")
print("=" * 80)

print(
    "Confident images:",
    len(tv_clean)
)

print(
    "Images copied:",
    copied
)

print(
    "Failed copies:",
    failed
)

print("\nCAT_CRT images:")

print(
    len(list(CRT_DIR.glob("*")))
)

print("\nCAT_LCD_LED images:")

print(
    len(list(LCD_DIR.glob("*")))
)

print("\nUnresolved images excluded:")

print(
    len(tv_labels) - len(tv_clean)
)

print("\nOutput directory:")
print(TV_OUTPUT_DIR)

print("\nMetadata:")
print(clean_metadata_path)

print("\n" + "=" * 80)

if (
    copied == len(tv_clean)
    and failed == 0
):
    print("✅ CLEAN TELEVISION DATASET BUILT SUCCESSFULLY")
else:
    print("⚠️ SOME IMAGES FAILED TO COPY")

print("=" * 80)
```

```python
# ============================================================
# STEP 39 — AUDIT CURRENT ECOBRIDGE CLASS SOURCES
# ============================================================

from pathlib import Path
import pandas as pd

print("=" * 80)
print("STEP 39 — CURRENT ECOBRIDGE DATASET SOURCE AUDIT")
print("=" * 80)

# ECOBRIDGE target classes
ECOBRIDGE_CLASSES = [
    "CAT_CRT",
    "CAT_LCD_LED",
    "CAT_PCB",
    "CAT_CABLES",
    "CAT_BATTERY",
    "CAT_MOTORS",
    "CAT_PLASTICS",
    "CAT_OTHER"
]

# ------------------------------------------------------------
# SOURCE 1 — CLEAN TELEVISION DATA
# ------------------------------------------------------------

tv_dir = BASE_DIR / "processed_television_clean"

tv_counts = {}

for cls in ["CAT_CRT", "CAT_LCD_LED"]:

    class_dir = tv_dir / cls

    if class_dir.exists():

        count = len([
            p for p in class_dir.iterdir()
            if p.is_file()
        ])

    else:

        count = 0

    tv_counts[cls] = count


# ------------------------------------------------------------
# SOURCE 2 — CLEAN BATTERY DATA
# ------------------------------------------------------------

battery_dir = (
    BASE_DIR /
    "processed_battery_clean"
)

battery_counts = {}

for split in ["train", "valid", "test"]:

    split_dir = battery_dir / split

    if not split_dir.exists():
        continue

    for p in split_dir.rglob("*"):

        if p.is_file():

            battery_counts["CAT_BATTERY"] = (
                battery_counts.get(
                    "CAT_BATTERY",
                    0
                ) + 1
            )


# ------------------------------------------------------------
# SOURCE 3 — ARCHIVE DATASET
# ------------------------------------------------------------

archive_root = (
    BASE_DIR /
    "modified-dataset"
)

archive_mapping = {
    "Battery": "CAT_BATTERY",
    "PCB": "CAT_PCB",
    "Keyboard": "CAT_OTHER",
    "Microwave": "CAT_OTHER",
    "Mobile": "CAT_OTHER",
    "Mouse": "CAT_OTHER",
    "Player": "CAT_OTHER",
    "Printer": "CAT_OTHER",
    "Washing Machine": "CAT_OTHER"
}

archive_counts = {
    cls: 0
    for cls in ECOBRIDGE_CLASSES
}

for source_class, target_class in archive_mapping.items():

    for split in [
        "train",
        "val",
        "test"
    ]:

        class_dir = (
            archive_root /
            split /
            source_class
        )

        if not class_dir.exists():
            continue

        image_count = len([
            p for p in class_dir.iterdir()
            if p.is_file()
        ])

        archive_counts[target_class] += image_count


# ------------------------------------------------------------
# PRINT RESULTS
# ------------------------------------------------------------

print("\n" + "-" * 80)
print("CLEAN TELEVISION SOURCE")
print("-" * 80)

for cls, count in tv_counts.items():

    print(
        f"{cls:<20} {count:>5}"
    )


print("\n" + "-" * 80)
print("CLEAN BATTERY SOURCE")
print("-" * 80)

for cls, count in battery_counts.items():

    print(
        f"{cls:<20} {count:>5}"
    )


print("\n" + "-" * 80)
print("ARCHIVE SOURCE")
print("-" * 80)

for cls, count in archive_counts.items():

    print(
        f"{cls:<20} {count:>5}"
    )


# ------------------------------------------------------------
# COMBINED CURRENT AVAILABILITY
# ------------------------------------------------------------

combined_counts = {
    cls: 0
    for cls in ECOBRIDGE_CLASSES
}

for cls in ECOBRIDGE_CLASSES:

    combined_counts[cls] = (
        tv_counts.get(cls, 0)
        + battery_counts.get(cls, 0)
        + archive_counts.get(cls, 0)
    )


print("\n" + "=" * 80)
print("CURRENT COMBINED IMAGE AVAILABILITY")
print("=" * 80)

for cls, count in combined_counts.items():

    print(
        f"{cls:<20} {count:>5}"
    )

print("=" * 80)
```

```python
# ============================================================
# STEP 40 — SEARCH EXISTING DATA FOR MISSING CLASSES
# ============================================================

from pathlib import Path
import os

print("=" * 80)
print("STEP 40 — SEARCH EXISTING DATA FOR MISSING CLASSES")
print("=" * 80)

SEARCH_TERMS = [
    "cable",
    "cables",
    "wire",
    "wires",
    "motor",
    "motors",
    "magnet",
    "magnets",
    "plastic",
    "plastics"
]

IMAGE_EXTENSIONS = {
    ".jpg",
    ".jpeg",
    ".png",
    ".bmp",
    ".webp"
}

# Directories that are already known to contain dataset material
SEARCH_ROOTS = [
    BASE_DIR / "datasets",
    BASE_DIR / "ctsoc-ewaste-main",
]

found_files = []

for root in SEARCH_ROOTS:

    if not root.exists():
        continue

    print("\nSearching:")
    print(root)

    for path in root.rglob("*"):

        if not path.is_file():
            continue

        if path.suffix.lower() not in IMAGE_EXTENSIONS:
            continue

        name_lower = path.name.lower()
        parent_lower = str(path.parent).lower()

        matched_terms = [
            term
            for term in SEARCH_TERMS
            if term in name_lower
            or term in parent_lower
        ]

        if matched_terms:

            found_files.append({
                "path": str(path),
                "matched_terms": ", ".join(
                    matched_terms
                )
            })


# ------------------------------------------------------------
# RESULTS
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("SEARCH RESULTS")
print("=" * 80)

print(
    "Potentially relevant images found:",
    len(found_files)
)

if found_files:

    for item in found_files[:200]:

        print(
            f"{item['matched_terms']:30s} "
            f"{item['path']}"
        )

else:

    print(
        "No filenames or directory names matched "
        "the missing-class search terms."
    )

print("\n" + "=" * 80)
print("STEP 40 COMPLETE")
print("=" * 80)
```

```python
# ============================================================
# STEP 41 — AUDIT ORIGINAL MISSING-CLASS IMAGES
# ============================================================

from pathlib import Path
import pandas as pd
from PIL import Image
import hashlib

print("=" * 80)
print("STEP 41 — ORIGINAL MISSING-CLASS IMAGE AUDIT")
print("=" * 80)

# ------------------------------------------------------------
# ORIGINAL ECOBRIDGE VISION CLASS DIRECTORIES
# ------------------------------------------------------------

source_dirs = {
    "CAT_CABLES": BASE_DIR / "datasets/ai_ml/vision/train_classes/CABLES",
    "CAT_MOTORS": BASE_DIR / "datasets/ai_ml/vision/train_classes/MOTORS_MAGNETS",
    "CAT_PLASTICS": BASE_DIR / "datasets/ai_ml/vision/train_classes/MIXED_PLASTICS",
}

# ------------------------------------------------------------
# ALSO CHECK VALIDATION DIRECTORIES
# ------------------------------------------------------------

validation_dirs = {
    "CAT_CABLES": BASE_DIR / "datasets/ai_ml/vision/val_classes/CABLES",
    "CAT_MOTORS": BASE_DIR / "datasets/ai_ml/vision/val_classes/MOTORS_MAGNETS",
    "CAT_PLASTICS": BASE_DIR / "datasets/ai_ml/vision/val_classes/MIXED_PLASTICS",
}

IMAGE_EXTENSIONS = {
    ".jpg", ".jpeg", ".png", ".bmp", ".webp"
}


# ------------------------------------------------------------
# COLLECT IMAGES
# ------------------------------------------------------------

records = []

all_dirs = []

for target_class, directory in source_dirs.items():
    all_dirs.append(("TRAIN", target_class, directory))

for target_class, directory in validation_dirs.items():
    all_dirs.append(("VALID", target_class, directory))


for split, target_class, directory in all_dirs:

    if not directory.exists():
        print(f"WARNING: Missing directory -> {directory}")
        continue

    for path in sorted(directory.iterdir()):

        if not path.is_file():
            continue

        if path.suffix.lower() not in IMAGE_EXTENSIONS:
            continue

        # ----------------------------------------------------
        # Image properties
        # ----------------------------------------------------

        try:
            with Image.open(path) as img:
                width, height = img.size
                mode = img.mode

                # Pixel hash helps detect exact duplicates
                rgb_img = img.convert("RGB")
                pixel_hash = hashlib.md5(
                    rgb_img.tobytes()
                ).hexdigest()

        except Exception as e:

            width = None
            height = None
            mode = None
            pixel_hash = None

            print(
                "ERROR reading:",
                path,
                "|",
                e
            )

        records.append({
            "target_class": target_class,
            "split": split,
            "filename": path.name,
            "path": str(path),
            "width": width,
            "height": height,
            "mode": mode,
            "pixel_hash": pixel_hash
        })


df_missing = pd.DataFrame(records)


# ------------------------------------------------------------
# BASIC COUNTS
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("IMAGE COUNTS")
print("=" * 80)

if len(df_missing) > 0:

    print(
        df_missing.groupby(
            ["target_class", "split"]
        ).size()
    )

    print("\nTotal image files:")
    print(len(df_missing))

else:

    print("No images found.")


# ------------------------------------------------------------
# EXACT DUPLICATE AUDIT
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("EXACT DUPLICATE AUDIT")
print("=" * 80)

if len(df_missing) > 0:

    duplicate_hashes = (
        df_missing
        .groupby("pixel_hash")
        .size()
        .sort_values(ascending=False)
    )

    duplicate_hashes = duplicate_hashes[
        duplicate_hashes > 1
    ]

    print(
        "Unique pixel hashes:",
        df_missing["pixel_hash"].nunique()
    )

    print(
        "Duplicate pixel groups:",
        len(duplicate_hashes)
    )

    if len(duplicate_hashes) > 0:

        print("\nDuplicate groups:")

        for h in duplicate_hashes.index:

            rows = df_missing[
                df_missing["pixel_hash"] == h
            ]

            print("\n--- Duplicate group ---")

            for p in rows["path"]:
                print(p)

    else:

        print("No exact pixel duplicates found.")


# ------------------------------------------------------------
# IMAGE DIMENSIONS
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("IMAGE DIMENSION SUMMARY")
print("=" * 80)

if len(df_missing) > 0:

    print(
        df_missing[
            ["target_class", "width", "height"]
        ]
        .groupby("target_class")
        .agg(
            images=("width", "count"),
            min_width=("width", "min"),
            max_width=("width", "max"),
            min_height=("height", "min"),
            max_height=("height", "max")
        )
    )


# ------------------------------------------------------------
# SAVE AUDIT
# ------------------------------------------------------------

audit_path = BASE_DIR / "missing_classes_original_audit.csv"

df_missing.to_csv(
    audit_path,
    index=False
)

print("\nAudit saved to:")
print(audit_path)

print("\n" + "=" * 80)
print("STEP 41 COMPLETE")
print("=" * 80)
```

```python
# ============================================================
# STEP 44 — FULL LOCAL IMAGE SOURCE INVENTORY
# ============================================================

from pathlib import Path
from collections import defaultdict

print("=" * 80)
print("STEP 44 — FULL LOCAL IMAGE SOURCE INVENTORY")
print("=" * 80)

IMAGE_EXTENSIONS = {
    ".jpg", ".jpeg", ".png", ".bmp", ".webp"
}

# Directories that should NOT be counted as training datasets
EXCLUDE_KEYWORDS = [
    "results",
    "result",
    "output",
    "logs",
    "screenshots",
    "dataset-preparation"
]

# ------------------------------------------------------------
# SEARCH ALL IMAGES
# ------------------------------------------------------------

image_by_parent = defaultdict(list)

for path in BASE_DIR.rglob("*"):

    if not path.is_file():
        continue

    if path.suffix.lower() not in IMAGE_EXTENSIONS:
        continue

    path_lower = str(path).lower()

    if any(
        keyword in path_lower
        for keyword in EXCLUDE_KEYWORDS
    ):
        continue

    image_by_parent[str(path.parent)].append(path)


# ------------------------------------------------------------
# PRINT DIRECTORIES WITH IMAGES
# ------------------------------------------------------------

print("\nDirectories containing images:")
print("-" * 80)

for parent, files in sorted(
    image_by_parent.items(),
    key=lambda x: len(x[1]),
    reverse=True
):

    print(
        f"{len(files):5d} images | {parent}"
    )


# ------------------------------------------------------------
# TOTAL
# ------------------------------------------------------------

all_images = [
    p
    for files in image_by_parent.values()
    for p in files
]

print("\n" + "=" * 80)
print("TOTAL LOCAL IMAGE FILES")
print("=" * 80)

print(len(all_images))

print("\n" + "=" * 80)
print("STEP 44 COMPLETE")
print("=" * 80)
```

```python
# ============================================================
# STEP 45 — AUDIT EXISTING ECOBRIDGE MASTER DATASET
# ============================================================

from pathlib import Path
from collections import defaultdict
import hashlib
from PIL import Image
import pandas as pd

print("=" * 80)
print("STEP 45 — EXISTING MASTER DATASET AUDIT")
print("=" * 80)

MASTER_DIR = BASE_DIR / "ECOBRIDGE_MASTER_DATASET"

CLASSES = [
    "CAT_CRT",
    "CAT_LCD_LED",
    "CAT_PCB",
    "CAT_CABLES",
    "CAT_BATTERY",
    "CAT_MOTORS",
    "CAT_PLASTICS",
    "CAT_OTHER"
]

SPLITS = ["train", "val", "test"]

IMAGE_EXTENSIONS = {
    ".jpg",
    ".jpeg",
    ".png",
    ".bmp",
    ".webp"
}

# ------------------------------------------------------------
# CHECK MASTER DIRECTORY
# ------------------------------------------------------------

print("\nMaster directory:")
print(MASTER_DIR)

if not MASTER_DIR.exists():
    print("\nERROR: Master dataset does not exist.")
else:

    records = []

    for split in SPLITS:

        split_dir = MASTER_DIR / split

        if not split_dir.exists():
            print(f"\nWARNING: Missing split directory: {split_dir}")
            continue

        for target_class in CLASSES:

            class_dir = split_dir / target_class

            if not class_dir.exists():
                continue

            for path in class_dir.iterdir():

                if not path.is_file():
                    continue

                if path.suffix.lower() not in IMAGE_EXTENSIONS:
                    continue

                try:
                    with Image.open(path) as img:

                        width, height = img.size

                        # Exact pixel hash
                        pixel_hash = hashlib.md5(
                            img.convert("RGB").tobytes()
                        ).hexdigest()

                except Exception as e:

                    width = None
                    height = None
                    pixel_hash = None

                    print(
                        "ERROR:",
                        path,
                        e
                    )

                records.append({
                    "split": split,
                    "target_class": target_class,
                    "filename": path.name,
                    "path": str(path),
                    "width": width,
                    "height": height,
                    "pixel_hash": pixel_hash
                })

    master_df = pd.DataFrame(records)

    # --------------------------------------------------------
    # CLASS × SPLIT COUNTS
    # --------------------------------------------------------

    print("\n" + "=" * 80)
    print("MASTER DATASET COUNTS")
    print("=" * 80)

    if len(master_df) > 0:

        counts = pd.crosstab(
            master_df["target_class"],
            master_df["split"]
        )

        for split in SPLITS:
            if split not in counts.columns:
                counts[split] = 0

        counts = counts[SPLITS]

        counts["TOTAL"] = counts.sum(axis=1)

        print(counts)

    else:

        print("No images found.")


    # --------------------------------------------------------
    # TOTAL
    # --------------------------------------------------------

    print("\n" + "=" * 80)
    print("TOTAL MASTER IMAGES")
    print("=" * 80)

    print(len(master_df))


    # --------------------------------------------------------
    # EXACT DUPLICATE AUDIT
    # --------------------------------------------------------

    print("\n" + "=" * 80)
    print("EXACT DUPLICATE AUDIT")
    print("=" * 80)

    if len(master_df) > 0:

        unique_hashes = master_df["pixel_hash"].nunique()

        duplicate_hash_groups = (
            master_df
            .groupby("pixel_hash")
            .size()
        )

        duplicate_hash_groups = duplicate_hash_groups[
            duplicate_hash_groups > 1
        ]

        print(
            "Total image records:",
            len(master_df)
        )

        print(
            "Unique pixel hashes:",
            unique_hashes
        )

        print(
            "Duplicate pixel groups:",
            len(duplicate_hash_groups)
        )

        # ----------------------------------------------------
        # CROSS-SPLIT DUPLICATES
        # ----------------------------------------------------

        cross_split_duplicates = 0

        for pixel_hash, group in master_df.groupby(
            "pixel_hash"
        ):

            if group["split"].nunique() > 1:

                cross_split_duplicates += 1

                print(
                    "\nCROSS-SPLIT DUPLICATE:"
                )

                print(
                    group[
                        [
                            "split",
                            "target_class",
                            "filename"
                        ]
                    ].to_string(index=False)
                )

        print(
            "\nCross-split duplicate groups:",
            cross_split_duplicates
        )

    # --------------------------------------------------------
    # CLASS BALANCE
    # --------------------------------------------------------

    print("\n" + "=" * 80)
    print("CLASS BALANCE")
    print("=" * 80)

    if len(master_df) > 0:

        class_totals = (
            master_df
            .groupby("target_class")
            .size()
            .sort_values(ascending=False)
        )

        print(class_totals)

    # --------------------------------------------------------
    # IMAGE DIMENSIONS
    # --------------------------------------------------------

    print("\n" + "=" * 80)
    print("IMAGE DIMENSIONS")
    print("=" * 80)

    if len(master_df) > 0:

        dimension_summary = (
            master_df
            .groupby("target_class")
            .agg(
                images=("filename", "count"),
                min_width=("width", "min"),
                max_width=("width", "max"),
                min_height=("height", "min"),
                max_height=("height", "max")
            )
        )

        print(dimension_summary)

    # --------------------------------------------------------
    # SAVE AUDIT
    # --------------------------------------------------------

    audit_path = BASE_DIR / "master_dataset_audit.csv"

    master_df.to_csv(
        audit_path,
        index=False
    )

    print("\nAudit saved to:")
    print(audit_path)

print("\n" + "=" * 80)
print("STEP 45 COMPLETE")
print("=" * 80)
```

```python
# ============================================================
# STEP 46 — MASTER DATASET SOURCE PROVENANCE AUDIT
# ============================================================

from pathlib import Path
import pandas as pd
from collections import Counter

print("=" * 80)
print("STEP 46 — MASTER DATASET SOURCE PROVENANCE AUDIT")
print("=" * 80)

MASTER_DIR = BASE_DIR / "ECOBRIDGE_MASTER_DATASET"

CLASSES = [
    "CAT_CRT",
    "CAT_LCD_LED",
    "CAT_PCB",
    "CAT_CABLES",
    "CAT_BATTERY",
    "CAT_MOTORS",
    "CAT_PLASTICS",
    "CAT_OTHER"
]

SPLITS = ["train", "val", "test"]

IMAGE_EXTENSIONS = {
    ".jpg",
    ".jpeg",
    ".png",
    ".bmp",
    ".webp"
}


# ------------------------------------------------------------
# SOURCE IDENTIFICATION FROM FILENAME
# ------------------------------------------------------------

def identify_source(filename):

    name = filename.upper()

    if "MODIFIED_DATASET" in name:
        return "ARCHIVE_MODIFIED_DATASET"

    if "BATTERY" in name or "POWERBANK" in name:
        return "BATTERY_ROBOFLOW_OR_ARCHIVE"

    if "TELEVISION" in name or "TV_" in name:
        return "TELEVISION_SOURCE"

    if "CRT" in name:
        return "CRT_SOURCE"

    if "LCD" in name:
        return "LCD_SOURCE"

    if "PCB" in name:
        return "PCB_SOURCE"

    if "CABLE" in name:
        return "CABLE_SOURCE"

    if "MOTOR" in name or "MAGNET" in name:
        return "MOTOR_SOURCE"

    if "PLASTIC" in name:
        return "PLASTIC_SOURCE"

    return "UNKNOWN_SOURCE"


# ------------------------------------------------------------
# COLLECT MASTER FILES
# ------------------------------------------------------------

records = []

for split in SPLITS:

    split_dir = MASTER_DIR / split

    if not split_dir.exists():
        continue

    for target_class in CLASSES:

        class_dir = split_dir / target_class

        if not class_dir.exists():
            continue

        for path in class_dir.iterdir():

            if not path.is_file():
                continue

            if path.suffix.lower() not in IMAGE_EXTENSIONS:
                continue

            source = identify_source(path.name)

            records.append({
                "split": split,
                "target_class": target_class,
                "filename": path.name,
                "source_guess": source,
                "path": str(path)
            })


df_provenance = pd.DataFrame(records)


# ------------------------------------------------------------
# SOURCE × CLASS
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("SOURCE × CLASS")
print("=" * 80)

if len(df_provenance) > 0:

    source_class_table = pd.crosstab(
        df_provenance["target_class"],
        df_provenance["source_guess"]
    )

    print(source_class_table)


# ------------------------------------------------------------
# SOURCE × SPLIT
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("SOURCE × SPLIT")
print("=" * 80)

if len(df_provenance) > 0:

    source_split_table = pd.crosstab(
        df_provenance["source_guess"],
        df_provenance["split"]
    )

    print(source_split_table)


# ------------------------------------------------------------
# EXAMPLE FILENAMES FOR EACH CLASS
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("SAMPLE FILENAMES BY CLASS")
print("=" * 80)

for target_class in CLASSES:

    subset = df_provenance[
        df_provenance["target_class"] == target_class
    ]

    print("\n" + "-" * 60)
    print(target_class)

    if len(subset) == 0:
        print("No files")
        continue

    for filename in subset["filename"].head(10):
        print(" ", filename)


# ------------------------------------------------------------
# UNKNOWN SOURCES
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("UNKNOWN SOURCE FILES")
print("=" * 80)

unknown = df_provenance[
    df_provenance["source_guess"] == "UNKNOWN_SOURCE"
]

print(
    "Unknown source count:",
    len(unknown)
)

for filename in unknown["filename"].head(50):
    print(filename)


# ------------------------------------------------------------
# SAVE
# ------------------------------------------------------------

provenance_path = (
    BASE_DIR /
    "master_dataset_provenance_audit.csv"
)

df_provenance.to_csv(
    provenance_path,
    index=False
)

print("\nProvenance audit saved to:")
print(provenance_path)

print("\n" + "=" * 80)
print("STEP 46 COMPLETE")
print("=" * 80)
```

```python
# ============================================================
# STEP 47 — TRUSTED SOURCE DUPLICATE & LEAKAGE AUDIT
# ============================================================

from pathlib import Path
from PIL import Image
import hashlib
import pandas as pd

print("=" * 80)
print("STEP 47 — TRUSTED SOURCE DUPLICATE & LEAKAGE AUDIT")
print("=" * 80)

IMAGE_EXTENSIONS = {
    ".jpg", ".jpeg", ".png", ".bmp", ".webp"
}

records = []


# ============================================================
# HELPER
# ============================================================

def add_images(root, source_name, target_class, split=None):

    root = Path(root)

    if not root.exists():
        print("WARNING - missing:", root)
        return

    for path in root.rglob("*"):

        if not path.is_file():
            continue

        if path.suffix.lower() not in IMAGE_EXTENSIONS:
            continue

        try:

            with Image.open(path) as img:

                rgb = img.convert("RGB")

                pixel_hash = hashlib.md5(
                    rgb.tobytes()
                ).hexdigest()

                width, height = img.size

        except Exception as e:

            print("ERROR:", path, e)
            continue

        records.append({
            "source": source_name,
            "target_class": target_class,
            "split": split if split else "UNSPECIFIED",
            "filename": path.name,
            "path": str(path),
            "pixel_hash": pixel_hash,
            "width": width,
            "height": height
        })


# ============================================================
# 1. TELEVISION CLEAN DATA
# ============================================================

add_images(
    BASE_DIR / "processed_television_clean" / "CAT_CRT",
    "TELEVISION_MANUAL",
    "CAT_CRT"
)

add_images(
    BASE_DIR / "processed_television_clean" / "CAT_LCD_LED",
    "TELEVISION_MANUAL",
    "CAT_LCD_LED"
)


# ============================================================
# 2. CLEAN ROBoflow BATTERY CROPS
# ============================================================

for split in ["train", "valid", "test"]:

    add_images(
        BASE_DIR
        / "processed_battery_clean"
        / split
        / "CAT_BATTERY",

        "ROBOFLOW_BATTERY_CLEAN",

        "CAT_BATTERY",

        split
    )


# ============================================================
# 3. ARCHIVE DATASET
# ============================================================

archive_mapping = {

    "Battery": "CAT_BATTERY",
    "PCB": "CAT_PCB",

    "Keyboard": "CAT_OTHER",
    "Microwave": "CAT_OTHER",
    "Mobile": "CAT_OTHER",
    "Mouse": "CAT_OTHER",
    "Player": "CAT_OTHER",
    "Printer": "CAT_OTHER",
    "Washing Machine": "CAT_OTHER",

    # Television was manually labeled separately.
    # Therefore we do NOT use archive Television here.
}


archive_root = BASE_DIR / "modified-dataset"

for split in ["train", "val", "test"]:

    for source_class, target_class in archive_mapping.items():

        add_images(
            archive_root / split / source_class,

            f"ARCHIVE_{source_class}",

            target_class,

            split
        )


# ============================================================
# CREATE DATAFRAME
# ============================================================

df_sources = pd.DataFrame(records)

print("\n" + "=" * 80)
print("CANDIDATE SOURCE COUNTS")
print("=" * 80)

print(
    df_sources
    .groupby(
        ["source", "target_class", "split"]
    )
    .size()
)


# ============================================================
# TOTAL COUNTS PER TARGET CLASS
# ============================================================

print("\n" + "=" * 80)
print("TOTAL CANDIDATE IMAGES BY ECOBRIDGE CLASS")
print("=" * 80)

print(
    df_sources
    .groupby("target_class")
    .size()
    .sort_values(ascending=False)
)


# ============================================================
# EXACT DUPLICATES ACROSS ALL TRUSTED SOURCES
# ============================================================

print("\n" + "=" * 80)
print("EXACT DUPLICATE AUDIT")
print("=" * 80)

hash_counts = (
    df_sources
    .groupby("pixel_hash")
    .size()
    .sort_values(ascending=False)
)

duplicate_hashes = hash_counts[
    hash_counts > 1
]

print(
    "Total candidate image records:",
    len(df_sources)
)

print(
    "Unique pixel hashes:",
    df_sources["pixel_hash"].nunique()
)

print(
    "Duplicate pixel groups:",
    len(duplicate_hashes)
)


# ============================================================
# CROSS-SPLIT DUPLICATES
# ============================================================

print("\n" + "=" * 80)
print("CROSS-SPLIT DUPLICATE GROUPS")
print("=" * 80)

cross_split_groups = 0

for pixel_hash, group in df_sources.groupby(
    "pixel_hash"
):

    meaningful_splits = set(
        group["split"].tolist()
    )

    if (
        "train" in meaningful_splits
        and (
            "val" in meaningful_splits
            or "test" in meaningful_splits
        )
    ):

        cross_split_groups += 1

        print("\n--- CROSS-SPLIT DUPLICATE ---")

        print(
            group[
                [
                    "source",
                    "target_class",
                    "split",
                    "filename"
                ]
            ].to_string(index=False)
        )


print(
    "\nCross-split duplicate groups:",
    cross_split_groups
)


# ============================================================
# CROSS-SOURCE DUPLICATES
# ============================================================

print("\n" + "=" * 80)
print("CROSS-SOURCE DUPLICATE GROUPS")
print("=" * 80)

cross_source_groups = 0

for pixel_hash, group in df_sources.groupby(
    "pixel_hash"
):

    if group["source"].nunique() > 1:

        cross_source_groups += 1

        print("\n--- CROSS-SOURCE DUPLICATE ---")

        print(
            group[
                [
                    "source",
                    "target_class",
                    "split",
                    "filename"
                ]
            ].to_string(index=False)
        )


print(
    "\nCross-source duplicate groups:",
    cross_source_groups
)


# ============================================================
# CROSS-CLASS DUPLICATES
# ============================================================

print("\n" + "=" * 80)
print("CROSS-CLASS DUPLICATE GROUPS")
print("=" * 80)

cross_class_groups = 0

for pixel_hash, group in df_sources.groupby(
    "pixel_hash"
):

    if group["target_class"].nunique() > 1:

        cross_class_groups += 1

        print("\n--- CROSS-CLASS DUPLICATE ---")

        print(
            group[
                [
                    "source",
                    "target_class",
                    "split",
                    "filename"
                ]
            ].to_string(index=False)
        )


print(
    "\nCross-class duplicate groups:",
    cross_class_groups
)


# ============================================================
# SAVE AUDIT
# ============================================================

audit_path = (
    BASE_DIR /
    "trusted_source_duplicate_audit.csv"
)

df_sources.to_csv(
    audit_path,
    index=False
)

print("\nAudit saved to:")
print(audit_path)

print("\n" + "=" * 80)
print("STEP 47 COMPLETE")
print("=" * 80)
```

```python
# ============================================================
# STEP 48 — PERCEPTUAL DUPLICATE AUDIT
# ============================================================

from pathlib import Path
from PIL import Image
import numpy as np
import pandas as pd

print("=" * 80)
print("STEP 48 — PERCEPTUAL DUPLICATE AUDIT")
print("=" * 80)

# ------------------------------------------------------------
# Use the trusted-source audit created in Step 47
# ------------------------------------------------------------

audit_path = (
    BASE_DIR /
    "trusted_source_duplicate_audit.csv"
)

df = pd.read_csv(audit_path)

print("\nCandidate records:", len(df))


# ------------------------------------------------------------
# PERCEPTUAL HASH
# ------------------------------------------------------------

def average_hash(image_path, hash_size=16):

    try:

        img = Image.open(image_path).convert("L")

        img = img.resize(
            (hash_size, hash_size),
            Image.Resampling.LANCZOS
        )

        pixels = np.asarray(img)

        avg = pixels.mean()

        bits = pixels > avg

        return "".join(
            "1" if bit else "0"
            for bit in bits.flatten()
        )

    except Exception:

        return None


# ------------------------------------------------------------
# COMPUTE PERCEPTUAL HASHES
# ------------------------------------------------------------

print("\nComputing perceptual hashes...")

perceptual_hashes = []

for i, path in enumerate(df["path"]):

    h = average_hash(path)

    perceptual_hashes.append(h)

    if (i + 1) % 500 == 0:
        print(
            f"Processed {i + 1}/{len(df)}"
        )

df["perceptual_hash"] = perceptual_hashes


# ------------------------------------------------------------
# HAMMING DISTANCE
# ------------------------------------------------------------

def hamming_distance(hash1, hash2):

    if hash1 is None or hash2 is None:
        return 999

    return sum(
        a != b
        for a, b in zip(hash1, hash2)
    )


# ------------------------------------------------------------
# SEARCH FOR VERY SIMILAR IMAGES
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("SEARCHING FOR HIGHLY SIMILAR IMAGES")
print("=" * 80)

valid_df = df[
    df["perceptual_hash"].notna()
].reset_index(drop=True)


similar_pairs = []

# Threshold:
# 0  = identical perceptual hash
# small value = visually very similar
THRESHOLD = 4

for i in range(len(valid_df)):

    h1 = valid_df.loc[
        i,
        "perceptual_hash"
    ]

    for j in range(i + 1, len(valid_df)):

        h2 = valid_df.loc[
            j,
            "perceptual_hash"
        ]

        distance = hamming_distance(
            h1,
            h2
        )

        if distance <= THRESHOLD:

            similar_pairs.append({
                "distance": distance,

                "source_1":
                    valid_df.loc[i, "source"],

                "class_1":
                    valid_df.loc[i, "target_class"],

                "split_1":
                    valid_df.loc[i, "split"],

                "filename_1":
                    valid_df.loc[i, "filename"],

                "path_1":
                    valid_df.loc[i, "path"],

                "source_2":
                    valid_df.loc[j, "source"],

                "class_2":
                    valid_df.loc[j, "target_class"],

                "split_2":
                    valid_df.loc[j, "split"],

                "filename_2":
                    valid_df.loc[j, "filename"],

                "path_2":
                    valid_df.loc[j, "path"]
            })


similar_df = pd.DataFrame(similar_pairs)


# ------------------------------------------------------------
# RESULTS
# ------------------------------------------------------------

print(
    "\nPotentially similar image pairs:",
    len(similar_df)
)

if len(similar_df) > 0:

    print("\nClosest/similar pairs:")

    display(
        similar_df[
            [
                "distance",
                "source_1",
                "class_1",
                "split_1",
                "filename_1",
                "source_2",
                "class_2",
                "split_2",
                "filename_2"
            ]
        ]
        .sort_values("distance")
        .head(100)
    )

else:

    print(
        "No highly similar image pairs found "
        f"at threshold <= {THRESHOLD}."
    )


# ------------------------------------------------------------
# CROSS-SPLIT PERCEPTUAL DUPLICATES
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("CROSS-SPLIT PERCEPTUAL DUPLICATES")
print("=" * 80)

cross_split = []

for _, row in similar_df.iterrows():

    splits = {
        row["split_1"],
        row["split_2"]
    }

    if (
        "train" in splits
        and (
            "val" in splits
            or "test" in splits
        )
    ):

        cross_split.append(row)


cross_split_df = pd.DataFrame(
    cross_split
)

print(
    "Potential cross-split similar pairs:",
    len(cross_split_df)
)

if len(cross_split_df) > 0:

    display(
        cross_split_df[
            [
                "distance",
                "source_1",
                "class_1",
                "split_1",
                "filename_1",
                "source_2",
                "class_2",
                "split_2",
                "filename_2"
            ]
        ]
        .sort_values("distance")
        .head(100)
    )


# ------------------------------------------------------------
# SAVE
# ------------------------------------------------------------

hash_path = (
    BASE_DIR /
    "trusted_source_perceptual_hashes.csv"
)

pairs_path = (
    BASE_DIR /
    "trusted_source_perceptual_similar_pairs.csv"
)

df.to_csv(
    hash_path,
    index=False
)

similar_df.to_csv(
    pairs_path,
    index=False
)

print("\nPerceptual hashes saved to:")
print(hash_path)

print("\nSimilar-pair report saved to:")
print(pairs_path)

print("\n" + "=" * 80)
print("STEP 48 COMPLETE")
print("=" * 80)
```

```python
# ============================================================
# STEP 48B — FAST PERCEPTUAL DUPLICATE AUDIT
# ============================================================

from pathlib import Path
from PIL import Image
import numpy as np
import pandas as pd

print("=" * 80)
print("STEP 48B — FAST PERCEPTUAL DUPLICATE AUDIT")
print("=" * 80)

audit_path = (
    BASE_DIR /
    "trusted_source_duplicate_audit.csv"
)

df = pd.read_csv(audit_path)

print("\nCandidate records:", len(df))


# ------------------------------------------------------------
# CREATE SMALL IMAGE SIGNATURE
# ------------------------------------------------------------

def image_signature(path):

    try:
        img = Image.open(path).convert("RGB")

        # Very small normalized representation
        img = img.resize(
            (16, 16),
            Image.Resampling.LANCZOS
        )

        arr = np.asarray(img).astype(np.float32)

        # Normalize brightness/contrast
        mean = arr.mean()
        std = arr.std()

        if std > 0:
            arr = (arr - mean) / std

        return arr.flatten()

    except Exception:
        return None


# ------------------------------------------------------------
# COMPUTE SIGNATURES
# ------------------------------------------------------------

signatures = []

for i, path in enumerate(df["path"]):

    signatures.append(
        image_signature(path)
    )

    if (i + 1) % 500 == 0:
        print(
            f"Processed {i + 1}/{len(df)}"
        )

df["signature"] = signatures


# ------------------------------------------------------------
# REMOVE FAILED IMAGES
# ------------------------------------------------------------

valid_indices = [
    i
    for i, s in enumerate(signatures)
    if s is not None
]

print(
    "\nValid image signatures:",
    len(valid_indices)
)


# ------------------------------------------------------------
# BUILD MATRIX
# ------------------------------------------------------------

matrix = np.vstack([
    signatures[i]
    for i in valid_indices
])

valid_df = df.iloc[
    valid_indices
].reset_index(drop=True)


# ------------------------------------------------------------
# NORMALIZE EACH VECTOR
# ------------------------------------------------------------

norms = np.linalg.norm(
    matrix,
    axis=1,
    keepdims=True
)

norms[norms == 0] = 1

matrix = matrix / norms


# ------------------------------------------------------------
# BLOCK-WISE SIMILARITY SEARCH
# ------------------------------------------------------------

print("\nSearching for highly similar images...")

THRESHOLD = 0.995

similar_pairs = []

block_size = 500

for start in range(
    0,
    len(matrix),
    block_size
):

    end = min(
        start + block_size,
        len(matrix)
    )

    block = matrix[start:end]

    similarity = block @ matrix.T

    for local_i in range(
        similarity.shape[0]
    ):

        global_i = start + local_i

        # Only compare with later images
        for j in range(
            global_i + 1,
            len(matrix)
        ):

            score = similarity[
                local_i,
                j
            ]

            if score >= THRESHOLD:

                similar_pairs.append({

                    "similarity": float(score),

                    "source_1":
                        valid_df.loc[
                            global_i,
                            "source"
                        ],

                    "class_1":
                        valid_df.loc[
                            global_i,
                            "target_class"
                        ],

                    "split_1":
                        valid_df.loc[
                            global_i,
                            "split"
                        ],

                    "filename_1":
                        valid_df.loc[
                            global_i,
                            "filename"
                        ],

                    "path_1":
                        valid_df.loc[
                            global_i,
                            "path"
                        ],

                    "source_2":
                        valid_df.loc[
                            j,
                            "source"
                        ],

                    "class_2":
                        valid_df.loc[
                            j,
                            "target_class"
                        ],

                    "split_2":
                        valid_df.loc[
                            j,
                            "split"
                        ],

                    "filename_2":
                        valid_df.loc[
                            j,
                            "filename"
                        ],

                    "path_2":
                        valid_df.loc[
                            j,
                            "path"
                        ]
                })

    print(
        f"Compared {end}/{len(matrix)} images"
    )


# ------------------------------------------------------------
# RESULTS
# ------------------------------------------------------------

similar_df = pd.DataFrame(
    similar_pairs
)

print("\n" + "=" * 80)
print("PERCEPTUAL SIMILARITY RESULTS")
print("=" * 80)

print(
    "Potentially similar pairs:",
    len(similar_df)
)

if len(similar_df) > 0:

    display(
        similar_df
        .sort_values(
            "similarity",
            ascending=False
        )
        .head(100)
    )

else:

    print(
        "No highly similar pairs found."
    )


# ------------------------------------------------------------
# CROSS-SPLIT SIMILARITY
# ------------------------------------------------------------

if len(similar_df) > 0:

    cross_split = similar_df[
        (
            (
                similar_df["split_1"] == "train"
            )
            &
            (
                similar_df["split_2"].isin(
                    ["val", "test"]
                )
            )
        )
        |
        (
            (
                similar_df["split_2"] == "train"
            )
            &
            (
                similar_df["split_1"].isin(
                    ["val", "test"]
                )
            )
        )
    ]

else:

    cross_split = pd.DataFrame()


print("\n" + "=" * 80)
print("CROSS-SPLIT PERCEPTUAL SIMILARITY")
print("=" * 80)

print(
    "Potential cross-split pairs:",
    len(cross_split)
)

if len(cross_split) > 0:

    display(
        cross_split
        .sort_values(
            "similarity",
            ascending=False
        )
        .head(100)
    )


# ------------------------------------------------------------
# SAVE
# ------------------------------------------------------------

output_path = (
    BASE_DIR /
    "trusted_source_perceptual_similar_pairs_fast.csv"
)

similar_df.to_csv(
    output_path,
    index=False
)

print("\nSaved to:")
print(output_path)

print("\n" + "=" * 80)
print("STEP 48B COMPLETE")
print("=" * 80)
```

```python
# ============================================================
# STEP 49 — BUILD PERCEPTUAL SIMILARITY GROUPS
# ============================================================

from pathlib import Path
import pandas as pd

print("=" * 80)
print("STEP 49 — BUILD PERCEPTUAL SIMILARITY GROUPS")
print("=" * 80)

PAIR_FILE = (
    BASE_DIR /
    "trusted_source_perceptual_similar_pairs_fast.csv"
)

df_pairs = pd.read_csv(PAIR_FILE)

print(
    "\nSimilarity pairs found:",
    len(df_pairs)
)


# ============================================================
# UNION-FIND
# ============================================================

parent = {}


def find(x):

    if parent[x] != x:
        parent[x] = find(parent[x])

    return parent[x]


def union(a, b):

    if a not in parent:
        parent[a] = a

    if b not in parent:
        parent[b] = b

    root_a = find(a)
    root_b = find(b)

    if root_a != root_b:
        parent[root_b] = root_a


# ============================================================
# CREATE UNIQUE IMAGE IDs
# ============================================================

def make_id(source, filename):

    return f"{source}::{filename}"


# ============================================================
# ADD ALL PAIRS
# ============================================================

for _, row in df_pairs.iterrows():

    id1 = make_id(
        row["source_1"],
        row["filename_1"]
    )

    id2 = make_id(
        row["source_2"],
        row["filename_2"]
    )

    union(id1, id2)


# ============================================================
# BUILD GROUPS
# ============================================================

groups = {}

for item in parent:

    root = find(item)

    groups.setdefault(
        root,
        []
    ).append(item)


# Only groups containing 2+ images
similar_groups = [
    members
    for members in groups.values()
    if len(members) > 1
]


print(
    "\nSimilarity groups:",
    len(similar_groups)
)


# ============================================================
# DISPLAY GROUPS
# ============================================================

for i, members in enumerate(
    similar_groups,
    start=1
):

    print(
        f"\nGROUP {i} "
        f"({len(members)} images)"
    )

    for member in members:
        print(
            " ",
            member
        )


# ============================================================
# CREATE GROUP TABLE
# ============================================================

group_records = []

for group_number, members in enumerate(
    similar_groups,
    start=1
):

    for member in members:

        source, filename = member.split(
            "::",
            1
        )

        group_records.append({
            "similarity_group":
                f"SIM_GROUP_{group_number:04d}",

            "source":
                source,

            "filename":
                filename
        })


groups_df = pd.DataFrame(
    group_records
)


# ============================================================
# SAVE
# ============================================================

output_path = (
    BASE_DIR /
    "trusted_source_similarity_groups.csv"
)

groups_df.to_csv(
    output_path,
    index=False
)

print("\n" + "=" * 80)
print("SIMILARITY GROUP SUMMARY")
print("=" * 80)

if len(groups_df) > 0:

    print(
        groups_df
        .groupby("similarity_group")
        .size()
        .sort_values(ascending=False)
    )

print("\nSaved to:")
print(output_path)

print("\n" + "=" * 80)
print("STEP 49 COMPLETE")
print("=" * 80)
```

```python
# ============================================================
# STEP 50 — BUILD VERIFIED 5-CLASS CANDIDATE DATASET
# ============================================================

from pathlib import Path
import shutil
import pandas as pd

print("=" * 80)
print("STEP 50 — BUILD VERIFIED 5-CLASS CANDIDATE DATASET")
print("=" * 80)

# ------------------------------------------------------------
# TARGET CLASSES
# ------------------------------------------------------------

VERIFIED_CLASSES = [
    "CAT_CRT",
    "CAT_LCD_LED",
    "CAT_PCB",
    "CAT_BATTERY",
    "CAT_OTHER"
]

# ------------------------------------------------------------
# SOURCE DIRECTORIES
# ------------------------------------------------------------

TV_ROOT = (
    BASE_DIR /
    "processed_television_clean"
)

BATTERY_ROOT = (
    BASE_DIR /
    "processed_battery_clean"
)

ARCHIVE_ROOT = (
    BASE_DIR /
    "modified-dataset"
)

# ------------------------------------------------------------
# OUTPUT
# ------------------------------------------------------------

CANDIDATE_ROOT = (
    BASE_DIR /
    "ECOBRIDGE_VERIFIED_5CLASS"
)

if CANDIDATE_ROOT.exists():

    shutil.rmtree(
        CANDIDATE_ROOT
    )

CANDIDATE_ROOT.mkdir(
    parents=True,
    exist_ok=True
)

for split in [
    "all"
]:

    for cls in VERIFIED_CLASSES:

        (
            CANDIDATE_ROOT /
            split /
            cls
        ).mkdir(
            parents=True,
            exist_ok=True
        )


# ------------------------------------------------------------
# RECORDS
# ------------------------------------------------------------

records = []


def copy_image(
    source_path,
    target_class,
    source_name
):

    source_path = Path(source_path)

    destination_dir = (
        CANDIDATE_ROOT /
        "all" /
        target_class
    )

    destination = (
        destination_dir /
        source_path.name
    )

    # Prevent filename collisions
    if destination.exists():

        stem = source_path.stem
        suffix = source_path.suffix

        counter = 1

        while destination.exists():

            destination = (
                destination_dir /
                f"{stem}_{counter}{suffix}"
            )

            counter += 1

    shutil.copy2(
        source_path,
        destination
    )

    records.append({

        "target_class":
            target_class,

        "source":
            source_name,

        "original_path":
            str(source_path),

        "candidate_path":
            str(destination)
    })


# ============================================================
# 1. CRT
# ============================================================

crt_dir = (
    TV_ROOT /
    "CAT_CRT"
)

for path in sorted(
    crt_dir.iterdir()
):

    if path.is_file():

        copy_image(
            path,
            "CAT_CRT",
            "TELEVISION_MANUAL"
        )


# ============================================================
# 2. LCD / LED
# ============================================================

lcd_dir = (
    TV_ROOT /
    "CAT_LCD_LED"
)

for path in sorted(
    lcd_dir.iterdir()
):

    if path.is_file():

        copy_image(
            path,
            "CAT_LCD_LED",
            "TELEVISION_MANUAL"
        )


# ============================================================
# 3. CLEAN BATTERY
# ============================================================

battery_count = 0

for split in [
    "train",
    "valid",
    "test"
]:

    source_dir = (
        BATTERY_ROOT /
        split /
        "CAT_BATTERY"
    )

    if not source_dir.exists():
        continue

    for path in sorted(
        source_dir.iterdir()
    ):

        if path.is_file():

            copy_image(
                path,
                "CAT_BATTERY",
                "ROBOFLOW_BATTERY_CLEAN"
            )

            battery_count += 1


# ============================================================
# 4. ARCHIVE BATTERY
# ============================================================

for split in [
    "train",
    "val",
    "test"
]:

    source_dir = (
        ARCHIVE_ROOT /
        split /
        "Battery"
    )

    if not source_dir.exists():
        continue

    for path in sorted(
        source_dir.iterdir()
    ):

        if path.is_file():

            copy_image(
                path,
                "CAT_BATTERY",
                "ARCHIVE_Battery"
            )


# ============================================================
# 5. ARCHIVE PCB
# ============================================================

for split in [
    "train",
    "val",
    "test"
]:

    source_dir = (
        ARCHIVE_ROOT /
        split /
        "PCB"
    )

    if not source_dir.exists():
        continue

    for path in sorted(
        source_dir.iterdir()
    ):

        if path.is_file():

            copy_image(
                path,
                "CAT_PCB",
                "ARCHIVE_PCB"
            )


# ============================================================
# 6. ARCHIVE OTHER
# ============================================================

OTHER_SOURCE_CLASSES = [
    "Keyboard",
    "Microwave",
    "Mobile",
    "Mouse",
    "Player",
    "Printer",
    "Washing Machine"
]

for source_class in OTHER_SOURCE_CLASSES:

    for split in [
        "train",
        "val",
        "test"
    ]:

        source_dir = (
            ARCHIVE_ROOT /
            split /
            source_class
        )

        if not source_dir.exists():
            continue

        for path in sorted(
            source_dir.iterdir()
        ):

            if path.is_file():

                copy_image(
                    path,
                    "CAT_OTHER",
                    f"ARCHIVE_{source_class}"
                )


# ============================================================
# SAVE METADATA
# ============================================================

metadata_path = (
    CANDIDATE_ROOT /
    "candidate_metadata.csv"
)

metadata_df = pd.DataFrame(
    records
)

metadata_df.to_csv(
    metadata_path,
    index=False
)


# ============================================================
# SUMMARY
# ============================================================

print("\n" + "=" * 80)
print("VERIFIED CANDIDATE COUNTS")
print("=" * 80)

print(
    metadata_df
    .groupby(
        ["target_class", "source"]
    )
    .size()
)

print("\n" + "=" * 80)
print("TOTALS BY CLASS")
print("=" * 80)

print(
    metadata_df
    .groupby("target_class")
    .size()
    .sort_values(
        ascending=False
    )
)

print("\nTotal candidate images:")
print(len(metadata_df))

print("\nDataset created at:")
print(CANDIDATE_ROOT)

print("\nMetadata:")
print(metadata_path)

print("\n" + "=" * 80)
print("STEP 50 COMPLETE")
print("=" * 80)
```

```python
# ============================================================
# STEP 55 — INSPECT ELECTRICAL CABLE SOURCE
# ============================================================

from pathlib import Path
from PIL import Image
import pandas as pd

cable_source = (
    EXTERNAL_DIR
    / "trashbox_repo"
    / "TrashBox_train_dataset_subfolders"
    / "e-waste"
    / "electrical cables"
)

image_exts = {".jpg", ".jpeg", ".png", ".bmp", ".webp"}

cable_images = sorted([
    p for p in cable_source.iterdir()
    if p.is_file() and p.suffix.lower() in image_exts
])

print("=" * 70)
print("STEP 55 — ELECTRICAL CABLE SOURCE INSPECTION")
print("=" * 70)

print(f"Source folder : {cable_source}")
print(f"Images found  : {len(cable_images)}")

print("\nFirst 20 filenames:")
for p in cable_images[:20]:
    print(" ", p.name)

# ------------------------------------------------------------
# Basic image validity check
# ------------------------------------------------------------

valid = 0
invalid = 0
sizes = []

for p in cable_images:
    try:
        with Image.open(p) as img:
            img.verify()

        with Image.open(p) as img:
            sizes.append(img.size)

        valid += 1

    except Exception:
        invalid += 1

print("\nImage validity:")
print(f"Valid   : {valid}")
print(f"Invalid : {invalid}")

if sizes:
    width_counts = pd.Series([s[0] for s in sizes]).value_counts().head(10)
    height_counts = pd.Series([s[1] for s in sizes]).value_counts().head(10)

    print("\nMost common widths:")
    print(width_counts.to_string())

    print("\nMost common heights:")
    print(height_counts.to_string())

print("=" * 70)
print("✅ Inspection complete.")
print("Next: visual audit before accepting these images into CAT_CABLES.")
```

```python
# ============================================================
# STEP 63 — EXPAND WIKIMEDIA ELECTRIC MOTOR CANDIDATES
# ============================================================

import requests
import pandas as pd

WIKI_API = "https://commons.wikimedia.org/w/api.php"

headers = {
    "User-Agent": "ECOBRIDGE-EWaste-Research/1.0 (dataset inspection)"
}

all_motor_records = []
continue_token = None

MAX_FILES = 300

while len(all_motor_records) < MAX_FILES:

    params = {
        "action": "query",
        "format": "json",
        "generator": "categorymembers",
        "gcmtitle": "Category:Electric motors",
        "gcmtype": "file",
        "gcmlimit": "100",
        "prop": "imageinfo",
        "iiprop": "url|size|mime",
        "iiurlwidth": "600"
    }

    if continue_token:
        params.update(continue_token)

    response = requests.get(
        WIKI_API,
        params=params,
        headers=headers,
        timeout=30
    )

    response.raise_for_status()

    data = response.json()

    pages = data.get("query", {}).get("pages", {})

    if not pages:
        break

    for page in pages.values():

        imageinfo = page.get("imageinfo", [{}])[0]

        mime = imageinfo.get("mime", "")

        if mime in ["image/jpeg", "image/png"]:

            all_motor_records.append({
                "title": page.get("title", ""),
                "url": imageinfo.get("url"),
                "thumb_url": imageinfo.get("thumburl"),
                "width": imageinfo.get("width"),
                "height": imageinfo.get("height"),
                "mime": mime
            })

            if len(all_motor_records) >= MAX_FILES:
                break

    continue_token = data.get("continue")

    if not continue_token:
        break

# ------------------------------------------------------------
# Remove duplicate URLs
# ------------------------------------------------------------

motor_expanded_df = (
    pd.DataFrame(all_motor_records)
    .drop_duplicates(subset=["url"])
    .reset_index(drop=True)
)

print("=" * 70)
print("STEP 63 — EXPANDED WIKIMEDIA MOTOR CANDIDATES")
print("=" * 70)

print(f"Unique image candidates: {len(motor_expanded_df)}")

print("\nImage formats:")
print(
    motor_expanded_df["mime"]
    .value_counts()
    .to_string()
)

print("\nResolution summary:")
print(
    motor_expanded_df[
        ["width", "height"]
    ].describe().round(1).to_string()
)

print("\nFirst 40 candidate filenames:")
for i, title in enumerate(
    motor_expanded_df["title"].head(40)
):
    print(f"[{i:03d}] {title}")

print("=" * 70)
print("⚠️ Candidates only — nothing has been added to ECOBRIDGE.")
print("⚠️ We will visually curate the source before acceptance.")
print("=" * 70)
```

```python
# ============================================================
# STEP 65 — CONSERVATIVE MOTOR CANDIDATE SHORTLIST
# ============================================================

import re
import pandas as pd

# Words that strongly indicate an actual motor or motor component.
# This is only a PRE-FILTER. It does NOT mean the image is accepted.
motor_keywords = [
    "motor",
    "motors",
    "electric motor",
    "electromotor",
    "moteur",
    "antrieb",
    "motoren",
    "stator",
    "rotor",
    "stepper",
    "hubmotor",
    "capstan motor",
    "motor assembly",
    "motor component",
    "motor part"
]

# Words that commonly indicate whole machines / unrelated material.
exclude_keywords = [
    "diagram",
    "voltage",
    "poster",
    "banner",
    "logo",
    "bridge",
    "tram test stand",
    "car",
    "e-up",
    "amg",
    "railroad journal",
    "advertisement",
    "oil",
    "generator set",
    "generator",
    "convertor",
    "converter"
]

def normalize(text):
    return re.sub(r"[^a-z0-9 ]+", " ", text.lower())

candidate_rows = []

for idx, row in motor_expanded_df.iterrows():

    title = normalize(row["title"])

    has_motor_word = any(
        keyword in title
        for keyword in motor_keywords
    )

    has_exclusion = any(
        keyword in title
        for keyword in exclude_keywords
    )

    if has_motor_word and not has_exclusion:
        candidate_rows.append(idx)

motor_shortlist_df = (
    motor_expanded_df
    .iloc[candidate_rows]
    .copy()
    .reset_index()
)

print("=" * 70)
print("STEP 65 — CONSERVATIVE MOTOR CANDIDATE SHORTLIST")
print("=" * 70)

print(f"Original Wikimedia candidates : {len(motor_expanded_df)}")
print(f"Shortlisted candidates        : {len(motor_shortlist_df)}")

print("\nFirst 100 shortlisted candidates:")

for i, row in motor_shortlist_df.head(100).iterrows():
    print(
        f"[{i:03d}] "
        f"OriginalIndex={row['index']:03d} | "
        f"{row['title']}"
    )

print("=" * 70)
print("IMPORTANT")
print("=" * 70)
print("This shortlist is NOT the final CAT_MOTORS dataset.")
print("It is only a title-based pre-filter for visual inspection.")
print("Every accepted image will still require visual verification.")
print("=" * 70)
```

```python
# ============================================================
# STEP 67 — TARGETED WIKIMEDIA MOTOR SUBCATEGORIES
# ============================================================

import requests
import pandas as pd

WIKI_API = "https://commons.wikimedia.org/w/api.php"

headers = {
    "User-Agent": "ECOBRIDGE-EWaste-Research/1.0 (dataset inspection)"
}

# These are targeted motor categories rather than the broad
# "Electric motors" category.
target_categories = [
    "Category:Hard disk motors",
    "Category:Optical disc drive motors",
    "Category:Wheel hub motors",
    "Category:Permanent magnet DC electric motors",
    "Category:Electric motor parts and components"
]

target_records = []

for category in target_categories:

    continue_token = None

    while True:

        params = {
            "action": "query",
            "format": "json",
            "generator": "categorymembers",
            "gcmtitle": category,
            "gcmtype": "file",
            "gcmlimit": "100",
            "prop": "imageinfo",
            "iiprop": "url|size|mime",
            "iiurlwidth": "600"
        }

        if continue_token:
            params.update(continue_token)

        response = requests.get(
            WIKI_API,
            params=params,
            headers=headers,
            timeout=30
        )

        response.raise_for_status()

        data = response.json()

        pages = data.get(
            "query",
            {}
        ).get(
            "pages",
            {}
        )

        if not pages:
            break

        for page in pages.values():

            imageinfo = page.get(
                "imageinfo",
                [{}]
            )[0]

            mime = imageinfo.get(
                "mime",
                ""
            )

            if mime in [
                "image/jpeg",
                "image/png"
            ]:

                target_records.append({
                    "source_category": category,
                    "title": page.get(
                        "title",
                        ""
                    ),
                    "url": imageinfo.get(
                        "url"
                    ),
                    "thumb_url": imageinfo.get(
                        "thumburl"
                    ),
                    "width": imageinfo.get(
                        "width"
                    ),
                    "height": imageinfo.get(
                        "height"
                    ),
                    "mime": mime
                })

        continue_token = data.get(
            "continue"
        )

        if not continue_token:
            break

# ------------------------------------------------------------
# Remove duplicate files
# ------------------------------------------------------------

target_motor_df = (
    pd.DataFrame(target_records)
    .drop_duplicates(
        subset=["url"]
    )
    .reset_index(drop=True)
)

print("=" * 70)
print("STEP 67 — TARGETED MOTOR SOURCES")
print("=" * 70)

print(
    f"Unique candidate images: "
    f"{len(target_motor_df)}"
)

print("\nCandidates by source category:")

print(
    target_motor_df[
        "source_category"
    ]
    .value_counts()
    .to_string()
)

print("\nFirst 100 candidates:")

for i, row in target_motor_df.head(100).iterrows():

    print(
        f"[{i:03d}] "
        f"{row['source_category']} | "
        f"{row['title']}"
    )

print("=" * 70)
print("⚠️ These are candidate sources only.")
print("⚠️ No images have been added to ECOBRIDGE.")
print("=" * 70)
```

```python
# =============================================================================
# STEP 68 — DOWNLOAD REPRESENTATIVE TARGETED MOTOR CANDIDATES
# =============================================================================

import os
import requests
from PIL import Image
from io import BytesIO
import matplotlib.pyplot as plt
import math

OUT_DIR = "/content/ECOBRIDGE_DATA/external_sources/motor_audit_step68"
os.makedirs(OUT_DIR, exist_ok=True)

# -------------------------------------------------------------------------
# Use the candidate list created in STEP 67
# -------------------------------------------------------------------------
if "motor_candidates" not in globals():
    raise NameError(
        "motor_candidates was not found. Please run STEP 67 first."
    )

print("Total targeted motor candidates:", len(motor_candidates))

# -------------------------------------------------------------------------
# Select 48 evenly spaced candidates for visual audit
# -------------------------------------------------------------------------
n_audit = min(48, len(motor_candidates))

if len(motor_candidates) <= n_audit:
    selected = motor_candidates
else:
    indices = [
        round(i * (len(motor_candidates) - 1) / (n_audit - 1))
        for i in range(n_audit)
    ]
    selected = [motor_candidates[i] for i in indices]

print("Selected for visual audit:", len(selected))

# -------------------------------------------------------------------------
# Download images
# -------------------------------------------------------------------------
downloaded = []

headers = {
    "User-Agent": "ECOBRIDGE-EWaste-Research/1.0 (dataset inspection)"
}

for i, item in enumerate(selected):

    url = item.get("url") or item.get("image_url")

    if not url:
        print(f"Skipping {i}: no URL")
        continue

    try:
        response = requests.get(
            url,
            headers=headers,
            timeout=20
        )
        response.raise_for_status()

        img = Image.open(BytesIO(response.content)).convert("RGB")

        filename = f"motor_audit_{i:03d}.jpg"
        save_path = os.path.join(OUT_DIR, filename)

        img.save(save_path, quality=95)

        downloaded.append({
            "audit_index": i,
            "source_category": item.get("source_category", ""),
            "title": item.get("title", item.get("filename", "")),
            "url": url,
            "local_path": save_path
        })

    except Exception as e:
        print(f"Failed {i}: {str(e)[:100]}")

print("\nDownloaded successfully:", len(downloaded))

# -------------------------------------------------------------------------
# Display contact sheet
# -------------------------------------------------------------------------
cols = 6
rows = math.ceil(len(downloaded) / cols)

plt.figure(figsize=(18, rows * 3))

for plot_i, item in enumerate(downloaded):

    img = Image.open(item["local_path"])

    ax = plt.subplot(rows, cols, plot_i + 1)
    ax.imshow(img)
    ax.axis("off")

    title = item["title"]
    if len(title) > 32:
        title = title[:29] + "..."

    ax.set_title(
        f"[{item['audit_index']:02d}] {title}",
        fontsize=7
    )

plt.tight_layout()
plt.show()

# -------------------------------------------------------------------------
# Print exact audit list
# -------------------------------------------------------------------------
print("\n" + "=" * 80)
print("STEP 68 AUDIT LIST")
print("=" * 80)

for item in downloaded:
    print(
        f"[{item['audit_index']:02d}] "
        f"{item['source_category']} | "
        f"{item['title']}"
    )

print("\nAudit folder:")
print(OUT_DIR)
```

```python
# =============================================================================
# STEP 68C — DOWNLOAD MOTOR THUMBNAILS WITH RATE-LIMIT PROTECTION
# =============================================================================

import os
import time
import requests
from PIL import Image
from io import BytesIO
import matplotlib.pyplot as plt
import math

OUT_DIR = "/content/ECOBRIDGE_DATA/external_sources/motor_audit_step68"
os.makedirs(OUT_DIR, exist_ok=True)

print("Total motor candidates:", len(motor_candidates))

downloaded = []

headers = {
    "User-Agent": (
        "Mozilla/5.0 (X11; Linux x86_64) "
        "AppleWebKit/537.36 Chrome/120 Safari/537.36 "
        "ECOBRIDGE research"
    ),
    "Accept": "image/avif,image/webp,image/apng,image/svg+xml,image/*,*/*;q=0.8"
}

session = requests.Session()
session.headers.update(headers)

# -------------------------------------------------------------------------
# Download thumbnail instead of original image
# -------------------------------------------------------------------------
for i, row in motor_candidates.iterrows():

    # Prefer thumbnail URL
    url = row["thumb_url"]

    title = row["title"]

    try:

        # Delay between requests to reduce Wikimedia load
        time.sleep(2)

        response = session.get(
            url,
            timeout=30
        )

        if response.status_code == 429:
            print(f"Rate limited at [{i:02d}]. Stopping safely.")
            break

        response.raise_for_status()

        img = Image.open(
            BytesIO(response.content)
        ).convert("RGB")

        filename = f"motor_audit_{i:03d}.jpg"
        save_path = os.path.join(
            OUT_DIR,
            filename
        )

        img.save(
            save_path,
            quality=90
        )

        downloaded.append({
            "audit_index": int(i),
            "source_index": int(row["index"]),
            "title": title,
            "url": url,
            "local_path": save_path
        })

        print(f"Downloaded [{i:02d}]")

    except Exception as e:

        print(
            f"Failed [{i:02d}] "
            f"{str(e)[:120]}"
        )

print("\n" + "=" * 80)
print("DOWNLOAD SUMMARY")
print("=" * 80)

print("Successfully downloaded:", len(downloaded))

# -------------------------------------------------------------------------
# Create contact sheet
# -------------------------------------------------------------------------
if downloaded:

    cols = 5
    rows = math.ceil(len(downloaded) / cols)

    plt.figure(
        figsize=(18, rows * 3.5)
    )

    for plot_i, item in enumerate(downloaded):

        img = Image.open(
            item["local_path"]
        )

        ax = plt.subplot(
            rows,
            cols,
            plot_i + 1
        )

        ax.imshow(img)
        ax.axis("off")

        title = item["title"]

        if len(title) > 30:
            title = title[:27] + "..."

        ax.set_title(
            f"[{item['audit_index']:02d}] {title}",
            fontsize=7
        )

    plt.tight_layout()
    plt.show()

else:
    print("\nNo images downloaded.")
    print("Wikimedia is still rate-limiting the requests.")
```

```python
# =============================================================================
# STEP 71 — VISUAL AUDIT OF TRASHBOX PLASTIC IMAGES
# =============================================================================

import os
import glob
import random
from PIL import Image
import matplotlib.pyplot as plt
import math

PLASTIC_DIR = (
    "/content/ECOBRIDGE_DATA/"
    "external_sources/trashbox_repo/"
    "TrashBox_train_set/plastic"
)

# -------------------------------------------------------------------------
# Find all plastic images
# -------------------------------------------------------------------------

plastic_files = []

for ext in ["*.jpg", "*.jpeg", "*.png", "*.webp"]:
    plastic_files.extend(
        glob.glob(
            os.path.join(PLASTIC_DIR, ext)
        )
    )

plastic_files = sorted(set(plastic_files))

print("=" * 80)
print("STEP 71 — TRASHBOX PLASTIC SOURCE AUDIT")
print("=" * 80)

print("Plastic images found:", len(plastic_files))

# -------------------------------------------------------------------------
# Select 48 random images
# -------------------------------------------------------------------------

random.seed(42)

audit_count = min(48, len(plastic_files))

audit_files = random.sample(
    plastic_files,
    audit_count
)

print("Images selected for audit:", len(audit_files))

# -------------------------------------------------------------------------
# Display contact sheet
# -------------------------------------------------------------------------

cols = 6
rows = math.ceil(len(audit_files) / cols)

plt.figure(
    figsize=(18, rows * 3)
)

for i, path in enumerate(audit_files):

    try:
        img = Image.open(path).convert("RGB")

        ax = plt.subplot(
            rows,
            cols,
            i + 1
        )

        ax.imshow(img)
        ax.axis("off")

        filename = os.path.basename(path)

        if len(filename) > 22:
            filename = filename[:19] + "..."

        ax.set_title(
            f"[{i:02d}] {filename}",
            fontsize=7
        )

    except Exception as e:

        print(
            f"Could not open {path}: {e}"
        )

plt.tight_layout()
plt.show()

# -------------------------------------------------------------------------
# Print selected file paths
# -------------------------------------------------------------------------

print("\n" + "=" * 80)
print("AUDIT FILES")
print("=" * 80)

for i, path in enumerate(audit_files):

    print(
        f"[{i:02d}] {path}"
    )
```

```python
# =============================================================================
# STEP 72 — CHECK FOR WEEE PLASTIC DATASET LOCALLY
# =============================================================================

import os

search_roots = [
    "/content",
    "/content/ECOBRIDGE_DATA"
]

keywords = [
    "RGB-MFCO",
    "MFCO",
    "WEEE",
    "PLA",
    "plastic"
]

matches = []

for search_root in search_roots:

    if not os.path.exists(search_root):
        continue

    for root, dirs, files in os.walk(search_root):

        # Avoid extremely large / irrelevant directories
        dirs[:] = [
            d for d in dirs
            if d not in [
                ".git",
                "__pycache__",
                "node_modules"
            ]
        ]

        for file in files:

            lower = file.lower()

            if any(
                keyword.lower() in lower
                for keyword in keywords
            ):

                matches.append(
                    os.path.join(root, file)
                )

# Remove duplicates
matches = sorted(set(matches))

print("=" * 80)
print("STEP 72 — LOCAL WEEE DATASET SEARCH")
print("=" * 80)

print("Matches found:", len(matches))

for path in matches[:200]:
    print(path)

if len(matches) > 200:
    print("\nShowing first 200 matches only.")
```

```python
# =============================================================================
# STEP 72A — TARGETED WEEE DATASET SEARCH
# =============================================================================

import os

ROOT = "/content"

target_keywords = [
    "weee",
    "mfco",
    "rgb-mfco",
    "rgb_mfco",
    "electronic_waste",
    "ewaste"
]

matches = []

for root, dirs, files in os.walk(ROOT):

    # Check directory names
    for d in dirs:
        d_lower = d.lower()

        if any(k in d_lower for k in target_keywords):
            matches.append(
                ("DIR", os.path.join(root, d))
            )

    # Check filenames
    for file in files:
        file_lower = file.lower()

        if any(k in file_lower for k in target_keywords):
            matches.append(
                ("FILE", os.path.join(root, file))
            )

matches = sorted(set(matches))

print("=" * 80)
print("STEP 72A — TARGETED WEEE DATASET SEARCH")
print("=" * 80)

print("WEEE-related matches:", len(matches))

for kind, path in matches[:300]:
    print(f"[{kind}] {path}")

if len(matches) > 300:
    print("\nShowing first 300 matches only.")
```

```python
# =============================================================================
# STEP 74 — LOCATE RGB-MFCO DOWNLOAD LINK
# =============================================================================

import requests
from bs4 import BeautifulSoup

URL = "https://doi.org/10.48328/tudatalib-1743"

headers = {
    "User-Agent": "Mozilla/5.0 ECOBRIDGE dataset research"
}

print("=" * 80)
print("STEP 74 — RGB-MFCO DATASET LINK CHECK")
print("=" * 80)

try:

    response = requests.get(
        URL,
        headers=headers,
        timeout=30,
        allow_redirects=True
    )

    print("Status:", response.status_code)
    print("Final URL:", response.url)

    soup = BeautifulSoup(
        response.text,
        "html.parser"
    )

    links = []

    for a in soup.find_all("a", href=True):

        href = a["href"]
        text = a.get_text(" ", strip=True)

        combined = (
            text + " " + href
        ).lower()

        if (
            "rgb-mfco" in combined
            or ".zip" in combined
            or "download" in combined
        ):
            links.append(
                (text, href)
            )

    print("\nPotential dataset/download links:")

    for text, href in links:

        print(
            f"\nTEXT: {text}\n"
            f"URL:  {href}"
        )

    print(
        "\nTotal potential links:",
        len(links)
    )

except Exception as e:

    print("ERROR:", e)
```

```python
# =============================================================================
# STEP 81 — DEEP IMAGE INVENTORY FOR MOTOR + PLASTIC SOURCES
# =============================================================================

import os
from collections import Counter

ROOT = "/content/ECOBRIDGE_DATA"

IMAGE_EXTS = {
    ".jpg", ".jpeg", ".png", ".bmp", ".webp"
}

keywords = [
    "motor",
    "motors",
    "plastic",
    "plastics",
    "weee",
    "ewaste",
    "e-waste",
    "electronic",
    "scrap"
]

print("=" * 80)
print("STEP 81 — DEEP IMAGE INVENTORY")
print("=" * 80)

results = []

for root, dirs, files in os.walk(ROOT):

    for file in files:

        ext = os.path.splitext(file)[1].lower()

        if ext not in IMAGE_EXTS:
            continue

        full_path = os.path.join(root, file)

        relative = os.path.relpath(full_path, ROOT).lower()

        if any(k in relative or k in file.lower() for k in keywords):

            results.append(
                (
                    relative,
                    os.path.getsize(full_path)
                )
            )

print("\nTotal keyword-matching images:", len(results))

# -------------------------------------------------------------------------
# Group by top-level source
# -------------------------------------------------------------------------

groups = Counter()

for path, size in results:

    parts = path.split(os.sep)

    if len(parts) >= 2:
        groups[os.path.join(parts[0], parts[1])] += 1
    else:
        groups[parts[0]] += 1

print("\nIMAGE COUNTS BY SOURCE/FOLDER:")
print("-" * 80)

for folder, count in groups.most_common():

    print(f"{count:6d} | {folder}")

# -------------------------------------------------------------------------
# Show motor-related paths
# -------------------------------------------------------------------------

print("\n" + "=" * 80)
print("MOTOR-RELATED IMAGE PATHS")
print("=" * 80)

motor_results = [
    (path, size)
    for path, size in results
    if any(k in path.lower() for k in ["motor", "motors"])
]

print("Count:", len(motor_results))

for path, size in motor_results[:200]:
    print(f"{size/1024:8.1f} KB | {path}")

# -------------------------------------------------------------------------
# Show plastic-related paths
# -------------------------------------------------------------------------

print("\n" + "=" * 80)
print("PLASTIC-RELATED IMAGE PATHS")
print("=" * 80)

plastic_results = [
    (path, size)
    for path, size in results
    if any(k in path.lower() for k in ["plastic", "plastics"])
]

print("Count:", len(plastic_results))

for path, size in plastic_results[:200]:
    print(f"{size/1024:8.1f} KB | {path}")
```

```python
# =============================================================================
# STEP 82 — INVENTORY TRASHBOX E-WASTE SUBCATEGORIES
# =============================================================================

import os
from collections import Counter

TRASHBOX = "/content/ECOBRIDGE_DATA/external_sources/trashbox_repo"

IMAGE_EXTS = {".jpg", ".jpeg", ".png", ".bmp", ".webp"}

print("=" * 80)
print("STEP 82 — TRASHBOX E-WASTE SUBCATEGORY INVENTORY")
print("=" * 80)

# Find all folders containing e-waste in their path
ewaste_dirs = []

for root, dirs, files in os.walk(TRASHBOX):

    if "e-waste" in root.lower() or "ewaste" in root.lower():

        image_count = sum(
            1
            for f in files
            if os.path.splitext(f)[1].lower() in IMAGE_EXTS
        )

        if image_count > 0:
            ewaste_dirs.append((root, image_count))

print("\nE-WASTE IMAGE FOLDERS:")
print("-" * 80)

for path, count in sorted(ewaste_dirs, key=lambda x: x[0].lower()):

    relative = os.path.relpath(path, TRASHBOX)

    print(f"{count:6d} | {relative}")

print("\nTotal e-waste folders:", len(ewaste_dirs))
print("Total e-waste images:",
      sum(count for _, count in ewaste_dirs))
```

```python
# =============================================================================
# STEP 84 — DOWNLOAD XBAT+ RGB TRAIN + INSPECT LABELS
# =============================================================================

import requests
import zipfile
import os

BASE = "/content/ECOBRIDGE_DATA/external_sources/xbat_plus"
os.makedirs(BASE, exist_ok=True)

ZIP_URL = (
    "https://zenodo.org/records/18022530/files/"
    "raw_XBAT+_v1.0_RGB_train.zip"
)

ZIP_PATH = os.path.join(BASE, "xbat_rgb_train.zip")

print("=" * 80)
print("STEP 84 — XBAT+ RGB TRAINING DATA")
print("=" * 80)

print("\nDownloading XBAT+ RGB training ZIP...")

r = requests.get(
    ZIP_URL,
    stream=True,
    timeout=120
)

print("HTTP status:", r.status_code)

if r.status_code != 200:

    print("\nDownload failed.")
    print(r.text[:1000])

else:

    total = int(r.headers.get("content-length", 0))

    with open(ZIP_PATH, "wb") as f:

        downloaded = 0

        for chunk in r.iter_content(chunk_size=1024 * 1024):

            if chunk:

                f.write(chunk)
                downloaded += len(chunk)

                if total:

                    pct = downloaded / total * 100

                    print(
                        f"\rDownloaded: "
                        f"{downloaded / (1024**2):.1f} MB "
                        f"({pct:.1f}%)",
                        end=""
                    )

    print("\n\nDownload complete.")

    print(
        "ZIP size:",
        f"{os.path.getsize(ZIP_PATH) / (1024**2):.2f} MB"
    )

    # -------------------------------------------------------------------------
    # Inspect ZIP WITHOUT extracting the image dataset
    # -------------------------------------------------------------------------

    print("\nZIP CONTENTS")
    print("-" * 80)

    with zipfile.ZipFile(ZIP_PATH, "r") as z:

        names = z.namelist()

        print("Total ZIP entries:", len(names))

        for name in names[:100]:
            print(name)

        # ---------------------------------------------------------------------
        # Find likely metadata / label files
        # ---------------------------------------------------------------------

        print("\n" + "-" * 80)
        print("POTENTIAL LABEL / METADATA FILES")
        print("-" * 80)

        label_files = []

        for name in names:

            lower = name.lower()

            if any(
                key in os.path.basename(lower)
                for key in [
                    "label",
                    "class",
                    "data",
                    "yaml",
                    "txt",
                    "csv"
                ]
            ):

                label_files.append(name)

        if label_files:

            for name in label_files[:100]:
                info = z.getinfo(name)

                print(
                    f"{info.file_size / 1024:.1f} KB | {name}"
                )

        else:

            print("No obvious label/metadata files found.")

        # ---------------------------------------------------------------------
        # Extract only small metadata files
        # ---------------------------------------------------------------------

        metadata_dir = os.path.join(BASE, "metadata")
        os.makedirs(metadata_dir, exist_ok=True)

        extracted = 0

        for name in label_files:

            info = z.getinfo(name)

            if info.file_size <= 2 * 1024 * 1024:

                try:
                    z.extract(name, metadata_dir)
                    extracted += 1
                except Exception as e:
                    print("Could not extract:", name, "|", e)

        print(
            "\nSmall metadata/label files extracted:",
            extracted
        )

print("\nSTEP 84 COMPLETE")
```

```python
# =============================================================================
# STEP 85 — ANALYZE XBAT+ CLASS IDS
# =============================================================================

import os
from collections import Counter, defaultdict

LABEL_ROOT = (
    "/content/ECOBRIDGE_DATA/external_sources/"
    "xbat_plus/metadata/raw_XBAT+_v1.0_RGB_train/"
    "train_data/labels"
)

print("=" * 80)
print("STEP 85 — XBAT+ CLASS ID ANALYSIS")
print("=" * 80)

class_counts = Counter()
class_examples = defaultdict(list)

total_files = 0
empty_files = 0
bad_files = 0

for root, dirs, files in os.walk(LABEL_ROOT):

    for file in files:

        if not file.endswith(".txt"):
            continue

        total_files += 1

        path = os.path.join(root, file)

        with open(path, "r", encoding="utf-8", errors="ignore") as f:
            lines = [line.strip() for line in f if line.strip()]

        if not lines:
            empty_files += 1
            continue

        for line in lines:

            parts = line.split()

            if len(parts) < 5:
                bad_files += 1
                continue

            try:
                class_id = int(parts[0])
            except:
                bad_files += 1
                continue

            class_counts[class_id] += 1

            if len(class_examples[class_id]) < 5:
                class_examples[class_id].append(
                    (file, line)
                )

print("\nLabel files found:", total_files)
print("Empty label files:", empty_files)
print("Bad annotation lines:", bad_files)

print("\n" + "-" * 80)
print("CLASS ID COUNTS")
print("-" * 80)

for class_id in sorted(class_counts):

    print(
        f"Class ID {class_id:2d} : "
        f"{class_counts[class_id]:4d} annotations"
    )

print("\n" + "-" * 80)
print("EXAMPLES PER CLASS ID")
print("-" * 80)

for class_id in sorted(class_examples):

    print(f"\nCLASS ID {class_id}")

    for filename, annotation in class_examples[class_id]:
        print(" ", filename)
        print("   ", annotation)

print("\n" + "=" * 80)
print("TOTAL ANNOTATIONS:", sum(class_counts.values()))
print("=" * 80)
```

```python
# =============================================================================
# STEP 86 — READ OFFICIAL XBAT+ METADATA
# =============================================================================

import requests
import zipfile
import os

BASE = "/content/ECOBRIDGE_DATA/external_sources/xbat_plus"

URL = (
    "https://zenodo.org/records/18022530/files/"
    "raw_XBAT+_v1.0_metadata.zip"
)

ZIP_PATH = os.path.join(BASE, "xbat_metadata.zip")

print("=" * 80)
print("STEP 86 — XBAT+ OFFICIAL METADATA")
print("=" * 80)

r = requests.get(URL, timeout=60)

print("HTTP status:", r.status_code)

if r.status_code == 200:

    with open(ZIP_PATH, "wb") as f:
        f.write(r.content)

    print(
        "Downloaded:",
        f"{len(r.content) / 1024:.2f} KB"
    )

    with zipfile.ZipFile(ZIP_PATH, "r") as z:

        names = z.namelist()

        print("\nZIP CONTENTS:")
        print("-" * 80)

        for name in names:
            print(name)

        print("\n" + "-" * 80)
        print("METADATA CONTENT:")
        print("-" * 80)

        for name in names:

            if name.endswith("/"):

                continue

            try:

                content = z.read(name).decode(
                    "utf-8",
                    errors="replace"
                )

                print(f"\n### {name}")
                print(content[:10000])

            except Exception as e:

                print(
                    f"\nCould not read {name}: {e}"
                )

else:

    print("\nDownload failed.")
    print(r.text[:1000])
```

```python
# =============================================================================
# STEP 88 — VISUAL AUDIT OF EXISTING MOTOR IMAGES
# =============================================================================

from PIL import Image, ImageDraw
import os
import math
import matplotlib.pyplot as plt

MOTOR_DIR = "/content/ECOBRIDGE_DATA/external_sources/motor_audit_step68"

files = sorted([
    f for f in os.listdir(MOTOR_DIR)
    if f.lower().endswith((".jpg", ".jpeg", ".png", ".webp"))
])

thumb_w = 300
thumb_h = 250
label_h = 45

cols = 2
rows = math.ceil(len(files) / cols)

sheet = Image.new(
    "RGB",
    (cols * thumb_w, rows * (thumb_h + label_h)),
    "white"
)

draw = ImageDraw.Draw(sheet)

for i, filename in enumerate(files):

    path = os.path.join(MOTOR_DIR, filename)

    img = Image.open(path).convert("RGB")
    img.thumbnail((thumb_w - 20, thumb_h - 20))

    x = (i % cols) * thumb_w
    y = (i // cols) * (thumb_h + label_h)

    px = x + (thumb_w - img.width) // 2
    py = y + 5

    sheet.paste(img, (px, py))

    draw.text(
        (x + 10, y + thumb_h + 5),
        f"{i:02d}  {filename}",
        fill="black"
    )

display(sheet)

print("\nTotal images:", len(files))
print("\nPlease inspect whether each image shows:")
print("1. A clearly identifiable electric motor/component")
print("2. A complete machine/vehicle rather than a motor")
print("3. A diagram or non-photo")
print("4. A motor that could realistically appear in e-waste")
```

```python
# =============================================================================
# STEP 89 — SEARCH ALL LOCAL DATA FOR MOTOR CANDIDATES
# =============================================================================

import os
import pandas as pd

ROOT = "/content/ECOBRIDGE_DATA"

motor_keywords = [
    "motor",
    "motors",
    "moteur",
    "electromotor",
    "electricmotor",
    "stepper",
    "rotor",
    "stator",
    "hubmotor",
    "dc_motor",
    "ac_motor"
]

records = []

for root, dirs, files in os.walk(ROOT):

    # Skip temporary/cache directories
    dirs[:] = [
        d for d in dirs
        if d not in [".git", "__pycache__", ".ipynb_checkpoints"]
    ]

    for fname in files:

        if not fname.lower().endswith(
            (".jpg", ".jpeg", ".png", ".webp", ".bmp")
        ):
            continue

        full_path = os.path.join(root, fname)
        searchable = (full_path + " " + fname).lower()

        matched = [
            kw for kw in motor_keywords
            if kw in searchable
        ]

        if matched:
            records.append({
                "filename": fname,
                "path": full_path,
                "matched_keywords": ", ".join(matched)
            })

motor_local_df = pd.DataFrame(records)

print("=" * 80)
print("STEP 89 — LOCAL MOTOR CANDIDATE SEARCH")
print("=" * 80)

print("\nTotal motor-keyword image paths:", len(motor_local_df))

if len(motor_local_df) > 0:

    print("\nSOURCE DIRECTORY COUNTS:")
    print("-" * 80)

    source_counts = (
        motor_local_df["path"]
        .apply(lambda x: os.path.dirname(x))
        .value_counts()
    )

    for source, count in source_counts.items():
        print(f"{count:4d} | {source}")

    print("\n" + "-" * 80)
    print("ALL CANDIDATES:")
    print("-" * 80)

    for i, row in motor_local_df.iterrows():
        print(
            f"{i:03d} | "
            f"{row['filename']:45} | "
            f"{row['matched_keywords']:25} | "
            f"{row['path']}"
        )

else:
    print("\nNo motor-keyword images found.")

print("\nSTEP 89 COMPLETE")
```

```python
# =============================================================================
# STEP 90 — CHECK LOCAL STORAGE FOR WEEE MATERIAL DATASET
# =============================================================================

import os

ROOT = "/content/ECOBRIDGE_DATA"

keywords = [
    "mfco",
    "rgb-mfco",
    "wEEE",
    "weee",
    "plastic",
    "shredded"
]

matches = []

for root, dirs, files in os.walk(ROOT):

    dirs[:] = [
        d for d in dirs
        if d not in [".git", "__pycache__", ".ipynb_checkpoints"]
    ]

    for fname in files:

        full_path = os.path.join(root, fname)

        searchable = (full_path + " " + fname).lower()

        if any(k.lower() in searchable for k in keywords):

            matches.append({
                "filename": fname,
                "path": full_path,
                "size_mb": os.path.getsize(full_path) / (1024 * 1024)
            })

print("=" * 80)
print("STEP 90 — LOCAL WEEE MATERIAL DATASET CHECK")
print("=" * 80)

print("\nMatching files:", len(matches))

for i, item in enumerate(matches):

    print(
        f"{i:03d} | "
        f"{item['size_mb']:8.2f} MB | "
        f"{item['path']}"
    )

print("\nSTEP 90 COMPLETE")
```

```python
# =============================================================================
# STEP 94 — FINAL 6-CLASS DUPLICATE & INTEGRITY AUDIT
# =============================================================================

import os
import hashlib
import pandas as pd
from PIL import Image
from collections import defaultdict

BASE = "/content/ECOBRIDGE_DATA"
FINAL_DIR = os.path.join(BASE, "ECOBRIDGE_FINAL_6CLASS")

CLASSES = [
    "CAT_BATTERY",
    "CAT_PCB",
    "CAT_CRT",
    "CAT_LCD_LED",
    "CAT_CABLES",
    "CAT_OTHER"
]

IMAGE_EXTENSIONS = (".jpg", ".jpeg", ".png", ".webp", ".bmp")

records = []

print("=" * 80)
print("STEP 94 — FINAL 6-CLASS DATASET AUDIT")
print("=" * 80)

# -------------------------------------------------------------------------
# Scan all source images
# -------------------------------------------------------------------------

for class_name in CLASSES:

    source_dir = {
        "CAT_BATTERY": os.path.join(
            BASE, "ECOBRIDGE_VERIFIED_5CLASS", "all", "CAT_BATTERY"
        ),
        "CAT_PCB": os.path.join(
            BASE, "ECOBRIDGE_VERIFIED_5CLASS", "all", "CAT_PCB"
        ),
        "CAT_CRT": os.path.join(
            BASE, "ECOBRIDGE_VERIFIED_5CLASS", "all", "CAT_CRT"
        ),
        "CAT_LCD_LED": os.path.join(
            BASE, "ECOBRIDGE_VERIFIED_5CLASS", "all", "CAT_LCD_LED"
        ),
        "CAT_OTHER": os.path.join(
            BASE, "ECOBRIDGE_VERIFIED_5CLASS", "all", "CAT_OTHER"
        ),
        "CAT_CABLES": os.path.join(
            BASE,
            "external_sources",
            "trashbox_repo",
            "TrashBox_train_dataset_subfolders",
            "e-waste",
            "electrical cables"
        )
    }[class_name]

    files = [
        os.path.join(source_dir, f)
        for f in os.listdir(source_dir)
        if f.lower().endswith(IMAGE_EXTENSIONS)
    ]

    print(f"\n{class_name}: {len(files)} images")

    for path in sorted(files):

        try:

            # -------------------------------------------------------------
            # Verify image can actually be opened
            # -------------------------------------------------------------
            with Image.open(path) as img:
                img.verify()

            # Re-open after verify for metadata
            with Image.open(path) as img:
                width, height = img.size
                mode = img.mode

            # -------------------------------------------------------------
            # Exact pixel/file hash
            # -------------------------------------------------------------
            with open(path, "rb") as f:
                file_hash = hashlib.sha256(f.read()).hexdigest()

            records.append({
                "class_name": class_name,
                "filename": os.path.basename(path),
                "source_path": path,
                "width": width,
                "height": height,
                "mode": mode,
                "sha256": file_hash
            })

        except Exception as e:

            records.append({
                "class_name": class_name,
                "filename": os.path.basename(path),
                "source_path": path,
                "width": None,
                "height": None,
                "mode": None,
                "sha256": None,
                "error": str(e)
            })

# -------------------------------------------------------------------------
# Create metadata table
# -------------------------------------------------------------------------

audit_df = pd.DataFrame(records)

metadata_path = os.path.join(
    FINAL_DIR,
    "final_6class_source_audit.csv"
)

audit_df.to_csv(metadata_path, index=False)

# -------------------------------------------------------------------------
# Basic integrity statistics
# -------------------------------------------------------------------------

print("\n" + "=" * 80)
print("INTEGRITY SUMMARY")
print("=" * 80)

bad_images = audit_df[
    audit_df["sha256"].isna()
]

print(f"Total images scanned : {len(audit_df)}")
print(f"Readable images      : {audit_df['sha256'].notna().sum()}")
print(f"Unreadable images    : {len(bad_images)}")

print("\nImage dimensions:")
print(
    audit_df[
        audit_df["sha256"].notna()
    ][["width", "height"]]
    .value_counts()
    .head(15)
)

# -------------------------------------------------------------------------
# Exact duplicate detection
# -------------------------------------------------------------------------

valid_df = audit_df[
    audit_df["sha256"].notna()
].copy()

hash_groups = (
    valid_df
    .groupby("sha256")
    .agg(
        duplicate_count=("sha256", "size"),
        classes=("class_name", lambda x: sorted(set(x))),
        files=("filename", list)
    )
    .reset_index()
)

duplicate_groups = hash_groups[
    hash_groups["duplicate_count"] > 1
].copy()

print("\n" + "=" * 80)
print("EXACT DUPLICATE ANALYSIS")
print("=" * 80)

print(f"Unique image hashes : {hash_groups.shape[0]}")
print(f"Duplicate groups    : {len(duplicate_groups)}")
print(
    f"Images involved     : "
    f"{duplicate_groups['duplicate_count'].sum() if len(duplicate_groups) else 0}"
)

# -------------------------------------------------------------------------
# Cross-class exact duplicates
# -------------------------------------------------------------------------

cross_class = duplicate_groups[
    duplicate_groups["classes"].apply(len) > 1
].copy()

print(f"\nCross-class duplicate groups: {len(cross_class)}")

if len(cross_class) > 0:

    print("\nWARNING — SAME IMAGE APPEARS UNDER MULTIPLE CLASSES")

    for _, row in cross_class.head(20).iterrows():

        print("\nHash:", row["sha256"][:16] + "...")
        print("Classes:", row["classes"])
        print("Files:", row["files"])

else:

    print("✓ No exact duplicate appears under multiple classes.")

# -------------------------------------------------------------------------
# Cross-source / within-class duplicates
# -------------------------------------------------------------------------

print("\n" + "=" * 80)
print("WITHIN-CLASS DUPLICATES")
print("=" * 80)

if len(duplicate_groups) > 0:

    for _, row in duplicate_groups.head(20).iterrows():

        print(
            f"{row['duplicate_count']} copies | "
            f"classes={row['classes']} | "
            f"files={row['files'][:4]}"
        )

else:

    print("✓ No exact duplicates found.")

# -------------------------------------------------------------------------
# Class balance
# -------------------------------------------------------------------------

print("\n" + "=" * 80)
print("FINAL CLASS COUNTS")
print("=" * 80)

class_counts = (
    valid_df["class_name"]
    .value_counts()
    .reindex(CLASSES)
)

for class_name, count in class_counts.items():

    print(
        f"{class_name:15} : {count:5d}"
    )

print(
    f"\nTOTAL VALID IMAGES: {class_counts.sum()}"
)

print("\nMetadata saved to:")
print(metadata_path)

print("\nSTEP 94 COMPLETE")
```

```python
# =============================================================================
# STEP 96 — SOURCE-LEVEL PROVENANCE ANALYSIS
# =============================================================================

import os
import pandas as pd

BASE = "/content/ECOBRIDGE_DATA"

metadata_path = os.path.join(
    BASE,
    "ECOBRIDGE_VERIFIED_5CLASS",
    "candidate_metadata.csv"
)

df = pd.read_csv(metadata_path)

print("=" * 80)
print("STEP 96 — SOURCE-LEVEL PROVENANCE ANALYSIS")
print("=" * 80)

# -------------------------------------------------------------------------
# 1. Overall source distribution
# -------------------------------------------------------------------------

print("\n" + "-" * 80)
print("SOURCE DISTRIBUTION")
print("-" * 80)

source_counts = (
    df["source"]
    .value_counts()
    .rename_axis("source")
    .reset_index(name="images")
)

print(source_counts.to_string(index=False))

# -------------------------------------------------------------------------
# 2. Source × target class
# -------------------------------------------------------------------------

print("\n" + "-" * 80)
print("SOURCE × TARGET CLASS")
print("-" * 80)

source_class = pd.crosstab(
    df["source"],
    df["target_class"]
)

print(source_class.to_string())

# -------------------------------------------------------------------------
# 3. Original path structure
# -------------------------------------------------------------------------

print("\n" + "-" * 80)
print("ORIGINAL PATH EXAMPLES")
print("-" * 80)

for source in df["source"].unique():

    subset = df[df["source"] == source]

    print(f"\nSOURCE: {source}")
    print(f"Rows: {len(subset)}")

    for path in subset["original_path"].head(5):

        print(" ", path)

# -------------------------------------------------------------------------
# 4. Candidate filename patterns
# -------------------------------------------------------------------------

print("\n" + "-" * 80)
print("CANDIDATE FILENAME EXAMPLES")
print("-" * 80)

df["candidate_filename"] = df["candidate_path"].apply(
    lambda x: os.path.basename(str(x))
)

for source in df["source"].unique():

    subset = df[df["source"] == source]

    print(f"\nSOURCE: {source}")

    for filename in subset["candidate_filename"].head(10):

        print(" ", filename)

# -------------------------------------------------------------------------
# 5. Original filename examples
# -------------------------------------------------------------------------

print("\n" + "-" * 80)
print("ORIGINAL FILENAME EXAMPLES")
print("-" * 80)

df["original_filename"] = df["original_path"].apply(
    lambda x: os.path.basename(str(x))
)

for source in df["source"].unique():

    subset = df[df["source"] == source]

    print(f"\nSOURCE: {source}")

    for filename in subset["original_filename"].head(10):

        print(" ", filename)

# -------------------------------------------------------------------------
# 6. Check whether source contains multiple target classes
# -------------------------------------------------------------------------

print("\n" + "-" * 80)
print("SOURCE CLASS CARDINALITY")
print("-" * 80)

source_class_counts = (
    df.groupby("source")["target_class"]
    .nunique()
    .sort_values(ascending=False)
)

for source, count in source_class_counts.items():

    classes = sorted(
        df.loc[
            df["source"] == source,
            "target_class"
        ].unique()
    )

    print(
        f"{source:30} | "
        f"{count} target class(es) | "
        f"{classes}"
    )

# -------------------------------------------------------------------------
# 7. Verify candidate paths are unique
# -------------------------------------------------------------------------

print("\n" + "-" * 80)
print("CANDIDATE PATH UNIQUENESS")
print("-" * 80)

print(
    "Total candidate rows:",
    len(df)
)

print(
    "Unique candidate paths:",
    df["candidate_path"].nunique()
)

print(
    "Duplicate candidate paths:",
    len(df) - df["candidate_path"].nunique()
)

# -------------------------------------------------------------------------
# 8. Save enhanced provenance inspection
# -------------------------------------------------------------------------

inspection_path = os.path.join(
    BASE,
    "ECOBRIDGE_FINAL_6CLASS",
    "source_provenance_inspection.csv"
)

df.to_csv(
    inspection_path,
    index=False
)

print("\nSaved:")
print(inspection_path)

print("\nSTEP 96 COMPLETE")
```

```python
# =============================================================================
# STEP 97 — CREATE LEAKAGE-AWARE SPLIT PLAN
# =============================================================================

import os
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split

# -------------------------------------------------------------------------
# Paths
# -------------------------------------------------------------------------

BASE = "/content/ECOBRIDGE_DATA"

metadata_path = os.path.join(
    BASE,
    "ECOBRIDGE_VERIFIED_5CLASS",
    "candidate_metadata.csv"
)

split_plan_path = os.path.join(
    BASE,
    "ECOBRIDGE_FINAL_6CLASS",
    "split_plan.csv"
)

# -------------------------------------------------------------------------
# Load metadata
# -------------------------------------------------------------------------

df = pd.read_csv(metadata_path)

# -------------------------------------------------------------------------
# IMPORTANT:
# The verified 5-class metadata contains:
#   CAT_BATTERY
#   CAT_PCB
#   CAT_CRT
#   CAT_LCD_LED
#   CAT_OTHER
#
# CAT_CABLES is handled separately because it comes from TrashBox.
# -------------------------------------------------------------------------

print("=" * 80)
print("STEP 97 — LEAKAGE-AWARE SPLIT PLAN")
print("=" * 80)

# -------------------------------------------------------------------------
# Add source filename information
# -------------------------------------------------------------------------

df["filename"] = df["candidate_path"].apply(
    lambda x: os.path.basename(str(x))
)

# -------------------------------------------------------------------------
# Identify original source split for Roboflow Battery
# -------------------------------------------------------------------------

def get_battery_original_split(path):

    path = str(path).replace("\\", "/").lower()

    if "/train/" in path:
        return "original_train"

    if "/valid/" in path or "/val/" in path:
        return "original_val"

    if "/test/" in path:
        return "original_test"

    return "unknown"


df["original_source_split"] = "not_applicable"

battery_mask = (
    df["source"] == "ROBOFLOW_BATTERY_CLEAN"
)

df.loc[battery_mask, "original_source_split"] = (
    df.loc[battery_mask, "original_path"]
    .apply(get_battery_original_split)
)

# -------------------------------------------------------------------------
# Show Roboflow provenance
# -------------------------------------------------------------------------

print("\n" + "-" * 80)
print("ROBOFLOW BATTERY ORIGINAL SPLIT")
print("-" * 80)

print(
    df.loc[battery_mask, "original_source_split"]
    .value_counts()
)

# -------------------------------------------------------------------------
# Split function
# -------------------------------------------------------------------------

def stratified_split(group, random_state=42):

    group = group.copy()

    # -------------------------------------------------------------
    # First: 10% test
    # -------------------------------------------------------------

    train_val, test = train_test_split(
        group,
        test_size=0.10,
        random_state=random_state,
        stratify=group["target_class"]
    )

    # -------------------------------------------------------------
    # Then: 10% validation overall
    # 10 / 90 = 0.1111 of train_val
    # -------------------------------------------------------------

    train, val = train_test_split(
        train_val,
        test_size=(0.10 / 0.90),
        random_state=random_state,
        stratify=train_val["target_class"]
    )

    train = train.copy()
    val = val.copy()
    test = test.copy()

    train["split"] = "train"
    val["split"] = "val"
    test["split"] = "test"

    return pd.concat(
        [train, val, test],
        ignore_index=True
    )

# -------------------------------------------------------------------------
# Create split plan for verified 5-class pool
# -------------------------------------------------------------------------

split_parts = []

for source in sorted(df["source"].unique()):

    source_df = df[
        df["source"] == source
    ].copy()

    print(
        f"\nSource: {source}"
    )

    print(
        "Images:",
        len(source_df)
    )

    # -------------------------------------------------------------
    # Roboflow Battery:
    # Preserve original train/val/test boundaries.
    # -------------------------------------------------------------

    if source == "ROBOFLOW_BATTERY_CLEAN":

        source_df["split"] = source_df[
            "original_source_split"
        ].map({
            "original_train": "train",
            "original_val": "val",
            "original_test": "test"
        })

        if source_df["split"].isna().any():

            print(
                "WARNING: Unknown Roboflow source split detected."
            )

        else:

            print(
                source_df["split"]
                .value_counts()
                .sort_index()
                .to_dict()
            )

        split_parts.append(source_df)

    # -------------------------------------------------------------
    # Other verified sources:
    # Stratify within the source.
    # -------------------------------------------------------------

    else:

        # For each source, all rows generally belong to one target class.
        # Stratification remains explicit for safety.

        result = stratified_split(
            source_df,
            random_state=42
        )

        print(
            result["split"]
            .value_counts()
            .sort_index()
            .to_dict()
        )

        split_parts.append(result)

# -------------------------------------------------------------------------
# Combine
# -------------------------------------------------------------------------

split_df = pd.concat(
    split_parts,
    ignore_index=True
)

# -------------------------------------------------------------------------
# Add CAT_CABLES
# -------------------------------------------------------------------------

cables_dir = os.path.join(
    BASE,
    "external_sources",
    "trashbox_repo",
    "TrashBox_train_dataset_subfolders",
    "e-waste",
    "electrical cables"
)

cable_files = sorted([
    os.path.join(cables_dir, f)
    for f in os.listdir(cables_dir)
    if f.lower().endswith(
        (".jpg", ".jpeg", ".png", ".webp", ".bmp")
    )
])

cable_df = pd.DataFrame({
    "target_class": "CAT_CABLES",
    "source": "TRASHBOX_ELECTRICAL_CABLES",
    "original_path": cable_files,
    "candidate_path": cable_files,
    "filename": [
        os.path.basename(x)
        for x in cable_files
    ],
    "original_source_split": "not_applicable"
})

# Stratified split for cables
cable_df = stratified_split(
    cable_df,
    random_state=42
)

# -------------------------------------------------------------------------
# Add cables
# -------------------------------------------------------------------------

split_df = pd.concat(
    [split_df, cable_df],
    ignore_index=True
)

# -------------------------------------------------------------------------
# Create stable row ID
# -------------------------------------------------------------------------

split_df.insert(
    0,
    "sample_id",
    [
        f"ECO_{i:05d}"
        for i in range(len(split_df))
    ]
)

# -------------------------------------------------------------------------
# Check total
# -------------------------------------------------------------------------

print("\n" + "=" * 80)
print("FINAL SPLIT COUNTS")
print("=" * 80)

print(
    split_df["split"]
    .value_counts()
    .sort_index()
)

print("\n" + "-" * 80)
print("CLASS × SPLIT")
print("-" * 80)

class_split = pd.crosstab(
    split_df["target_class"],
    split_df["split"]
)

print(class_split)

# -------------------------------------------------------------------------
# Source × split
# -------------------------------------------------------------------------

print("\n" + "-" * 80)
print("SOURCE × SPLIT")
print("-" * 80)

source_split = pd.crosstab(
    split_df["source"],
    split_df["split"]
)

print(source_split)

# -------------------------------------------------------------------------
# Verify every candidate path is assigned exactly once
# -------------------------------------------------------------------------

print("\n" + "-" * 80)
print("ASSIGNMENT CHECK")
print("-" * 80)

print(
    "Total rows:",
    len(split_df)
)

print(
    "Unique candidate paths:",
    split_df["candidate_path"].nunique()
)

print(
    "Duplicate candidate paths:",
    len(split_df)
    - split_df["candidate_path"].nunique()
)

print(
    "Missing split values:",
    split_df["split"].isna().sum()
)

# -------------------------------------------------------------------------
# Save
# -------------------------------------------------------------------------

split_df.to_csv(
    split_plan_path,
    index=False
)

print("\nSplit plan saved to:")
print(split_plan_path)

print("\nSTEP 97 COMPLETE")
```

```python
# =============================================================================
# STEP 98 — EXACT-DUPLICATE-SAFE FINAL SPLIT
# =============================================================================

import os
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split

BASE = "/content/ECOBRIDGE_DATA"

split_plan_path = os.path.join(
    BASE,
    "ECOBRIDGE_FINAL_6CLASS",
    "split_plan.csv"
)

audit_path = os.path.join(
    BASE,
    "ECOBRIDGE_FINAL_6CLASS",
    "final_6class_source_audit.csv"
)

output_path = os.path.join(
    BASE,
    "ECOBRIDGE_FINAL_6CLASS",
    "split_plan_duplicate_safe.csv"
)

# -------------------------------------------------------------------------
# Load existing split plan and audit hashes
# -------------------------------------------------------------------------

split_df = pd.read_csv(split_plan_path)
audit_df = pd.read_csv(audit_path)

print("=" * 80)
print("STEP 98 — EXACT-DUPLICATE-SAFE SPLIT")
print("=" * 80)

# -------------------------------------------------------------------------
# Merge SHA256 hashes
# -------------------------------------------------------------------------

hash_lookup = audit_df[
    ["source_path", "sha256"]
].copy()

hash_lookup = hash_lookup.rename(
    columns={"source_path": "candidate_path"}
)

split_df = split_df.drop(
    columns=["sha256"],
    errors="ignore"
)

split_df = split_df.merge(
    hash_lookup,
    on="candidate_path",
    how="left",
    validate="one_to_one"
)

print("\nHash coverage:")
print(
    "Rows:",
    len(split_df)
)

print(
    "Missing hashes:",
    split_df["sha256"].isna().sum()
)

# -------------------------------------------------------------------------
# Find exact duplicate groups
# -------------------------------------------------------------------------

valid = split_df[
    split_df["sha256"].notna()
].copy()

hash_counts = (
    valid["sha256"]
    .value_counts()
)

duplicate_hashes = set(
    hash_counts[
        hash_counts > 1
    ].index
)

duplicate_df = valid[
    valid["sha256"].isin(duplicate_hashes)
].copy()

print("\n" + "-" * 80)
print("DUPLICATE GROUPS")
print("-" * 80)

print(
    "Exact duplicate groups:",
    len(duplicate_hashes)
)

print(
    "Images in duplicate groups:",
    len(duplicate_df)
)

# -------------------------------------------------------------------------
# Check current leakage
# -------------------------------------------------------------------------

current_duplicate_split = (
    duplicate_df
    .groupby("sha256")["split"]
    .nunique()
)

leaking_groups = current_duplicate_split[
    current_duplicate_split > 1
]

print(
    "Duplicate groups crossing splits:",
    len(leaking_groups)
)

# -------------------------------------------------------------------------
# Preserve Roboflow original split
# -------------------------------------------------------------------------

roboflow_mask = (
    split_df["source"] == "ROBOFLOW_BATTERY_CLEAN"
)

# Roboflow already has an authoritative source split.
# We do NOT redistribute those images.
#
# We only repair exact-duplicate leakage among the other sources.
#
# Since the earlier audit found no cross-source exact duplicates,
# duplicate groups outside Roboflow belong to a single source.
# -------------------------------------------------------------------------

non_roboflow = split_df[
    ~roboflow_mask
].copy()

roboflow = split_df[
    roboflow_mask
].copy()

# -------------------------------------------------------------------------
# Rebuild non-Roboflow assignments using duplicate groups
# -------------------------------------------------------------------------

rng = np.random.RandomState(42)

new_parts = []

for source in sorted(
    non_roboflow["source"].unique()
):

    source_df = non_roboflow[
        non_roboflow["source"] == source
    ].copy()

    print(
        f"\nProcessing source: {source}"
    )

    # -------------------------------------------------------------
    # Create group ID:
    # exact duplicate images share one group.
    # unique images each become their own group.
    # -------------------------------------------------------------

    source_df["group_id"] = source_df["sha256"]

    missing_hash_mask = source_df["group_id"].isna()

    source_df.loc[
        missing_hash_mask,
        "group_id"
    ] = [
        f"UNIQUE_{i}"
        for i in source_df.index[missing_hash_mask]
    ]

    # -------------------------------------------------------------
    # One row per group
    # -------------------------------------------------------------

    groups = (
        source_df
        .groupby("group_id", as_index=False)
        .agg(
            group_size=("group_id", "size"),
            target_class=("target_class", "first")
        )
    )

    # Randomize group order reproducibly
    groups = groups.sample(
        frac=1,
        random_state=42
    ).reset_index(drop=True)

    # -------------------------------------------------------------
    # Greedy assignment toward 80/10/10
    # -------------------------------------------------------------

    total_images = len(source_df)

    target_counts = {
        "train": total_images * 0.80,
        "val": total_images * 0.10,
        "test": total_images * 0.10
    }

    assigned_counts = {
        "train": 0,
        "val": 0,
        "test": 0
    }

    group_assignments = {}

    for _, row in groups.iterrows():

        group_size = int(row["group_size"])

        # Choose split with largest remaining capacity
        remaining = {
            split: target_counts[split] - assigned_counts[split]
            for split in target_counts
        }

        selected_split = max(
            remaining,
            key=remaining.get
        )

        group_assignments[
            row["group_id"]
        ] = selected_split

        assigned_counts[
            selected_split
        ] += group_size

    # -------------------------------------------------------------
    # Apply assignments
    # -------------------------------------------------------------

    source_df["split"] = source_df[
        "group_id"
    ].map(group_assignments)

    new_parts.append(
        source_df
    )

    print(
        "Images:",
        total_images
    )

    print(
        "Groups:",
        len(groups)
    )

    print(
        "Final:",
        source_df["split"]
        .value_counts()
        .sort_index()
        .to_dict()
    )

# -------------------------------------------------------------------------
# Recombine
# -------------------------------------------------------------------------

fixed_non_roboflow = pd.concat(
    new_parts,
    ignore_index=True
)

final_df = pd.concat(
    [
        fixed_non_roboflow,
        roboflow
    ],
    ignore_index=True
)

# -------------------------------------------------------------------------
# Restore original row order by sample ID
# -------------------------------------------------------------------------

final_df = final_df.sort_values(
    "sample_id"
).reset_index(drop=True)

# -------------------------------------------------------------------------
# Verify every candidate appears exactly once
# -------------------------------------------------------------------------

print("\n" + "=" * 80)
print("FINAL ASSIGNMENT CHECK")
print("=" * 80)

print(
    "Total rows:",
    len(final_df)
)

print(
    "Unique candidate paths:",
    final_df["candidate_path"].nunique()
)

print(
    "Duplicate candidate paths:",
    len(final_df)
    - final_df["candidate_path"].nunique()
)

print(
    "Missing splits:",
    final_df["split"].isna().sum()
)

# -------------------------------------------------------------------------
# Verify exact duplicate groups do NOT cross splits
# -------------------------------------------------------------------------

duplicate_split_counts = (
    final_df[
        final_df["sha256"].notna()
    ]
    .groupby("sha256")["split"]
    .nunique()
)

remaining_leakage = duplicate_split_counts[
    duplicate_split_counts > 1
]

print(
    "\nExact duplicate groups crossing splits:",
    len(remaining_leakage)
)

if len(remaining_leakage) == 0:
    print(
        "✓ NO EXACT DUPLICATE GROUP CROSSES TRAIN/VAL/TEST"
    )
else:
    print(
        "⚠️ DUPLICATE LEAKAGE STILL EXISTS"
    )

# -------------------------------------------------------------------------
# Class × split
# -------------------------------------------------------------------------

print("\n" + "-" * 80)
print("CLASS × SPLIT")
print("-" * 80)

class_split = pd.crosstab(
    final_df["target_class"],
    final_df["split"]
)

print(class_split)

# -------------------------------------------------------------------------
# Overall split
# -------------------------------------------------------------------------

print("\n" + "-" * 80)
print("OVERALL SPLIT")
print("-" * 80)

overall = (
    final_df["split"]
    .value_counts()
    .sort_index()
)

print(overall)

print(
    "\nPercentages:"
)

print(
    (
        overall / len(final_df) * 100
    ).round(2)
)

# -------------------------------------------------------------------------
# Source × split
# -------------------------------------------------------------------------

print("\n" + "-" * 80)
print("SOURCE × SPLIT")
print("-" * 80)

print(
    pd.crosstab(
        final_df["source"],
        final_df["split"]
    )
)

# -------------------------------------------------------------------------
# Save
# -------------------------------------------------------------------------

final_df.to_csv(
    output_path,
    index=False
)

print("\nSaved final duplicate-safe split plan:")
print(output_path)

print("\nSTEP 98 COMPLETE")
```

```python
# =============================================================================
# STEP 99 — NEAR-DUPLICATE SPLIT LEAKAGE AUDIT
# =============================================================================

import os
import hashlib
import numpy as np
import pandas as pd
from PIL import Image
from tqdm.auto import tqdm

BASE = "/content/ECOBRIDGE_DATA"

split_path = os.path.join(
    BASE,
    "ECOBRIDGE_FINAL_6CLASS",
    "split_plan_duplicate_safe.csv"
)

print("=" * 80)
print("STEP 99 — NEAR-DUPLICATE SPLIT LEAKAGE AUDIT")
print("=" * 80)

# -------------------------------------------------------------------------
# Load split plan
# -------------------------------------------------------------------------

df = pd.read_csv(split_path)

print(
    "Total images:",
    len(df)
)

# -------------------------------------------------------------------------
# We use a compact perceptual hash.
#
# Images with identical/similar visual structure tend to have small
# Hamming distances even when their file bytes are different.
# -------------------------------------------------------------------------

def average_hash(path, hash_size=16):

    try:

        with Image.open(path) as img:

            img = img.convert("L")
            img = img.resize(
                (hash_size, hash_size)
            )

            pixels = np.asarray(
                img,
                dtype=np.float32
            )

        mean = pixels.mean()

        bits = pixels > mean

        return bits.flatten()

    except Exception:
        return None


print("\nComputing perceptual hashes...")

hashes = []

for path in tqdm(
    df["candidate_path"],
    total=len(df)
):

    hashes.append(
        average_hash(path)
    )

df["phash_bits"] = hashes

valid = df[
    df["phash_bits"].notna()
].copy()

print(
    "\nImages successfully hashed:",
    len(valid)
)

print(
    "Hash failures:",
    len(df) - len(valid)
)

# -------------------------------------------------------------------------
# Compare only cross-split images within the SAME CLASS.
#
# We don't want visually similar but legitimately different categories
# to be treated as leakage.
# -------------------------------------------------------------------------

def hamming_distance(a, b):

    return int(
        np.sum(a != b)
    )

# -------------------------------------------------------------------------
# To keep Colab runtime manageable, compare images inside each class.
# -------------------------------------------------------------------------

CANDIDATE_THRESHOLD = 12

near_pairs = []

print("\nSearching for close perceptual matches...")

for class_name in sorted(
    valid["target_class"].unique()
):

    class_df = valid[
        valid["target_class"] == class_name
    ].reset_index(drop=True)

    n = len(class_df)

    print(
        f"\n{class_name}: {n} images"
    )

    for i in range(n):

        hash_i = class_df.loc[
            i,
            "phash_bits"
        ]

        split_i = class_df.loc[
            i,
            "split"
        ]

        for j in range(i + 1, n):

            split_j = class_df.loc[
                j,
                "split"
            ]

            # Same split is not relevant for leakage.
            if split_i == split_j:
                continue

            hash_j = class_df.loc[
                j,
                "phash_bits"
            ]

            distance = hamming_distance(
                hash_i,
                hash_j
            )

            if distance <= CANDIDATE_THRESHOLD:

                near_pairs.append({
                    "class_name": class_name,
                    "distance": distance,
                    "split_a": split_i,
                    "split_b": split_j,
                    "file_a": class_df.loc[
                        i,
                        "candidate_path"
                    ],
                    "file_b": class_df.loc[
                        j,
                        "candidate_path"
                    ]
                })

# -------------------------------------------------------------------------
# Results
# -------------------------------------------------------------------------

near_df = pd.DataFrame(
    near_pairs
)

print("\n" + "=" * 80)
print("NEAR-DUPLICATE RESULTS")
print("=" * 80)

print(
    "Cross-split perceptually similar pairs:",
    len(near_df)
)

if len(near_df) > 0:

    print(
        "\nDistance distribution:"
    )

    print(
        near_df["distance"]
        .value_counts()
        .sort_index()
        .head(30)
    )

    print(
        "\nPairs with distance <= 6:"
    )

    very_close = near_df[
        near_df["distance"] <= 6
    ]

    print(
        "Count:",
        len(very_close)
    )

    if len(very_close) > 0:

        display(
            very_close[
                [
                    "class_name",
                    "distance",
                    "split_a",
                    "split_b",
                    "file_a",
                    "file_b"
                ]
            ].head(30)
        )

else:

    print(
        "✓ No cross-split perceptually similar pairs "
        "under the threshold."
    )

# -------------------------------------------------------------------------
# Save results
# -------------------------------------------------------------------------

near_path = os.path.join(
    BASE,
    "ECOBRIDGE_FINAL_6CLASS",
    "near_duplicate_cross_split_audit.csv"
)

near_df.to_csv(
    near_path,
    index=False
)

print(
    "\nSaved audit:",
    near_path
)

print("\nSTEP 99 COMPLETE")
```

```python
# =============================================================================
# STEP 100 — VISUAL AUDIT OF STRONG NEAR-DUPLICATES
# =============================================================================

import os
import math
import pandas as pd
import matplotlib.pyplot as plt
from PIL import Image

BASE = "/content/ECOBRIDGE_DATA"

audit_path = os.path.join(
    BASE,
    "ECOBRIDGE_FINAL_6CLASS",
    "near_duplicate_cross_split_audit.csv"
)

output_dir = os.path.join(
    BASE,
    "ECOBRIDGE_FINAL_6CLASS",
    "near_duplicate_contact_sheets"
)

os.makedirs(
    output_dir,
    exist_ok=True
)

near_df = pd.read_csv(audit_path)

# -------------------------------------------------------------------------
# Only the strongest potential matches
# -------------------------------------------------------------------------

strong = near_df[
    near_df["distance"] <= 6
].copy()

strong = strong.sort_values(
    ["distance", "class_name"]
).reset_index(drop=True)

print("=" * 80)
print("STEP 100 — STRONG NEAR-DUPLICATE VISUAL AUDIT")
print("=" * 80)

print(
    "Strong pairs (distance <= 6):",
    len(strong)
)

print("\nBy class:")
print(
    strong["class_name"]
    .value_counts()
)

print("\nBy distance:")
print(
    strong["distance"]
    .value_counts()
    .sort_index()
)

# -------------------------------------------------------------------------
# Create contact sheets in batches of 20 pairs.
# Each row = one pair.
# Left  = image A
# Right = image B
# -------------------------------------------------------------------------

BATCH_SIZE = 20

sheet_paths = []

for batch_start in range(
    0,
    len(strong),
    BATCH_SIZE
):

    batch = strong.iloc[
        batch_start:batch_start + BATCH_SIZE
    ]

    rows = len(batch)

    fig, axes = plt.subplots(
        rows,
        2,
        figsize=(10, max(3, rows * 2.2))
    )

    if rows == 1:
        axes = [axes]

    for row_idx, (_, record) in enumerate(
        batch.iterrows()
    ):

        path_a = record["file_a"]
        path_b = record["file_b"]

        try:

            img_a = Image.open(path_a).convert("RGB")

            axes[row_idx][0].imshow(img_a)
            axes[row_idx][0].set_title(
                f"{record['class_name']} | "
                f"d={record['distance']} | "
                f"{record['split_a']}\n"
                f"{os.path.basename(path_a)}",
                fontsize=8
            )
            axes[row_idx][0].axis("off")

        except Exception as e:

            axes[row_idx][0].text(
                0.5,
                0.5,
                f"ERROR\n{e}",
                ha="center",
                va="center"
            )
            axes[row_idx][0].axis("off")

        try:

            img_b = Image.open(path_b).convert("RGB")

            axes[row_idx][1].imshow(img_b)
            axes[row_idx][1].set_title(
                f"{record['class_name']} | "
                f"d={record['distance']} | "
                f"{record['split_b']}\n"
                f"{os.path.basename(path_b)}",
                fontsize=8
            )
            axes[row_idx][1].axis("off")

        except Exception as e:

            axes[row_idx][1].text(
                0.5,
                0.5,
                f"ERROR\n{e}",
                ha="center",
                va="center"
            )
            axes[row_idx][1].axis("off")

    plt.tight_layout()

    batch_number = (
        batch_start // BATCH_SIZE
    ) + 1

    output_path = os.path.join(
        output_dir,
        f"near_duplicate_batch_{batch_number}.png"
    )

    plt.savefig(
        output_path,
        dpi=120,
        bbox_inches="tight"
    )

    plt.show()

    plt.close(fig)

    sheet_paths.append(
        output_path
    )

print("\n" + "=" * 80)
print("CONTACT SHEETS CREATED")
print("=" * 80)

for path in sheet_paths:
    print(path)

print("\nSTEP 100 COMPLETE")
```

```python
# =============================================================================
# STEP 101 — BUILD FINAL TRAIN / VALIDATION / TEST DATASET
# =============================================================================

import os
import shutil
import pandas as pd
from tqdm.auto import tqdm

BASE = "/content/ECOBRIDGE_DATA"

SPLIT_PATH = os.path.join(
    BASE,
    "ECOBRIDGE_FINAL_6CLASS",
    "split_plan_duplicate_safe.csv"
)

FINAL_DIR = os.path.join(
    BASE,
    "ECOBRIDGE_FINAL_6CLASS"
)

CLASSES = [
    "CAT_BATTERY",
    "CAT_PCB",
    "CAT_CRT",
    "CAT_LCD_LED",
    "CAT_CABLES",
    "CAT_OTHER"
]

# -------------------------------------------------------------------------
# Load final split plan
# -------------------------------------------------------------------------

df = pd.read_csv(SPLIT_PATH)

print("=" * 80)
print("STEP 101 — BUILD FINAL DATASET")
print("=" * 80)

print(
    "Images in split plan:",
    len(df)
)

# -------------------------------------------------------------------------
# Clean old train/val/test directories
# -------------------------------------------------------------------------

for split in ["train", "val", "test"]:

    split_dir = os.path.join(
        FINAL_DIR,
        split
    )

    if os.path.exists(split_dir):

        shutil.rmtree(
            split_dir
        )

    os.makedirs(
        split_dir,
        exist_ok=True
    )

# -------------------------------------------------------------------------
# Create class directories
# -------------------------------------------------------------------------

for split in ["train", "val", "test"]:

    for class_name in CLASSES:

        os.makedirs(
            os.path.join(
                FINAL_DIR,
                split,
                class_name
            ),
            exist_ok=True
        )

# -------------------------------------------------------------------------
# Copy images
# -------------------------------------------------------------------------

copy_records = []

errors = []

for _, row in tqdm(
    df.iterrows(),
    total=len(df),
    desc="Copying images"
):

    source = row["candidate_path"]
    split = row["split"]
    class_name = row["target_class"]

    filename = os.path.basename(
        source
    )

    destination_dir = os.path.join(
        FINAL_DIR,
        split,
        class_name
    )

    destination = os.path.join(
        destination_dir,
        filename
    )

    # -------------------------------------------------------------
    # Avoid accidental filename collisions
    # -------------------------------------------------------------

    if os.path.exists(destination):

        base_name, extension = os.path.splitext(
            filename
        )

        destination = os.path.join(
            destination_dir,
            f"{row['sample_id']}_{base_name}{extension}"
        )

    try:

        shutil.copy2(
            source,
            destination
        )

        copy_records.append({
            "sample_id": row["sample_id"],
            "target_class": class_name,
            "split": split,
            "source_dataset": row["source"],
            "original_path": row["original_path"],
            "source_path": source,
            "final_path": destination,
            "sha256": row["sha256"]
        })

    except Exception as e:

        errors.append({
            "sample_id": row["sample_id"],
            "source": source,
            "destination": destination,
            "error": str(e)
        })

# -------------------------------------------------------------------------
# Save copy manifest
# -------------------------------------------------------------------------

copy_df = pd.DataFrame(
    copy_records
)

manifest_path = os.path.join(
    FINAL_DIR,
    "final_dataset_manifest.csv"
)

copy_df.to_csv(
    manifest_path,
    index=False
)

# -------------------------------------------------------------------------
# Save errors
# -------------------------------------------------------------------------

error_path = os.path.join(
    FINAL_DIR,
    "dataset_copy_errors.csv"
)

pd.DataFrame(
    errors
).to_csv(
    error_path,
    index=False
)

# -------------------------------------------------------------------------
# Final filesystem count
# -------------------------------------------------------------------------

print("\n" + "=" * 80)
print("COPY SUMMARY")
print("=" * 80)

print(
    "Expected images:",
    len(df)
)

print(
    "Successfully copied:",
    len(copy_df)
)

print(
    "Copy errors:",
    len(errors)
)

# -------------------------------------------------------------------------
# Count actual files in final folders
# -------------------------------------------------------------------------

print("\n" + "-" * 80)
print("FINAL DIRECTORY COUNTS")
print("-" * 80)

filesystem_counts = []

for split in ["train", "val", "test"]:

    for class_name in CLASSES:

        folder = os.path.join(
            FINAL_DIR,
            split,
            class_name
        )

        count = sum(
            1
            for f in os.listdir(folder)
            if f.lower().endswith(
                (".jpg", ".jpeg", ".png", ".webp", ".bmp")
            )
        )

        filesystem_counts.append({
            "split": split,
            "class": class_name,
            "count": count
        })

        print(
            f"{split:5} | "
            f"{class_name:15} | "
            f"{count}"
        )

counts_df = pd.DataFrame(
    filesystem_counts
)

# -------------------------------------------------------------------------
# Verify filesystem total
# -------------------------------------------------------------------------

actual_total = counts_df["count"].sum()

print("\n" + "=" * 80)
print("FINAL DATASET VERIFICATION")
print("=" * 80)

print(
    "Expected total:",
    len(df)
)

print(
    "Actual total:",
    actual_total
)

print(
    "Difference:",
    actual_total - len(df)
)

if (
    len(errors) == 0
    and actual_total == len(df)
):

    print(
        "\n✓ FINAL DATASET BUILT SUCCESSFULLY"
    )

else:

    print(
        "\n⚠️ DATASET BUILD NEEDS INVESTIGATION"
    )

print("\nManifest:")
print(manifest_path)

print("\nSTEP 101 COMPLETE")
```

```python
# =============================================================================
# STEP 102 — FINAL TENSORFLOW DATASET SANITY CHECK
# =============================================================================

import os
import numpy as np
import tensorflow as tf

BASE = "/content/ECOBRIDGE_DATA/ECOBRIDGE_FINAL_6CLASS"

CLASSES = [
    "CAT_BATTERY",
    "CAT_CABLES",
    "CAT_CRT",
    "CAT_LCD_LED",
    "CAT_OTHER",
    "CAT_PCB"
]

print("=" * 80)
print("STEP 102 — FINAL TENSORFLOW DATASET SANITY CHECK")
print("=" * 80)

# -------------------------------------------------------------------------
# TensorFlow version
# -------------------------------------------------------------------------

print("\nTensorFlow version:")
print(tf.__version__)

# -------------------------------------------------------------------------
# GPU check
# -------------------------------------------------------------------------

print("\nGPU devices:")
gpus = tf.config.list_physical_devices("GPU")

if gpus:
    for gpu in gpus:
        print(" ", gpu)
else:
    print("⚠️ No GPU detected")

# -------------------------------------------------------------------------
# Dataset directories
# -------------------------------------------------------------------------

train_dir = os.path.join(BASE, "train")
val_dir = os.path.join(BASE, "val")
test_dir = os.path.join(BASE, "test")

print("\nDataset directories:")
print("Train:", train_dir)
print("Val  :", val_dir)
print("Test :", test_dir)

# -------------------------------------------------------------------------
# Load datasets
# -------------------------------------------------------------------------

IMG_SIZE = (224, 224)
BATCH_SIZE = 32
SEED = 42

print("\nLoading training dataset...")

train_ds = tf.keras.utils.image_dataset_from_directory(
    train_dir,
    labels="inferred",
    label_mode="int",
    class_names=CLASSES,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    shuffle=True,
    seed=SEED
)

print("\nLoading validation dataset...")

val_ds = tf.keras.utils.image_dataset_from_directory(
    val_dir,
    labels="inferred",
    label_mode="int",
    class_names=CLASSES,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    shuffle=False
)

print("\nLoading test dataset...")

test_ds = tf.keras.utils.image_dataset_from_directory(
    test_dir,
    labels="inferred",
    label_mode="int",
    class_names=CLASSES,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    shuffle=False
)

# -------------------------------------------------------------------------
# Print class mapping
# -------------------------------------------------------------------------

print("\n" + "-" * 80)
print("CLASS MAPPING")
print("-" * 80)

for index, class_name in enumerate(
    train_ds.class_names
):

    print(
        f"{index} -> {class_name}"
    )

# -------------------------------------------------------------------------
# Count examples from TensorFlow datasets
# -------------------------------------------------------------------------

def count_dataset(ds):

    total = 0

    for images, labels in ds:
        total += images.shape[0]

    return total

print("\n" + "-" * 80)
print("TENSORFLOW DATASET COUNTS")
print("-" * 80)

train_count = count_dataset(train_ds)
val_count = count_dataset(val_ds)
test_count = count_dataset(test_ds)

print("Train:", train_count)
print("Val  :", val_count)
print("Test :", test_count)
print("Total:", train_count + val_count + test_count)

# -------------------------------------------------------------------------
# Inspect tensor shape and dtype
# -------------------------------------------------------------------------

print("\n" + "-" * 80)
print("BATCH INSPECTION")
print("-" * 80)

for images, labels in train_ds.take(1):

    print("Image tensor shape:", images.shape)
    print("Image dtype:", images.dtype)

    print(
        "Label tensor shape:",
        labels.shape
    )

    print(
        "Label dtype:",
        labels.dtype
    )

    print(
        "Pixel minimum:",
        float(tf.reduce_min(images))
    )

    print(
        "Pixel maximum:",
        float(tf.reduce_max(images))
    )

    print(
        "Sample labels:",
        labels[:10].numpy()
    )

# -------------------------------------------------------------------------
# Check for NaN / Inf
# -------------------------------------------------------------------------

print("\n" + "-" * 80)
print("NUMERICAL SANITY CHECK")
print("-" * 80)

nan_found = False
inf_found = False

for images, labels in train_ds:

    if bool(tf.reduce_any(tf.math.is_nan(images))):
        nan_found = True

    if bool(tf.reduce_any(tf.math.is_inf(images))):
        inf_found = True

print("NaN values found:", nan_found)
print("Inf values found:", inf_found)

# -------------------------------------------------------------------------
# Verify expected counts
# -------------------------------------------------------------------------

EXPECTED_TRAIN = 3470
EXPECTED_VAL = 418
EXPECTED_TEST = 448

print("\n" + "=" * 80)
print("EXPECTED COUNT VERIFICATION")
print("=" * 80)

print(
    "Train:",
    train_count,
    "| Expected:",
    EXPECTED_TRAIN,
    "| Match:",
    train_count == EXPECTED_TRAIN
)

print(
    "Val:",
    val_count,
    "| Expected:",
    EXPECTED_VAL,
    "| Match:",
    val_count == EXPECTED_VAL
)

print(
    "Test:",
    test_count,
    "| Expected:",
    EXPECTED_TEST,
    "| Match:",
    test_count == EXPECTED_TEST
)

print("\nSTEP 102 COMPLETE")
```

```python
# =============================================================================
# STEP 103 — ECOBRIDGE MOBILE-NET V2 TRAINING PIPELINE
# =============================================================================

import os
import numpy as np
import tensorflow as tf

from sklearn.utils.class_weight import compute_class_weight

print("=" * 80)
print("STEP 103 — ECOBRIDGE TRAINING PIPELINE")
print("=" * 80)

# =============================================================================
# 1. CONFIGURATION
# =============================================================================

BASE = "/content/ECOBRIDGE_DATA/ECOBRIDGE_FINAL_6CLASS"

TRAIN_DIR = os.path.join(BASE, "train")
VAL_DIR = os.path.join(BASE, "val")
TEST_DIR = os.path.join(BASE, "test")

CLASSES = [
    "CAT_BATTERY",
    "CAT_CABLES",
    "CAT_CRT",
    "CAT_LCD_LED",
    "CAT_OTHER",
    "CAT_PCB"
]

NUM_CLASSES = len(CLASSES)

IMG_SIZE = (224, 224)
BATCH_SIZE = 32
SEED = 42

print("\nConfiguration:")
print("Classes:", CLASSES)
print("Number of classes:", NUM_CLASSES)
print("Image size:", IMG_SIZE)
print("Batch size:", BATCH_SIZE)

# =============================================================================
# 2. LOAD DATASETS
# =============================================================================

print("\n" + "-" * 80)
print("LOADING DATASETS")
print("-" * 80)

train_ds = tf.keras.utils.image_dataset_from_directory(
    TRAIN_DIR,
    labels="inferred",
    label_mode="int",
    class_names=CLASSES,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    shuffle=True,
    seed=SEED
)

val_ds = tf.keras.utils.image_dataset_from_directory(
    VAL_DIR,
    labels="inferred",
    label_mode="int",
    class_names=CLASSES,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    shuffle=False
)

test_ds = tf.keras.utils.image_dataset_from_directory(
    TEST_DIR,
    labels="inferred",
    label_mode="int",
    class_names=CLASSES,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    shuffle=False
)

print("\nDatasets loaded successfully.")

# =============================================================================
# 3. DATA AUGMENTATION
# =============================================================================
#
# Augmentation is applied ONLY during training.
# Validation and test images remain unchanged.
#
# This helps reduce overfitting and improve robustness to real-world
# differences in orientation, lighting and framing.
# =============================================================================

data_augmentation = tf.keras.Sequential(
    [
        tf.keras.layers.RandomFlip(
            "horizontal"
        ),

        tf.keras.layers.RandomRotation(
            0.08
        ),

        tf.keras.layers.RandomZoom(
            0.10
        ),

        tf.keras.layers.RandomContrast(
            0.10
        )
    ],
    name="eco_augmentation"
)

print("\nData augmentation:")
print(data_augmentation)

# =============================================================================
# 4. MOBILE-NET V2 PREPROCESSING
# =============================================================================
#
# MobileNetV2 expects inputs preprocessed to approximately [-1, 1].
# =============================================================================

preprocess_input = tf.keras.applications.mobilenet_v2.preprocess_input

# =============================================================================
# 5. CLASS WEIGHTS
# =============================================================================
#
# Our dataset is intentionally imbalanced:
#
# OTHER       2100
# BATTERY     1090
# CABLES       554
# PCB          300
# LCD/LED      161
# CRT          131
#
# Class weighting prevents the large OTHER class from dominating
# the loss function.
# =============================================================================

class_counts = np.array([
    874,   # BATTERY
    443,   # CABLES
    105,   # CRT
    128,   # LCD_LED
    1680,  # OTHER
    240    # PCB
])

class_indices = np.arange(
    NUM_CLASSES
)

class_weights_array = compute_class_weight(
    class_weight="balanced",
    classes=class_indices,
    y=np.repeat(
        class_indices,
        class_counts
    )
)

class_weights = {
    int(i): float(weight)
    for i, weight in zip(
        class_indices,
        class_weights_array
    )
}

print("\n" + "-" * 80)
print("CLASS WEIGHTS")
print("-" * 80)

for index, class_name in enumerate(CLASSES):

    print(
        f"{index} | "
        f"{class_name:15} | "
        f"count={class_counts[index]:4d} | "
        f"weight={class_weights[index]:.4f}"
    )

# =============================================================================
# 6. BUILD PREPROCESSING PIPELINE
# =============================================================================

AUTOTUNE = tf.data.AUTOTUNE

def prepare_training(images, labels):

    images = data_augmentation(
        images,
        training=True
    )

    images = preprocess_input(
        images
    )

    return images, labels


def prepare_evaluation(images, labels):

    images = preprocess_input(
        images
    )

    return images, labels


train_prepared = (
    train_ds
    .map(
        prepare_training,
        num_parallel_calls=AUTOTUNE
    )
    .prefetch(AUTOTUNE)
)

val_prepared = (
    val_ds
    .map(
        prepare_evaluation,
        num_parallel_calls=AUTOTUNE
    )
    .prefetch(AUTOTUNE)
)

test_prepared = (
    test_ds
    .map(
        prepare_evaluation,
        num_parallel_calls=AUTOTUNE
    )
    .prefetch(AUTOTUNE)
)

print("\nPreprocessing pipelines created.")

# =============================================================================
# 7. LOAD PRETRAINED MOBILENETV2
# =============================================================================

print("\n" + "-" * 80)
print("LOADING PRETRAINED MOBILENETV2")
print("-" * 80)

base_model = tf.keras.applications.MobileNetV2(
    input_shape=(
        IMG_SIZE[0],
        IMG_SIZE[1],
        3
    ),
    include_top=False,
    weights="imagenet"
)

# Freeze backbone for initial training
base_model.trainable = False

print(
    "Base model:",
    base_model.name
)

print(
    "Base model trainable:",
    base_model.trainable
)

print(
    "Base model parameters:",
    base_model.count_params()
)

# =============================================================================
# 8. BUILD ECOBRIDGE CLASSIFIER
# =============================================================================

inputs = tf.keras.Input(
    shape=(
        IMG_SIZE[0],
        IMG_SIZE[1],
        3
    ),
    name="image"
)

x = base_model(
    inputs,
    training=False
)

x = tf.keras.layers.GlobalAveragePooling2D(
    name="global_average_pooling"
)(x)

x = tf.keras.layers.Dropout(
    0.30,
    name="dropout"
)(x)

outputs = tf.keras.layers.Dense(
    NUM_CLASSES,
    activation="softmax",
    name="classification_head"
)(x)

model = tf.keras.Model(
    inputs=inputs,
    outputs=outputs,
    name="ECOBRIDGE_MobileNetV2_6Class"
)

# =============================================================================
# 9. COMPILE
# =============================================================================

model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=1e-3
    ),
    loss=tf.keras.losses.SparseCategoricalCrossentropy(),
    metrics=[
        tf.keras.metrics.SparseCategoricalAccuracy(
            name="accuracy"
        )
    ]
)

# =============================================================================
# 10. MODEL SUMMARY
# =============================================================================

print("\n" + "=" * 80)
print("MODEL SUMMARY")
print("=" * 80)

model.summary()

print("\n" + "=" * 80)
print("STEP 103 COMPLETE")
print("=" * 80)
```

```python
# =============================================================================
# STEP 104 — ECOBRIDGE MODEL SMOKE TEST
# =============================================================================

import numpy as np
import tensorflow as tf

print("=" * 80)
print("STEP 104 — MODEL SMOKE TEST")
print("=" * 80)

# -------------------------------------------------------------------------
# 1. Take one prepared training batch
# -------------------------------------------------------------------------

print("\nLoading one training batch...")

images, labels = next(
    iter(train_prepared)
)

print(
    "Input batch shape:",
    images.shape
)

print(
    "Label batch shape:",
    labels.shape
)

# -------------------------------------------------------------------------
# 2. Check preprocessed pixel range
# -------------------------------------------------------------------------

print("\n" + "-" * 80)
print("PREPROCESSING CHECK")
print("-" * 80)

print(
    "Minimum pixel value:",
    float(tf.reduce_min(images))
)

print(
    "Maximum pixel value:",
    float(tf.reduce_max(images))
)

# MobileNetV2 preprocessing should produce values approximately [-1, 1].

# -------------------------------------------------------------------------
# 3. Run forward pass
# -------------------------------------------------------------------------

print("\n" + "-" * 80)
print("FORWARD PASS")
print("-" * 80)

predictions = model(
    images,
    training=False
)

print(
    "Prediction tensor shape:",
    predictions.shape
)

print(
    "Expected shape:",
    (images.shape[0], 6)
)

# -------------------------------------------------------------------------
# 4. Check probability properties
# -------------------------------------------------------------------------

prediction_sums = tf.reduce_sum(
    predictions,
    axis=1
)

print(
    "\nFirst 5 probability sums:"
)

print(
    prediction_sums[:5].numpy()
)

print(
    "\nMinimum probability:",
    float(tf.reduce_min(predictions))
)

print(
    "Maximum probability:",
    float(tf.reduce_max(predictions))
)

# -------------------------------------------------------------------------
# 5. Check for NaN / Inf
# -------------------------------------------------------------------------

has_nan = bool(
    tf.reduce_any(
        tf.math.is_nan(predictions)
    )
)

has_inf = bool(
    tf.reduce_any(
        tf.math.is_inf(predictions)
    )
)

print("\nNaN predictions:", has_nan)
print("Inf predictions:", has_inf)

# -------------------------------------------------------------------------
# 6. Display sample predictions
# -------------------------------------------------------------------------

print("\n" + "-" * 80)
print("SAMPLE PREDICTIONS")
print("-" * 80)

class_names = train_ds.class_names

predicted_indices = tf.argmax(
    predictions,
    axis=1
).numpy()

confidence_values = tf.reduce_max(
    predictions,
    axis=1
).numpy()

actual_indices = labels.numpy()

for i in range(
    min(10, len(predicted_indices))
):

    actual = class_names[
        actual_indices[i]
    ]

    predicted = class_names[
        predicted_indices[i]
    ]

    confidence = confidence_values[i]

    print(
        f"{i:02d} | "
        f"Actual: {actual:15} | "
        f"Predicted: {predicted:15} | "
        f"Confidence: {confidence:.4f}"
    )

# -------------------------------------------------------------------------
# 7. Trainable parameter verification
# -------------------------------------------------------------------------

print("\n" + "-" * 80)
print("TRAINABLE PARAMETER CHECK")
print("-" * 80)

print(
    "Total parameters:",
    model.count_params()
)

print(
    "Trainable parameters:",
    sum(
        np.prod(v.shape)
        for v in model.trainable_variables
    )
)

print(
    "Non-trainable parameters:",
    sum(
        np.prod(v.shape)
        for v in model.non_trainable_variables
    )
)

# -------------------------------------------------------------------------
# 8. Final status
# -------------------------------------------------------------------------

print("\n" + "=" * 80)
print("SMOKE TEST STATUS")
print("=" * 80)

shape_ok = (
    predictions.shape[-1] == 6
)

probability_ok = bool(
    tf.reduce_all(
        tf.abs(
            prediction_sums - 1.0
        ) < 1e-5
    )
)

numerical_ok = (
    not has_nan
    and not has_inf
)

if (
    shape_ok
    and probability_ok
    and numerical_ok
):

    print(
        "✓ MODEL SMOKE TEST PASSED"
    )

else:

    print(
        "⚠️ MODEL SMOKE TEST FAILED — DO NOT TRAIN YET"
    )

print("\nSTEP 104 COMPLETE")
```

```python
# =============================================================================
# STEP 105 — ECOBRIDGE INITIAL TRAINING
# =============================================================================

import os
import tensorflow as tf

print("=" * 80)
print("STEP 105 — INITIAL TRAINING")
print("=" * 80)

# -------------------------------------------------------------------------
# TRAINING CONFIGURATION
# -------------------------------------------------------------------------

EPOCHS_INITIAL = 15

CHECKPOINT_DIR = "/content/ECOBRIDGE_DATA/ECOBRIDGE_FINAL_6CLASS/checkpoints"
os.makedirs(CHECKPOINT_DIR, exist_ok=True)

BEST_MODEL_PATH = os.path.join(
    CHECKPOINT_DIR,
    "ecobridge_mobilenetv2_initial_best.keras"
)

print("\nTraining configuration:")
print("Epochs:", EPOCHS_INITIAL)
print("Batch size:", 32)
print("Learning rate:", 0.001)
print("Backbone trainable:", base_model.trainable)
print("Best model path:", BEST_MODEL_PATH)

# -------------------------------------------------------------------------
# CALLBACKS
# -------------------------------------------------------------------------

checkpoint_callback = tf.keras.callbacks.ModelCheckpoint(
    filepath=BEST_MODEL_PATH,
    monitor="val_accuracy",
    mode="max",
    save_best_only=True,
    verbose=1
)

early_stopping_callback = tf.keras.callbacks.EarlyStopping(
    monitor="val_loss",
    patience=4,
    mode="min",
    restore_best_weights=True,
    verbose=1
)

reduce_lr_callback = tf.keras.callbacks.ReduceLROnPlateau(
    monitor="val_loss",
    factor=0.3,
    patience=2,
    min_lr=1e-6,
    verbose=1
)

# -------------------------------------------------------------------------
# TRAIN MODEL
# -------------------------------------------------------------------------

print("\n" + "=" * 80)
print("STARTING INITIAL TRAINING")
print("=" * 80)

history_initial = model.fit(
    train_prepared,
    validation_data=val_prepared,
    epochs=EPOCHS_INITIAL,
    class_weight=class_weights,
    callbacks=[
        checkpoint_callback,
        early_stopping_callback,
        reduce_lr_callback
    ],
    verbose=1
)

# -------------------------------------------------------------------------
# TRAINING SUMMARY
# -------------------------------------------------------------------------

print("\n" + "=" * 80)
print("INITIAL TRAINING COMPLETE")
print("=" * 80)

best_epoch = (
    int(
        tf.argmax(
            history_initial.history["val_accuracy"]
        ).numpy()
    ) + 1
)

best_val_accuracy = max(
    history_initial.history["val_accuracy"]
)

best_val_loss = min(
    history_initial.history["val_loss"]
)

best_train_accuracy = (
    history_initial.history["accuracy"][best_epoch - 1]
)

best_train_loss = (
    history_initial.history["loss"][best_epoch - 1]
)

print(f"\nBest epoch: {best_epoch}")

print(
    f"Training accuracy at best epoch: "
    f"{best_train_accuracy:.4f}"
)

print(
    f"Validation accuracy at best epoch: "
    f"{best_val_accuracy:.4f}"
)

print(
    f"Training loss at best epoch: "
    f"{best_train_loss:.4f}"
)

print(
    f"Validation loss at best epoch: "
    f"{best_val_loss:.4f}"
)

print(
    f"\nBest checkpoint saved at:\n{BEST_MODEL_PATH}"
)

print("\nSTEP 105 COMPLETE")
```

```python
# =============================================================================
# STEP 106 — TRAINING CURVES + BEST CHECKPOINT VERIFICATION
# =============================================================================

import os
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

print("=" * 80)
print("STEP 106 — TRAINING CURVES + CHECKPOINT VERIFICATION")
print("=" * 80)

# -------------------------------------------------------------------------
# 1. Verify checkpoint exists
# -------------------------------------------------------------------------

print("\nChecking best checkpoint...")

if os.path.exists(BEST_MODEL_PATH):
    checkpoint_size_mb = (
        os.path.getsize(BEST_MODEL_PATH) / (1024 ** 2)
    )

    print("✓ Best checkpoint found")
    print(f"Checkpoint size: {checkpoint_size_mb:.2f} MB")
    print(f"Path: {BEST_MODEL_PATH}")

else:
    raise FileNotFoundError(
        f"Best checkpoint not found:\n{BEST_MODEL_PATH}"
    )

# -------------------------------------------------------------------------
# 2. Extract training history
# -------------------------------------------------------------------------

history_data = history_initial.history

train_accuracy = history_data["accuracy"]
val_accuracy = history_data["val_accuracy"]

train_loss = history_data["loss"]
val_loss = history_data["val_loss"]

epochs = range(1, len(train_accuracy) + 1)

# -------------------------------------------------------------------------
# 3. Find best epochs
# -------------------------------------------------------------------------

best_val_accuracy_epoch = (
    int(np.argmax(val_accuracy)) + 1
)

best_val_loss_epoch = (
    int(np.argmin(val_loss)) + 1
)

best_val_accuracy = max(val_accuracy)
best_val_loss = min(val_loss)

print("\n" + "-" * 80)
print("BEST TRAINING RESULTS")
print("-" * 80)

print(
    f"Best validation accuracy: "
    f"{best_val_accuracy:.4f} "
    f"({best_val_accuracy * 100:.2f}%)"
)

print(
    f"Best validation accuracy epoch: "
    f"{best_val_accuracy_epoch}"
)

print(
    f"Best validation loss: "
    f"{best_val_loss:.4f}"
)

print(
    f"Best validation loss epoch: "
    f"{best_val_loss_epoch}"
)

# -------------------------------------------------------------------------
# 4. Plot accuracy
# -------------------------------------------------------------------------

plt.figure(figsize=(10, 5))

plt.plot(
    epochs,
    train_accuracy,
    marker="o",
    label="Training Accuracy"
)

plt.plot(
    epochs,
    val_accuracy,
    marker="o",
    label="Validation Accuracy"
)

plt.axvline(
    best_val_accuracy_epoch,
    linestyle="--",
    label=f"Best Val Accuracy: Epoch {best_val_accuracy_epoch}"
)

plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.title("ECOBRIDGE — Training vs Validation Accuracy")
plt.legend()
plt.grid(True, alpha=0.3)

plt.show()

# -------------------------------------------------------------------------
# 5. Plot loss
# -------------------------------------------------------------------------

plt.figure(figsize=(10, 5))

plt.plot(
    epochs,
    train_loss,
    marker="o",
    label="Training Loss"
)

plt.plot(
    epochs,
    val_loss,
    marker="o",
    label="Validation Loss"
)

plt.axvline(
    best_val_loss_epoch,
    linestyle="--",
    label=f"Lowest Val Loss: Epoch {best_val_loss_epoch}"
)

plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("ECOBRIDGE — Training vs Validation Loss")
plt.legend()
plt.grid(True, alpha=0.3)

plt.show()

# -------------------------------------------------------------------------
# 6. Reload BEST VALIDATION-ACCURACY checkpoint
# -------------------------------------------------------------------------

print("\n" + "-" * 80)
print("RELOADING BEST VALIDATION-ACCURACY CHECKPOINT")
print("-" * 80)

best_model = tf.keras.models.load_model(
    BEST_MODEL_PATH
)

print("✓ Best checkpoint loaded successfully")

# -------------------------------------------------------------------------
# 7. Verify loaded model
# -------------------------------------------------------------------------

print("\nLoaded model name:")
print(best_model.name)

print(
    "\nLoaded model parameters:",
    best_model.count_params()
)

print(
    "Loaded model trainable parameters:",
    sum(
        np.prod(v.shape)
        for v in best_model.trainable_variables
    )
)

# -------------------------------------------------------------------------
# 8. Evaluate best checkpoint on validation set
# -------------------------------------------------------------------------

print("\n" + "-" * 80)
print("BEST CHECKPOINT — VALIDATION EVALUATION")
print("-" * 80)

val_results = best_model.evaluate(
    val_prepared,
    verbose=1
)

for metric_name, metric_value in zip(
    best_model.metrics_names,
    val_results
):
    print(
        f"{metric_name}: {metric_value:.4f}"
    )

print("\n" + "=" * 80)
print("STEP 106 COMPLETE")
print("=" * 80)
```

```python
# =============================================================================
# STEP 107 — ECOBRIDGE MOBILENETV2 FINE-TUNING
# =============================================================================

import os
import numpy as np
import tensorflow as tf

print("=" * 80)
print("STEP 107 — MOBILENETV2 FINE-TUNING")
print("=" * 80)

# -------------------------------------------------------------------------
# 1. Start from the verified best validation-accuracy checkpoint
# -------------------------------------------------------------------------

FINE_TUNE_MODEL_PATH = os.path.join(
    CHECKPOINT_DIR,
    "ecobridge_mobilenetv2_finetuned_best.keras"
)

fine_model = tf.keras.models.load_model(
    BEST_MODEL_PATH
)

print("\n✓ Starting from best initial checkpoint")
print(f"Initial checkpoint: {BEST_MODEL_PATH}")

# -------------------------------------------------------------------------
# 2. Locate MobileNetV2 backbone
# -------------------------------------------------------------------------

backbone = None

for layer in fine_model.layers:

    if isinstance(layer, tf.keras.Model):

        if "mobilenetv2" in layer.name.lower():

            backbone = layer
            break

if backbone is None:
    raise RuntimeError(
        "MobileNetV2 backbone could not be located."
    )

print("\nBackbone found:")
print(backbone.name)

print(
    "Total backbone layers:",
    len(backbone.layers)
)

# -------------------------------------------------------------------------
# 3. Freeze everything first
# -------------------------------------------------------------------------

backbone.trainable = True

# Number of final layers to fine-tune
FINE_TUNE_LAYERS = 30

freeze_until = max(
    0,
    len(backbone.layers) - FINE_TUNE_LAYERS
)

for i, layer in enumerate(backbone.layers):

    # Freeze earlier layers
    if i < freeze_until:

        layer.trainable = False

    else:

        layer.trainable = True

    # Keep BatchNormalization frozen
    if isinstance(
        layer,
        tf.keras.layers.BatchNormalization
    ):

        layer.trainable = False

# -------------------------------------------------------------------------
# 4. Display fine-tuning configuration
# -------------------------------------------------------------------------

trainable_backbone_layers = sum(
    1
    for layer in backbone.layers
    if layer.trainable
)

frozen_backbone_layers = sum(
    1
    for layer in backbone.layers
    if not layer.trainable
)

print("\n" + "-" * 80)
print("FINE-TUNING CONFIGURATION")
print("-" * 80)

print(
    "Backbone total layers:",
    len(backbone.layers)
)

print(
    "Backbone trainable layers:",
    trainable_backbone_layers
)

print(
    "Backbone frozen layers:",
    frozen_backbone_layers
)

print(
    "Fine-tuning final layers:",
    FINE_TUNE_LAYERS
)

print(
    "BatchNormalization layers kept frozen: YES"
)

# -------------------------------------------------------------------------
# 5. Recompile with a very small learning rate
# -------------------------------------------------------------------------

FINE_TUNE_LR = 1e-5

fine_model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=FINE_TUNE_LR
    ),
    loss=tf.keras.losses.SparseCategoricalCrossentropy(),
    metrics=[
        tf.keras.metrics.SparseCategoricalAccuracy(
            name="accuracy"
        )
    ]
)

print("\nLearning rate:", FINE_TUNE_LR)

# -------------------------------------------------------------------------
# 6. Count trainable parameters
# -------------------------------------------------------------------------

trainable_params = sum(
    np.prod(v.shape)
    for v in fine_model.trainable_variables
)

non_trainable_params = sum(
    np.prod(v.shape)
    for v in fine_model.non_trainable_variables
)

print(
    "Trainable parameters:",
    trainable_params
)

print(
    "Non-trainable parameters:",
    non_trainable_params
)

print(
    "Total parameters:",
    fine_model.count_params()
)

# -------------------------------------------------------------------------
# 7. Fine-tuning callbacks
# -------------------------------------------------------------------------

fine_checkpoint_callback = tf.keras.callbacks.ModelCheckpoint(
    filepath=FINE_TUNE_MODEL_PATH,
    monitor="val_accuracy",
    mode="max",
    save_best_only=True,
    verbose=1
)

fine_early_stopping = tf.keras.callbacks.EarlyStopping(
    monitor="val_loss",
    patience=3,
    mode="min",
    restore_best_weights=True,
    verbose=1
)

fine_reduce_lr = tf.keras.callbacks.ReduceLROnPlateau(
    monitor="val_loss",
    factor=0.3,
    patience=1,
    min_lr=1e-7,
    verbose=1
)

# -------------------------------------------------------------------------
# 8. Fine-tune
# -------------------------------------------------------------------------

print("\n" + "=" * 80)
print("STARTING FINE-TUNING")
print("=" * 80)

history_finetune = fine_model.fit(
    train_prepared,
    validation_data=val_prepared,
    epochs=8,
    class_weight=class_weights,
    callbacks=[
        fine_checkpoint_callback,
        fine_early_stopping,
        fine_reduce_lr
    ],
    verbose=1
)

# -------------------------------------------------------------------------
# 9. Training summary
# -------------------------------------------------------------------------

print("\n" + "=" * 80)
print("FINE-TUNING COMPLETE")
print("=" * 80)

best_ft_epoch = (
    int(
        np.argmax(
            history_finetune.history["val_accuracy"]
        )
    ) + 1
)

best_ft_val_accuracy = max(
    history_finetune.history["val_accuracy"]
)

best_ft_val_loss = min(
    history_finetune.history["val_loss"]
)

print(
    f"\nBest fine-tuning epoch: {best_ft_epoch}"
)

print(
    f"Best fine-tuning validation accuracy: "
    f"{best_ft_val_accuracy:.4f} "
    f"({best_ft_val_accuracy * 100:.2f}%)"
)

print(
    f"Best fine-tuning validation loss: "
    f"{best_ft_val_loss:.4f}"
)

print(
    f"\nFine-tuned checkpoint:"
)

print(FINE_TUNE_MODEL_PATH)

print("\nSTEP 107 COMPLETE")
```

```python
# =============================================================================
# STEP 108 — FINAL HELD-OUT TEST EVALUATION
# =============================================================================

import os
import numpy as np
import tensorflow as tf

print("=" * 80)
print("STEP 108 — FINAL HELD-OUT TEST EVALUATION")
print("=" * 80)

# -------------------------------------------------------------------------
# 1. Load the BEST fine-tuned checkpoint
# -------------------------------------------------------------------------

FINAL_MODEL_PATH = FINE_TUNE_MODEL_PATH

if not os.path.exists(FINAL_MODEL_PATH):
    raise FileNotFoundError(
        f"Fine-tuned checkpoint not found:\n{FINAL_MODEL_PATH}"
    )

final_model = tf.keras.models.load_model(
    FINAL_MODEL_PATH
)

print("\n✓ Final fine-tuned model loaded")
print("Model:", final_model.name)
print("Path:", FINAL_MODEL_PATH)

# -------------------------------------------------------------------------
# 2. Verify model parameters
# -------------------------------------------------------------------------

print("\n" + "-" * 80)
print("MODEL INFORMATION")
print("-" * 80)

print(
    "Total parameters:",
    final_model.count_params()
)

print(
    "Trainable parameters:",
    sum(
        np.prod(v.shape)
        for v in final_model.trainable_variables
    )
)

print(
    "Non-trainable parameters:",
    sum(
        np.prod(v.shape)
        for v in final_model.non_trainable_variables
    )
)

# -------------------------------------------------------------------------
# 3. Verify test dataset size
# -------------------------------------------------------------------------

print("\n" + "-" * 80)
print("TEST DATASET")
print("-" * 80)

print("Expected test images: 448")

# Count test batches/images
test_image_count = 0

for batch_images, batch_labels in test_prepared:
    test_image_count += batch_images.shape[0]

print("Actual test images:", test_image_count)

if test_image_count != 448:
    raise ValueError(
        f"Expected 448 test images, found {test_image_count}"
    )

print("✓ Test set size verified")

# -------------------------------------------------------------------------
# 4. Evaluate on untouched test set
# -------------------------------------------------------------------------

print("\n" + "=" * 80)
print("EVALUATING FINAL MODEL ON TEST SET")
print("=" * 80)

test_results = final_model.evaluate(
    test_prepared,
    verbose=1
)

print("\n" + "-" * 80)
print("FINAL TEST RESULTS")
print("-" * 80)

for metric_name, metric_value in zip(
    final_model.metrics_names,
    test_results
):
    print(
        f"{metric_name}: {metric_value:.4f}"
    )

# -------------------------------------------------------------------------
# 5. Extract basic metrics
# -------------------------------------------------------------------------

test_loss = float(test_results[0])
test_accuracy = float(test_results[1])

print("\n" + "=" * 80)
print("HELD-OUT TEST SUMMARY")
print("=" * 80)

print(
    f"Test Loss:     {test_loss:.4f}"
)

print(
    f"Test Accuracy: {test_accuracy:.4f} "
    f"({test_accuracy * 100:.2f}%)"
)

print(
    "\nTest samples evaluated:",
    test_image_count
)

print(
    "\nIMPORTANT:"
)

print(
    "This is the first evaluation on the untouched test set."
)

print("\nSTEP 108 COMPLETE")
```

```python
# =============================================================================
# STEP 109 — ECOBRIDGE DETAILED TEST EVALUATION
# =============================================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf

from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

print("=" * 80)
print("STEP 109 — DETAILED TEST EVALUATION")
print("=" * 80)

# -------------------------------------------------------------------------
# 1. Class names
# -------------------------------------------------------------------------

class_names = [
    "CAT_BATTERY",
    "CAT_CABLES",
    "CAT_CRT",
    "CAT_LCD_LED",
    "CAT_OTHER",
    "CAT_PCB"
]

print("\nClasses:")
for i, name in enumerate(class_names):
    print(f"{i} | {name}")

# -------------------------------------------------------------------------
# 2. Generate predictions on the COMPLETE test set
# -------------------------------------------------------------------------

print("\n" + "-" * 80)
print("GENERATING TEST PREDICTIONS")
print("-" * 80)

y_true = []
y_pred = []
y_probabilities = []

for images, labels in test_prepared:

    predictions = final_model.predict(
        images,
        verbose=0
    )

    predicted_classes = np.argmax(
        predictions,
        axis=1
    )

    y_true.extend(
        labels.numpy()
    )

    y_pred.extend(
        predicted_classes
    )

    y_probabilities.extend(
        predictions
    )

y_true = np.array(y_true)
y_pred = np.array(y_pred)
y_probabilities = np.array(y_probabilities)

print(
    "Ground-truth samples:",
    len(y_true)
)

print(
    "Predicted samples:",
    len(y_pred)
)

print(
    "Probability matrix shape:",
    y_probabilities.shape
)

# -------------------------------------------------------------------------
# 3. Verify prediction count
# -------------------------------------------------------------------------

if len(y_true) != 448:
    raise ValueError(
        f"Expected 448 labels, found {len(y_true)}"
    )

if len(y_pred) != 448:
    raise ValueError(
        f"Expected 448 predictions, found {len(y_pred)}"
    )

print("\n✓ All 448 test images have predictions")

# -------------------------------------------------------------------------
# 4. Overall metrics
# -------------------------------------------------------------------------

accuracy = accuracy_score(
    y_true,
    y_pred
)

precision_macro = precision_score(
    y_true,
    y_pred,
    average="macro",
    zero_division=0
)

recall_macro = recall_score(
    y_true,
    y_pred,
    average="macro",
    zero_division=0
)

f1_macro = f1_score(
    y_true,
    y_pred,
    average="macro",
    zero_division=0
)

precision_weighted = precision_score(
    y_true,
    y_pred,
    average="weighted",
    zero_division=0
)

recall_weighted = recall_score(
    y_true,
    y_pred,
    average="weighted",
    zero_division=0
)

f1_weighted = f1_score(
    y_true,
    y_pred,
    average="weighted",
    zero_division=0
)

print("\n" + "=" * 80)
print("OVERALL TEST METRICS")
print("=" * 80)

print(
    f"Accuracy:           {accuracy:.4f} "
    f"({accuracy * 100:.2f}%)"
)

print(
    f"Macro Precision:    {precision_macro:.4f}"
)

print(
    f"Macro Recall:       {recall_macro:.4f}"
)

print(
    f"Macro F1-score:     {f1_macro:.4f}"
)

print(
    f"Weighted Precision: {precision_weighted:.4f}"
)

print(
    f"Weighted Recall:    {recall_weighted:.4f}"
)

print(
    f"Weighted F1-score:  {f1_weighted:.4f}"
)

# -------------------------------------------------------------------------
# 5. Classification report
# -------------------------------------------------------------------------

print("\n" + "=" * 80)
print("PER-CLASS CLASSIFICATION REPORT")
print("=" * 80)

report = classification_report(
    y_true,
    y_pred,
    labels=np.arange(len(class_names)),
    target_names=class_names,
    digits=4,
    zero_division=0
)

print(report)

# -------------------------------------------------------------------------
# 6. Store classification report as DataFrame
# -------------------------------------------------------------------------

report_dict = classification_report(
    y_true,
    y_pred,
    labels=np.arange(len(class_names)),
    target_names=class_names,
    output_dict=True,
    zero_division=0
)

report_df = pd.DataFrame(
    report_dict
).transpose()

print("\n" + "-" * 80)
print("CLASSIFICATION REPORT TABLE")
print("-" * 80)

display(
    report_df.round(4)
)

# -------------------------------------------------------------------------
# 7. Confusion matrix
# -------------------------------------------------------------------------

cm = confusion_matrix(
    y_true,
    y_pred,
    labels=np.arange(len(class_names))
)

print("\n" + "=" * 80)
print("CONFUSION MATRIX")
print("=" * 80)

print(
    "Rows = Actual class"
)

print(
    "Columns = Predicted class"
)

print("\nClass order:")
print(class_names)

print("\nConfusion Matrix:")
print(cm)

# -------------------------------------------------------------------------
# 8. Verify confusion matrix total
# -------------------------------------------------------------------------

print(
    "\nConfusion matrix total:",
    cm.sum()
)

if cm.sum() != 448:
    raise ValueError(
        f"Confusion matrix total should be 448, found {cm.sum()}"
    )

print("✓ Confusion matrix verified")

# -------------------------------------------------------------------------
# 9. Plot confusion matrix
# -------------------------------------------------------------------------

plt.figure(figsize=(10, 8))

plt.imshow(
    cm,
    interpolation="nearest"
)

plt.title(
    "ECOBRIDGE — Test Set Confusion Matrix"
)

plt.colorbar()

tick_marks = np.arange(
    len(class_names)
)

plt.xticks(
    tick_marks,
    class_names,
    rotation=45,
    ha="right"
)

plt.yticks(
    tick_marks,
    class_names
)

# Add numbers inside cells
threshold = cm.max() / 2.0

for i in range(cm.shape[0]):

    for j in range(cm.shape[1]):

        plt.text(
            j,
            i,
            str(cm[i, j]),
            horizontalalignment="center",
            verticalalignment="center",
            color="white" if cm[i, j] > threshold else "black",
            fontsize=12
        )

plt.ylabel(
    "Actual Class"
)

plt.xlabel(
    "Predicted Class"
)

plt.tight_layout()

plt.show()

# -------------------------------------------------------------------------
# 10. Per-class correct / incorrect counts
# -------------------------------------------------------------------------

print("\n" + "=" * 80)
print("PER-CLASS TEST COUNTS")
print("=" * 80)

for i, class_name in enumerate(class_names):

    total_actual = np.sum(
        y_true == i
    )

    correct = cm[i, i]

    incorrect = (
        total_actual - correct
    )

    class_recall = (
        correct / total_actual
        if total_actual > 0
        else 0
    )

    print(
        f"{class_name:15} | "
        f"Total={total_actual:3d} | "
        f"Correct={correct:3d} | "
        f"Incorrect={incorrect:3d} | "
        f"Recall={class_recall:.4f}"
    )

# -------------------------------------------------------------------------
# 11. Save detailed predictions for later analysis
# -------------------------------------------------------------------------

prediction_confidence = np.max(
    y_probabilities,
    axis=1
)

prediction_df = pd.DataFrame({
    "true_class_id": y_true,
    "true_class": [
        class_names[i]
        for i in y_true
    ],
    "predicted_class_id": y_pred,
    "predicted_class": [
        class_names[i]
        for i in y_pred
    ],
    "confidence": prediction_confidence
})

PREDICTION_CSV = (
    "/content/ECOBRIDGE_DATA/"
    "ECOBRIDGE_FINAL_6CLASS/"
    "test_predictions.csv"
)

prediction_df.to_csv(
    PREDICTION_CSV,
    index=False
)

print(
    f"\n✓ Test predictions saved to:\n{PREDICTION_CSV}"
)

# -------------------------------------------------------------------------
# 12. Final status
# -------------------------------------------------------------------------

print("\n" + "=" * 80)
print("STEP 109 COMPLETE")
print("=" * 80)

print(
    f"Final Test Accuracy: {accuracy * 100:.2f}%"
)

print(
    f"Macro F1-score: {f1_macro:.4f}"
)

print(
    f"Weighted F1-score: {f1_weighted:.4f}"
)

print(
    "Test samples: 448"
)
```

```python
# =============================================================================
# STEP 110 — ECOBRIDGE CONFIDENCE-AWARE PREDICTION ANALYSIS
# =============================================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print("=" * 80)
print("STEP 110 — CONFIDENCE-AWARE PREDICTION ANALYSIS")
print("=" * 80)

# -------------------------------------------------------------------------
# 1. Generate predictions on VALIDATION set
# -------------------------------------------------------------------------

print("\nGenerating validation predictions...")

val_true = []
val_pred = []
val_probs = []

for images, labels in val_prepared:

    predictions = final_model.predict(
        images,
        verbose=0
    )

    predicted_classes = np.argmax(
        predictions,
        axis=1
    )

    val_true.extend(
        labels.numpy()
    )

    val_pred.extend(
        predicted_classes
    )

    val_probs.extend(
        predictions
    )

val_true = np.array(val_true)
val_pred = np.array(val_pred)
val_probs = np.array(val_probs)

print(
    "Validation samples:",
    len(val_true)
)

# -------------------------------------------------------------------------
# 2. Calculate confidence
# -------------------------------------------------------------------------

val_confidence = np.max(
    val_probs,
    axis=1
)

val_correct = (
    val_true == val_pred
)

print("\n" + "-" * 80)
print("CONFIDENCE SUMMARY")
print("-" * 80)

print(
    f"Minimum confidence: "
    f"{val_confidence.min():.4f}"
)

print(
    f"Maximum confidence: "
    f"{val_confidence.max():.4f}"
)

print(
    f"Mean confidence: "
    f"{val_confidence.mean():.4f}"
)

print(
    f"Median confidence: "
    f"{np.median(val_confidence):.4f}"
)

# -------------------------------------------------------------------------
# 3. Confidence bins
# -------------------------------------------------------------------------

print("\n" + "-" * 80)
print("CONFIDENCE VS CORRECTNESS")
print("-" * 80)

confidence_bins = [
    (0.00, 0.50),
    (0.50, 0.60),
    (0.60, 0.70),
    (0.70, 0.80),
    (0.80, 0.90),
    (0.90, 1.01)
]

confidence_rows = []

for lower, upper in confidence_bins:

    mask = (
        (val_confidence >= lower)
        & (val_confidence < upper)
    )

    count = int(
        np.sum(mask)
    )

    if count > 0:

        accuracy = float(
            np.mean(
                val_correct[mask]
            )
        )

        avg_conf = float(
            np.mean(
                val_confidence[mask]
            )
        )

    else:

        accuracy = np.nan
        avg_conf = np.nan

    confidence_rows.append({
        "confidence_range":
            f"{lower:.0%}–{min(upper, 1.0):.0%}",
        "samples": count,
        "accuracy": accuracy,
        "avg_confidence": avg_conf
    })

confidence_df = pd.DataFrame(
    confidence_rows
)

display(
    confidence_df.round(4)
)

# -------------------------------------------------------------------------
# 4. Candidate manual-verification thresholds
# -------------------------------------------------------------------------

print("\n" + "-" * 80)
print("CANDIDATE CONFIDENCE THRESHOLDS")
print("-" * 80)

thresholds = [
    0.60,
    0.70,
    0.80,
    0.90
]

threshold_rows = []

for threshold in thresholds:

    high_conf_mask = (
        val_confidence >= threshold
    )

    low_conf_mask = (
        val_confidence < threshold
    )

    high_count = int(
        np.sum(high_conf_mask)
    )

    low_count = int(
        np.sum(low_conf_mask)
    )

    if high_count > 0:

        high_conf_accuracy = float(
            np.mean(
                val_correct[high_conf_mask]
            )
        )

    else:

        high_conf_accuracy = np.nan

    threshold_rows.append({
        "threshold": threshold,
        "auto_accept_candidates": high_count,
        "manual_review_candidates": low_count,
        "high_conf_accuracy": high_conf_accuracy
    })

threshold_df = pd.DataFrame(
    threshold_rows
)

display(
    threshold_df.round(4)
)

# -------------------------------------------------------------------------
# 5. Analyze incorrect predictions
# -------------------------------------------------------------------------

print("\n" + "-" * 80)
print("INCORRECT VALIDATION PREDICTIONS")
print("-" * 80)

incorrect_mask = ~val_correct

incorrect_confidences = (
    val_confidence[incorrect_mask]
)

print(
    "Incorrect predictions:",
    int(np.sum(incorrect_mask))
)

if len(incorrect_confidences) > 0:

    print(
        f"Mean confidence on incorrect predictions: "
        f"{incorrect_confidences.mean():.4f}"
    )

    print(
        f"Maximum confidence on incorrect predictions: "
        f"{incorrect_confidences.max():.4f}"
    )

    print(
        f"Median confidence on incorrect predictions: "
        f"{np.median(incorrect_confidences):.4f}"
    )

# -------------------------------------------------------------------------
# 6. Most uncertain validation predictions
# -------------------------------------------------------------------------

print("\n" + "-" * 80)
print("10 MOST UNCERTAIN VALIDATION PREDICTIONS")
print("-" * 80)

uncertain_indices = np.argsort(
    val_confidence
)[:10]

for idx in uncertain_indices:

    actual = class_names[
        val_true[idx]
    ]

    predicted = class_names[
        val_pred[idx]
    ]

    confidence = val_confidence[idx]

    status = (
        "CORRECT"
        if val_true[idx] == val_pred[idx]
        else "INCORRECT"
    )

    print(
        f"{idx:03d} | "
        f"Actual: {actual:15} | "
        f"Predicted: {predicted:15} | "
        f"Confidence: {confidence:.4f} | "
        f"{status}"
    )

# -------------------------------------------------------------------------
# 7. Plot confidence distribution
# -------------------------------------------------------------------------

plt.figure(figsize=(10, 5))

plt.hist(
    val_confidence,
    bins=20
)

plt.xlabel(
    "Prediction Confidence"
)

plt.ylabel(
    "Number of Validation Images"
)

plt.title(
    "ECOBRIDGE — Validation Prediction Confidence Distribution"
)

plt.grid(
    True,
    alpha=0.3
)

plt.show()

# -------------------------------------------------------------------------
# 8. Save confidence analysis
# -------------------------------------------------------------------------

confidence_analysis_path = (
    "/content/ECOBRIDGE_DATA/"
    "ECOBRIDGE_FINAL_6CLASS/"
    "validation_confidence_analysis.csv"
)

validation_confidence_df = pd.DataFrame({
    "actual_class": [
        class_names[i]
        for i in val_true
    ],
    "predicted_class": [
        class_names[i]
        for i in val_pred
    ],
    "confidence": val_confidence,
    "correct": val_correct
})

validation_confidence_df.to_csv(
    confidence_analysis_path,
    index=False
)

print(
    f"\n✓ Confidence analysis saved to:\n"
    f"{confidence_analysis_path}"
)

# -------------------------------------------------------------------------
# 9. Final status
# -------------------------------------------------------------------------

print("\n" + "=" * 80)
print("STEP 110 COMPLETE")
print("=" * 80)

print(
    "Confidence-aware validation analysis completed."
)

print(
    "Test set was NOT used to select the confidence threshold."
)
```

```python
# =============================================================================
# STEP 111 — ECOBRIDGE REAL-TIME IMAGE INFERENCE
# =============================================================================

import os
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt

from google.colab import files

print("=" * 80)
print("STEP 111 — ECOBRIDGE IMAGE INFERENCE")
print("=" * 80)

# -------------------------------------------------------------------------
# CONFIGURATION
# -------------------------------------------------------------------------

CONFIDENCE_THRESHOLD = 0.90

CLASS_NAMES = [
    "CAT_BATTERY",
    "CAT_CABLES",
    "CAT_CRT",
    "CAT_LCD_LED",
    "CAT_OTHER",
    "CAT_PCB"
]

DISPLAY_NAMES = {
    "CAT_BATTERY": "Battery",
    "CAT_CABLES": "Cables",
    "CAT_CRT": "CRT",
    "CAT_LCD_LED": "LCD / LED",
    "CAT_OTHER": "Other E-waste",
    "CAT_PCB": "PCB"
}

# -------------------------------------------------------------------------
# UPLOAD IMAGE
# -------------------------------------------------------------------------

print("\nUpload an e-waste image for ECOBRIDGE classification.")

uploaded = files.upload()

if len(uploaded) == 0:
    raise ValueError("No image was uploaded.")

image_path = list(uploaded.keys())[0]

print(
    f"\n✓ Image received: {image_path}"
)

# -------------------------------------------------------------------------
# LOAD IMAGE
# -------------------------------------------------------------------------

image = tf.keras.utils.load_img(
    image_path,
    target_size=(224, 224)
)

image_array = tf.keras.utils.img_to_array(
    image
)

# Add batch dimension
image_batch = np.expand_dims(
    image_array,
    axis=0
)

# -------------------------------------------------------------------------
# MOBILENETV2 PREPROCESSING
# -------------------------------------------------------------------------

image_preprocessed = tf.keras.applications.mobilenet_v2.preprocess_input(
    image_batch
)

# -------------------------------------------------------------------------
# MODEL PREDICTION
# -------------------------------------------------------------------------

probabilities = final_model.predict(
    image_preprocessed,
    verbose=0
)[0]

predicted_index = int(
    np.argmax(probabilities)
)

predicted_class = CLASS_NAMES[
    predicted_index
]

predicted_display_name = DISPLAY_NAMES[
    predicted_class
]

confidence = float(
    probabilities[predicted_index]
)

# -------------------------------------------------------------------------
# TOP-3 PREDICTIONS
# -------------------------------------------------------------------------

top_indices = np.argsort(
    probabilities
)[::-1][:3]

print("\n" + "=" * 80)
print("ECOBRIDGE AI RESULT")
print("=" * 80)

print(
    f"\nPredicted Category : "
    f"{predicted_display_name}"
)

print(
    f"Confidence         : "
    f"{confidence * 100:.2f}%"
)

# -------------------------------------------------------------------------
# CONFIDENCE-AWARE STATUS
# -------------------------------------------------------------------------

if confidence >= CONFIDENCE_THRESHOLD:

    verification_status = (
        "HIGH-CONFIDENCE AI SUGGESTION"
    )

    verification_message = (
        "AI confidence is above the prototype threshold. "
        "Collector/recycler verification should still be completed."
    )

else:

    verification_status = (
        "MANUAL VERIFICATION REQUIRED"
    )

    verification_message = (
        "AI confidence is below the prototype threshold. "
        "Please verify the material manually."
    )

print(
    f"Verification Status : "
    f"{verification_status}"
)

print(
    f"\n{verification_message}"
)

# -------------------------------------------------------------------------
# TOP-3 RESULTS
# -------------------------------------------------------------------------

print("\n" + "-" * 80)
print("TOP-3 AI SUGGESTIONS")
print("-" * 80)

for rank, index in enumerate(
    top_indices,
    start=1
):

    class_name = CLASS_NAMES[index]

    display_name = DISPLAY_NAMES[
        class_name
    ]

    probability = probabilities[index]

    print(
        f"{rank}. "
        f"{display_name:18} "
        f"{probability * 100:6.2f}%"
    )

# -------------------------------------------------------------------------
# ALL CLASS PROBABILITIES
# -------------------------------------------------------------------------

print("\n" + "-" * 80)
print("ALL CLASS PROBABILITIES")
print("-" * 80)

for i, class_name in enumerate(
    CLASS_NAMES
):

    print(
        f"{DISPLAY_NAMES[class_name]:18} : "
        f"{probabilities[i] * 100:6.2f}%"
    )

# -------------------------------------------------------------------------
# DISPLAY IMAGE
# -------------------------------------------------------------------------

plt.figure(figsize=(7, 7))

plt.imshow(
    image
)

plt.axis("off")

plt.title(
    f"ECOBRIDGE Prediction\n"
    f"{predicted_display_name} — "
    f"{confidence * 100:.2f}%"
)

plt.show()

# -------------------------------------------------------------------------
# FINAL DEMO MESSAGE
# -------------------------------------------------------------------------

print("\n" + "=" * 80)
print("ECOBRIDGE DEMO OUTPUT")
print("=" * 80)

print(
    f"Material: {predicted_display_name}"
)

print(
    f"AI Confidence: {confidence * 100:.2f}%"
)

print(
    f"Status: {verification_status}"
)

print(
    "\nNote:"
)

print(
    "The AI provides a classification suggestion."
)

print(
    "Final transaction value must be based on verified "
    "weight and agreed recycler price."
)

print("\nSTEP 111 COMPLETE")
```

```python
# =============================================================================
# STEP 112 — ECOBRIDGE MULTI-CLASS REAL IMAGE DEMO TEST
# =============================================================================

import os
import numpy as np
import pandas as pd
import tensorflow as tf
import matplotlib.pyplot as plt

from google.colab import files

print("=" * 80)
print("STEP 112 — ECOBRIDGE MULTI-CLASS REAL IMAGE DEMO TEST")
print("=" * 80)

CLASS_NAMES = [
    "CAT_BATTERY",
    "CAT_CABLES",
    "CAT_CRT",
    "CAT_LCD_LED",
    "CAT_OTHER",
    "CAT_PCB"
]

DISPLAY_NAMES = {
    "CAT_BATTERY": "Battery",
    "CAT_CABLES": "Cables",
    "CAT_CRT": "CRT",
    "CAT_LCD_LED": "LCD / LED",
    "CAT_OTHER": "Other E-waste",
    "CAT_PCB": "PCB"
}

CONFIDENCE_THRESHOLD = 0.90

# -------------------------------------------------------------------------
# Upload multiple images
# -------------------------------------------------------------------------

print("\nUpload up to 6 real e-waste images.")
print("Ideally provide one image for each category.")
print("\nRecommended order:")
print("1. Battery")
print("2. Cables")
print("3. CRT")
print("4. LCD / LED")
print("5. Other E-waste")
print("6. PCB")

uploaded = files.upload()

if len(uploaded) == 0:
    raise ValueError("No images were uploaded.")

print(
    f"\n✓ {len(uploaded)} image(s) received."
)

# -------------------------------------------------------------------------
# Prediction function
# -------------------------------------------------------------------------

def predict_ecobridge_image(image_path):

    image = tf.keras.utils.load_img(
        image_path,
        target_size=(224, 224)
    )

    image_array = tf.keras.utils.img_to_array(
        image
    )

    image_batch = np.expand_dims(
        image_array,
        axis=0
    )

    image_preprocessed = (
        tf.keras.applications.mobilenet_v2.preprocess_input(
            image_batch
        )
    )

    probabilities = final_model.predict(
        image_preprocessed,
        verbose=0
    )[0]

    predicted_index = int(
        np.argmax(probabilities)
    )

    predicted_class = CLASS_NAMES[
        predicted_index
    ]

    confidence = float(
        probabilities[predicted_index]
    )

    if confidence >= CONFIDENCE_THRESHOLD:
        status = "HIGH-CONFIDENCE AI SUGGESTION"
    else:
        status = "MANUAL VERIFICATION REQUIRED"

    return (
        predicted_class,
        confidence,
        probabilities,
        status
    )

# -------------------------------------------------------------------------
# Run predictions
# -------------------------------------------------------------------------

results = []

print("\n" + "=" * 80)
print("PREDICTION RESULTS")
print("=" * 80)

for filename in uploaded.keys():

    (
        predicted_class,
        confidence,
        probabilities,
        status
    ) = predict_ecobridge_image(
        filename
    )

    results.append({
        "image": filename,
        "predicted_class": predicted_class,
        "predicted_category":
            DISPLAY_NAMES[predicted_class],
        "confidence": confidence,
        "confidence_percent":
            confidence * 100,
        "status": status
    })

    print("\n" + "-" * 80)

    print(
        f"Image      : {filename}"
    )

    print(
        f"Prediction : "
        f"{DISPLAY_NAMES[predicted_class]}"
    )

    print(
        f"Confidence : "
        f"{confidence * 100:.2f}%"
    )

    print(
        f"Status     : {status}"
    )

# -------------------------------------------------------------------------
# Results table
# -------------------------------------------------------------------------

results_df = pd.DataFrame(
    results
)

print("\n" + "=" * 80)
print("SUMMARY TABLE")
print("=" * 80)

display(
    results_df[
        [
            "image",
            "predicted_category",
            "confidence_percent",
            "status"
        ]
    ].round(2)
)

# -------------------------------------------------------------------------
# Save demo results
# -------------------------------------------------------------------------

demo_results_path = (
    "/content/ECOBRIDGE_DATA/"
    "ECOBRIDGE_FINAL_6CLASS/"
    "real_image_demo_results.csv"
)

results_df.to_csv(
    demo_results_path,
    index=False
)

print(
    f"\n✓ Demo results saved to:\n"
    f"{demo_results_path}"
)

print("\n" + "=" * 80)
print("STEP 112 COMPLETE")
print("=" * 80)
```

```python
# =============================================================================
# STEP 113 — FINAL ECOBRIDGE AI INFERENCE ENGINE
# =============================================================================

import numpy as np
import tensorflow as tf

print("=" * 80)
print("STEP 113 — FINAL ECOBRIDGE AI INFERENCE ENGINE")
print("=" * 80)

# -------------------------------------------------------------------------
# CONFIGURATION
# -------------------------------------------------------------------------

ECOBRIDGE_CONFIDENCE_THRESHOLD = 0.90

ECOBRIDGE_CLASSES = [
    "CAT_BATTERY",
    "CAT_CABLES",
    "CAT_CRT",
    "CAT_LCD_LED",
    "CAT_OTHER",
    "CAT_PCB"
]

ECOBRIDGE_DISPLAY_NAMES = {
    "CAT_BATTERY": "Battery",
    "CAT_CABLES": "Cables",
    "CAT_CRT": "CRT",
    "CAT_LCD_LED": "LCD / LED",
    "CAT_OTHER": "Other E-waste",
    "CAT_PCB": "PCB"
}

# -------------------------------------------------------------------------
# INFERENCE FUNCTION
# -------------------------------------------------------------------------

def ecobridge_predict(image_path):
    """
    ECOBRIDGE AI classification engine.

    Input:
        image_path : path to an image

    Output:
        dictionary containing:
        - predicted category
        - confidence
        - verification status
        - top-3 predictions
    """

    # -------------------------------------------------------------
    # Load image
    # -------------------------------------------------------------

    image = tf.keras.utils.load_img(
        image_path,
        target_size=(224, 224)
    )

    image_array = tf.keras.utils.img_to_array(
        image
    )

    image_batch = np.expand_dims(
        image_array,
        axis=0
    )

    # -------------------------------------------------------------
    # MobileNetV2 preprocessing
    # -------------------------------------------------------------

    image_batch = (
        tf.keras.applications.mobilenet_v2.preprocess_input(
            image_batch
        )
    )

    # -------------------------------------------------------------
    # Prediction
    # -------------------------------------------------------------

    probabilities = final_model.predict(
        image_batch,
        verbose=0
    )[0]

    predicted_index = int(
        np.argmax(probabilities)
    )

    predicted_class = ECOBRIDGE_CLASSES[
        predicted_index
    ]

    confidence = float(
        probabilities[predicted_index]
    )

    # -------------------------------------------------------------
    # Confidence-aware verification
    # -------------------------------------------------------------

    if confidence >= ECOBRIDGE_CONFIDENCE_THRESHOLD:

        verification_status = (
            "HIGH-CONFIDENCE AI SUGGESTION"
        )

        requires_manual_verification = True

        message = (
            "AI confidence is above the prototype threshold. "
            "Collector/recycler verification is still required "
            "before completing the transaction."
        )

    else:

        verification_status = (
            "MANUAL VERIFICATION REQUIRED"
        )

        requires_manual_verification = True

        message = (
            "AI confidence is below the prototype threshold. "
            "Material must be manually verified."
        )

    # -------------------------------------------------------------
    # Top-3 predictions
    # -------------------------------------------------------------

    top_indices = np.argsort(
        probabilities
    )[::-1][:3]

    top_predictions = []

    for index in top_indices:

        class_name = ECOBRIDGE_CLASSES[
            index
        ]

        top_predictions.append({
            "category": class_name,
            "display_name":
                ECOBRIDGE_DISPLAY_NAMES[
                    class_name
                ],
            "confidence":
                float(probabilities[index])
        })

    # -------------------------------------------------------------
    # Result
    # -------------------------------------------------------------

    result = {
        "predicted_class":
            predicted_class,

        "predicted_category":
            ECOBRIDGE_DISPLAY_NAMES[
                predicted_class
            ],

        "confidence":
            confidence,

        "confidence_percent":
            confidence * 100,

        "verification_status":
            verification_status,

        "requires_manual_verification":
            requires_manual_verification,

        "message":
            message,

        "top_3_predictions":
            top_predictions
    }

    return result


# =============================================================================
# TEST THE FINAL ENGINE USING THE PREVIOUS IMAGE
# =============================================================================

test_image = "OIP (1).jpg"

result = ecobridge_predict(
    test_image
)

print("\n" + "=" * 80)
print("FINAL ECOBRIDGE AI RESULT")
print("=" * 80)

print(
    f"\nMaterial: "
    f"{result['predicted_category']}"
)

print(
    f"Confidence: "
    f"{result['confidence_percent']:.2f}%"
)

print(
    f"Status: "
    f"{result['verification_status']}"
)

print(
    f"\nManual verification required: "
    f"{result['requires_manual_verification']}"
)

print(
    f"\nMessage:\n"
    f"{result['message']}"
)

print("\n" + "-" * 80)
print("TOP-3 PREDICTIONS")
print("-" * 80)

for rank, prediction in enumerate(
    result["top_3_predictions"],
    start=1
):

    print(
        f"{rank}. "
        f"{prediction['display_name']:18} "
        f"{prediction['confidence'] * 100:.2f}%"
    )

print("\n" + "=" * 80)
print("STEP 113 COMPLETE")
print("=" * 80)

print(
    "\n✓ ECOBRIDGE inference engine is ready."
)
```

```python
# =============================================================================
# STEP 116 — ECOBRIDGE FINAL LIVE DEMO
# =============================================================================

from google.colab import files
from IPython.display import display
from PIL import Image
import numpy as np
import tensorflow as tf

print("=" * 80)
print("♻️ ECOBRIDGE — AI E-WASTE CLASSIFICATION DEMO")
print("=" * 80)

# -------------------------------------------------------------------------
# CONFIGURATION
# -------------------------------------------------------------------------

CONFIDENCE_THRESHOLD = 0.90

CLASS_NAMES = [
    "Battery",
    "Cables",
    "CRT",
    "LCD / LED",
    "Other E-waste",
    "PCB"
]

# -------------------------------------------------------------------------
# UPLOAD IMAGE
# -------------------------------------------------------------------------

print("\n📸 Upload an e-waste image...")

uploaded = files.upload()

if not uploaded:
    raise ValueError("No image uploaded.")

image_path = list(uploaded.keys())[0]

print(f"\n✓ Image uploaded: {image_path}")

# -------------------------------------------------------------------------
# DISPLAY IMAGE
# -------------------------------------------------------------------------

image = Image.open(image_path).convert("RGB")

display(image.resize((350, 350)))

# -------------------------------------------------------------------------
# PREPROCESS IMAGE
# -------------------------------------------------------------------------

image_resized = image.resize(
    (224, 224)
)

image_array = np.array(
    image_resized
).astype("float32")

image_batch = np.expand_dims(
    image_array,
    axis=0
)

image_batch = (
    tf.keras.applications.mobilenet_v2.preprocess_input(
        image_batch
    )
)

# -------------------------------------------------------------------------
# AI PREDICTION
# -------------------------------------------------------------------------

probabilities = final_model.predict(
    image_batch,
    verbose=0
)[0]

predicted_index = int(
    np.argmax(probabilities)
)

predicted_class = CLASS_NAMES[
    predicted_index
]

confidence = float(
    probabilities[predicted_index]
)

# -------------------------------------------------------------------------
# VERIFICATION STATUS
# -------------------------------------------------------------------------

if confidence >= CONFIDENCE_THRESHOLD:

    status = "HIGH-CONFIDENCE AI SUGGESTION"

    instruction = (
        "Proceed to collector/recycler verification."
    )

else:

    status = "MANUAL VERIFICATION REQUIRED"

    instruction = (
        "AI confidence is below 90%. "
        "Collector/recycler must verify the material."
    )

# -------------------------------------------------------------------------
# TOP-3 PREDICTIONS
# -------------------------------------------------------------------------

top_indices = np.argsort(
    probabilities
)[::-1][:3]

# -------------------------------------------------------------------------
# DISPLAY RESULT
# -------------------------------------------------------------------------

print("\n" + "=" * 80)
print("🤖 ECOBRIDGE AI RESULT")
print("=" * 80)

print(
    f"\nMaterial Classification : {predicted_class}"
)

print(
    f"AI Confidence           : "
    f"{confidence * 100:.2f}%"
)

print(
    f"Verification Status     : {status}"
)

print(
    f"\nAction                  : {instruction}"
)

# -------------------------------------------------------------------------
# TOP-3
# -------------------------------------------------------------------------

print("\n" + "-" * 80)
print("TOP-3 AI SUGGESTIONS")
print("-" * 80)

for rank, index in enumerate(
    top_indices,
    start=1
):

    print(
        f"{rank}. "
        f"{CLASS_NAMES[index]:18} "
        f"{probabilities[index] * 100:6.2f}%"
    )

# -------------------------------------------------------------------------
# IMPORTANT BUSINESS RULE
# -------------------------------------------------------------------------

print("\n" + "=" * 80)
print("♻️ ECOBRIDGE TRANSACTION RULE")
print("=" * 80)

print("""
AI provides a classification suggestion.

Final transaction value is NOT independently determined by AI.

Final value =
Verified Weight × Agreed Recycler Price

Collector/recycler verification remains part of the transaction workflow.
""")

print("=" * 80)
print("✓ ECOBRIDGE LIVE DEMO COMPLETE")
print("=" * 80)
```
